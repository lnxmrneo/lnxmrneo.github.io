## Тэмдэгт мөрүүдийг харьцуулах

Харьцангуй операторууд тэмдэгт мөрүүд дээр ажилладаг. Хоёр мөр байгаа эсэхийг харахын тулд
тэнцүү бол бид `==` операторыг ашиглаж болно.

```python
word = 'banana'

if word == 'banana':
    print('All right, banana.')
```

Бусад харилцааны үйлдлүүд нь үгсийг цагаан толгойн үсгийн дарааллаар оруулахад тустай
захиалга:

```python
def compare_word(word):
    if word < 'banana':
        print(word, 'comes before banana.')
    elif word > 'banana':
        print(word, 'comes after banana.')
    else:
        print('All right, banana.')
```

```python
compare_word('apple')
```

Python том жижиг үсгийг ижил аргаар зохицуулдаггүй
хүмүүс хийдэг. Бүх том үсэг нь жижиг үсгийн өмнө ирдэг
үсэг, тиймээс:

```python
compare_word('Pineapple')
```

Энэ асуудлыг шийдэхийн тулд бид харьцуулалт хийхээсээ өмнө мөрүүдийг бүх жижиг үсгээр зэрэг стандарт формат руу хөрвүүлж болно.
Хэрэв та хан боргоцойгоор зэвсэглэсэн хүнээс өөрийгөө хамгаалах шаардлагатай бол үүнийг санаарай.

## Файлд бичих

Стринг операторууд болон аргууд нь текст файлуудыг унших, бичихэд тустай.
Жишээ болгон бид Гутенберг төслийн (<https://www.gutenberg.org/ebooks/345>) -аас авах боломжтой Брам Стокерын *Дракула* романы тексттэй ажиллах болно.

```python
import os

if not os.path.exists('pg345.txt'):
    !wget https://www.gutenberg.org/cache/epub/345/pg345.txt
```

Би номыг `pg345.txt` нэртэй энгийн текст файлаар татаж авсан бөгөөд бид үүнийг дараах байдлаар унших боломжтой.

```python
reader = open('pg345.txt')
```

Энэхүү файл нь номын эх бичвэрээс гадна эхэнд нь номын тухай мэдээлэл, төгсгөлд нь лицензийн тухай мэдээлэл бүхий хэсгийг агуулна.
Текстийг боловсруулахын өмнө бид `***`-ээр эхэлсэн эхэн ба төгсгөлд байгаа тусгай мөрүүдийг олох замаар энэ нэмэлт материалыг устгаж болно.

Дараах функц нь мөр авч, тусгай мөрүүдийн нэг мөн эсэхийг шалгана.
Энэ нь `startswith` аргыг ашигладаг бөгөөд мөр нь өгөгдсөн тэмдэгтүүдийн дарааллаар эхэлж байгаа эсэхийг шалгадаг.

```python
def is_special_line(line):
    return line.startswith('*** ')
```

Бид энэ функцийг ашиглан файлын мөрүүдийг давтаж, зөвхөн тусгай мөрүүдийг хэвлэх боломжтой.

```python
for line in reader:
    if is_special_line(line):
        print(line.strip())
```

Одоо зөвхөн номын текстийг агуулсан `pg345_cleaned.txt` нэртэй шинэ файл үүсгэцгээе.
Номыг дахин уншихын тулд бид дахин уншихаар нээх хэрэгтэй.
Мөн шинэ файл бичихийн тулд бид үүнийг бичихээр нээж болно.

```python
reader = open('pg345.txt')
writer = open('pg345_cleaned.txt', 'w')
```

`open` нь "горим"-ыг тодорхойлсон нэмэлт параметрүүдийг авдаг -- энэ жишээнд `w` нь файлыг бичихээр нээж байгааг харуулж байна.
Хэрэв файл байхгүй бол түүнийг үүсгэх болно; Хэрэв энэ нь аль хэдийн байгаа бол агуулгыг солих болно.

Эхний алхам болгон бид эхний тусгай мөрийг олох хүртлээ файлыг давтах болно.

```python
for line in reader:
    if is_special_line(line):
        break
```

`break` заавар нь давталтноос "тасрах" бөгөөд өөрөөр хэлбэл файлын төгсгөлд хүрэхээс өмнө давталтыг шууд дуусгахад хүргэдэг.

Давталт гарах үед `line` нөхцөлийг үнэн болгосон тусгай мөрийг агуулна.

```python
line
```

`reader` нь файлын хаана байгааг хянаж байдаг тул бид хоёр дахь давталт ашиглан орхисон газраасаа үргэлжлүүлж болно.

Дараах давталт нь файлын үлдсэн хэсгийг нэг мөр болгон уншдаг.
Текстийн төгсгөлийг харуулсан тусгай мөрийг олох үед энэ нь давталтноос гарна.
Үгүй бол гаралтын файлд мөрийг бичнэ.

```python
for line in reader:
    if is_special_line(line):
        break
    writer.write(line)
```

Энэ давталт гарахад `line` хоёр дахь тусгай мөрийг агуулна.

```python
line
```

Энэ үед `reader` болон `writer` нээлттэй хэвээр байгаа бөгөөд энэ нь бид `reader`-аас мөрүүдийг унших эсвэл `writer` хүртэл бичих боломжтой гэсэн үг юм.
Дууссан гэдгээ харуулахын тулд бид `close` аргыг дуудаж хоёр файлыг хааж болно.

```python
reader.close()
writer.close()
```

Энэ процесс амжилттай болсон эсэхийг шалгахын тулд бид саяхан үүсгэсэн шинэ файлынхаа эхний хэдэн мөрийг уншиж болно.

```python
for line in open('pg345_cleaned.txt'):
    line = line.strip()
    if len(line) > 0:
        print(line)
    if line.endswith('Stoker'):
        break
```

## Хайх ба орлуулах

1901 онд Исланд хэлээр орчуулагдсан *Дракула* зохиолд нэг дүрийн нэрийг "Жонатан" байсныг "Томас" болгон өөрчилсөн байна.
Англи хувилбарт энэ өөрчлөлтийг хийхийн тулд бид номыг эргүүлж, `replace` аргыг ашиглан нэг нэрийг өөр нэрээр сольж, үр дүнг шинэ файлд бичиж болно.

Бид файлын цэвэрлэсэн хувилбар дахь мөрүүдийг тоолж эхэлнэ.

```python
total = 0
for line in open('pg345_cleaned.txt'):
    total += 1
    
total
```

Нэг мөрөнд "Жонатан" байгаа эсэхийг харахын тулд бид `in` операторыг ашиглаж болох бөгөөд энэ тэмдэгтүүдийн дараалал мөрний аль ч хэсэгт гарч байгаа эсэхийг шалгадаг.

```python
total = 0
for line in open('pg345_cleaned.txt'):
    if 'Jonathan' in line:
        total += 1

total
```

Нэрийг агуулсан 199 мөр байдаг боловч энэ нь нэг мөрөнд нэгээс олон удаа гарч ирж болох тул энэ нь нийт гарч ирэх тоо биш юм.
Нийт дүнг авахын тулд бид `count` аргыг ашиглаж болох бөгөөд энэ нь мөрөнд дараалал гарч ирэх тоог буцаана.

```python
total = 0
for line in open('pg345_cleaned.txt'):
    total += line.count('Jonathan')

total
```

Одоо бид `Jonathan`-г `Thomas`-ээр дараах байдлаар сольж болно:

```python
writer = open('pg345_replaced.txt', 'w')

for line in open('pg345_cleaned.txt'):
    line = line.replace('Jonathan', 'Thomas')
    writer.write(line)
```

Үр дүн нь Жонатан Харкерыг Томас гэж нэрлэдэг *Дракула*-н хувилбарыг агуулсан `pg345_replaced.txt` нэртэй шинэ файл юм.

```python
total = 0
for line in open('pg345_replaced.txt'):
    total += line.count('Thomas')

total
```

`re` нэртэй модуль нь регуляр илэрхийлэлтэй холбоотой функцүүдийг хангадаг.
Бид үүнийг ингэж импортлох боломжтой бөгөөд `search` функцийг ашиглан хээ нь бичвэрт байгаа эсэхийг шалгах боломжтой.

```python
import re

result = re.search(pattern, text)
result
```

Хэрэв загвар текст дээр гарч ирвэл `search` хайлтын үр дүнг агуулсан `Match` объектыг буцаана.
Бусад мэдээллийн дотор хайсан текстийг агуулсан `string` нэртэй хувьсагчтай.

```python
result.string
```

Энэ нь мөн `group` хэмээх функцээр хангадаг бөгөөд энэ нь загварт тохирсон текстийн хэсгийг буцаадаг.

```python
result.group()
```

Мөн энэ нь `span` хэмээх функцээр хангадаг бөгөөд энэ нь хэв маяг эхэлж, төгссөн текст дэх индексийг буцаадаг.

```python
result.span()
```

Хэрэв загвар нь текстэд харагдахгүй бол `search`-аас буцаах утга нь `None` байна.

```python
result = re.search('Count', text)
print(result)
```

Тиймээс бид хайлтын үр дүн `None` эсэхийг шалгаснаар хайлт амжилттай болсон эсэхийг шалгаж болно.

```python
result == None
```

Энэ бүгдийг нэгтгэж үзвэл, өгөгдсөн загварт тохирохыг олох хүртэл номын мөрүүдийг давж, `Match` объектыг буцаадаг функц байна.

```python
def find_first(pattern):
    for line in open('pg345_cleaned.txt'):
        result = re.search(pattern, line)
        if result != None:
            return result
```

Бид үүнийг дүрийн эхний дурдлагыг олохын тулд ашиглаж болно.

```python
result = find_first('Harker')
result.string
```

Энэ жишээний хувьд бид регуляр илэрхийлэл ашиглах шаардлагагүй байсан -- бид `in` оператортой ижил зүйлийг илүү хялбар хийж болох байсан.
Гэхдээ регуляр илэрхийлэл нь `in` операторын хийж чадахгүй зүйлийг хийж чадна.

Жишээлбэл, хээ нь `|` гэсэн босоо зураасны тэмдэгтийг агуулж байвал зүүн талын дараалал эсвэл баруун талын дараалалтай таарч болно.
Бид энэ номноос Мина Мюррейгийн тухай анхны дурдлагыг олохыг хүсч байгаа ч түүний нэр эсвэл овог нэрээр нь дурдагдсан эсэх нь тодорхойгүй байна гэж бодъё.
Бид аль ч нэртэй тохирох дараах загварыг ашиглаж болно.

```python
pattern = r'Mina|Murray'
result = find_first(pattern)
result.string
```

Бид үүнтэй төстэй загвар ашиглан тэмдэгтийг аль ч нэрээр нь хэдэн удаа дурддагийг харах боломжтой.
Номыг давтаж, өгөгдсөн загварт тохирох мөрийн тоог тоолох функц энд байна.

```python
def count_matches(pattern):
    count = 0
    for line in open('pg345_cleaned.txt'):
        result = re.search(pattern, line)
        if result != None:
            count += 1
    return count
```

Одоо Мина хэдэн удаа дурдагдсаныг харцгаая.
