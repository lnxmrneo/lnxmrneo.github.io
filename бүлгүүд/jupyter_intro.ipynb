## Notebook гэж юу вэ?

Jupyter дэвтэр нь нүднүүдээс тогтдог бөгөөд нүд бүр нь текст эсвэл код агуулсан байдаг.
Энэ нүдэнд текст агуулагдаж байна. 

Дараах нүдэнд код байна.

```python
print('Hello')
```

Сонгохын тулд өмнөх нүдэн дээр дарна уу.
Та зүүн талд тойрог дотор гурвалжин дүрс бүхий товчлуурыг харах ёстой бөгөөд энэ нь "Тоглох" дүрс юм.
Хэрэв та энэ товчлуурыг дарвал Jupyter нүдэн дэх кодыг ажиллуулж, үр дүнг харуулна.

Тэмдэглэлийн дэвтэрт кодыг анх удаа ажиллуулж эхлэхэд хэдхэн секунд зарцуулагдана.
Хэрэв энэ нь таны бичээгүй дэвтэр бол танд анхааруулах мессеж ирж магадгүй юм.
Хэрэв та итгэлтэй эх сурвалжаасаа дэвтэр ажиллуулж байгаа бол "Ямар ч байсан ажиллуулах" дээр дарж болно.

Та "Тоглох" товчийг дарахын оронд `Shift` товчийг удаан дарж, `Enter` дээр дарснаар кодыг нүдэнд ажиллуулж болно.

Хэрэв та энэ дэвтэрийг Colab дээр ажиллуулж байгаа бол зүүн дээд хэсэгт "+ Code" болон "+ Text" гэсэн товчлууруудыг харах ёстой.  Эхнийх нь кодын нүдийг, хоёр дахь нь текстийн нүдийг нэмнэ.
Хэрэв та тэдгээрийг туршиж үзэхийг хүсвэл дээр нь дарж энэ нүдийг сонгоод "+ Текст" товчийг дарна уу.
Үүний доор шинэ нүд гарч ирэх ёстой.

Нүдэнд зарим текст нэмнэ үү.
Та товчлууруудыг ашиглан форматлах эсвэл [Markdown](https://www.markdownguide.org/basic-syntax/) ашиглан текстийг тэмдэглэж болно.
Дуусаад `Shift` товчийг удаан дарж `Enter` товчийг дарснаар таны оруулсан текстийг форматлаад дараагийн нүд рүү шилжинэ.

Ямар ч үед Jupyter хоёр горимын аль нэгэнд байна:

* **командын горимд** та бүхэл нүдийг нэмэх, хасах зэрэг нүдэнд нөлөөлөх үйлдлүүдийг хийж болно.

* **засварлах горимд** та нүдний агуулгыг засах боломжтой.

Текст нүднүүдийн тусламжтайгаар та ямар горимд байгаа нь тодорхой болно.
Засварлах горимд нүд нь босоо байдлаар хуваагддаг бөгөөд зүүн талд нь засварлаж буй текст, баруун талд нь форматлагдсан текст байна.
Мөн та дээд талд текст засварлах хэрэгслүүдийг харах болно.
Тушаалын горимд та зөвхөн форматлагдсан текстийг харна.

Кодын нүднүүдийн хувьд ялгаа нь илүү нарийн байх боловч хэрэв нүдэнд курсор байгаа бол та засварлах горимд байна.

Засварлах горимоос тушаалын горимд шилжихийн тулд `ESC` товчийг дарна уу.
Командын горимоос засварлах горим руу шилжихийн тулд `Enter` товчийг дарна уу.

Тэмдэглэлийн дэвтэр дээр ажиллаж дууссаны дараа та цонхыг хааж болно, гэхдээ таны хийсэн өөрчлөлтүүд алга болно.
Хэрэв та хадгалахыг хүсэж байгаа ямар нэгэн өөрчлөлт хийвэл зүүн дээд хэсэгт байрлах Файл цэсийг нээнэ үү.
Та дэвтэрээ хадгалах хэд хэдэн аргыг харах болно.

* Хэрэв танд Google бүртгэл байгаа бол дэвтэрээ Драйвдаа хадгалах боломжтой.

* Хэрэв танд GitHub бүртгэл байгаа бол GitHub дээр хадгалах боломжтой.

* Эсвэл та зөөврийн компьютер дээрээ зөөврийн компьютерээ хадгалахыг хүсвэл "Татаж авах" гэснийг сонгоод "Татаж авах .ipynb"-г сонгоод ".ipynb" дагавар нь зөвхөн Python код агуулсан файлыг заадаг ".py"-ээс ялгаатай нь ".ipynb" дагавар нь тэмдэглэлийн дэвтрийн файл гэдгийг илтгэнэ.

## *Think Python* номын код

Тэмдэглэлийн дэвтэр бүрийн эхэнд та дараах кодтой нүдийг харах болно:

```python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://raw.githubusercontent.com/AllenDowney/ThinkPython/v3/thinkpython.py')

import thinkpython
```

Та энэ код хэрхэн ажилладагийг мэдэх шаардлагагүй, гэхдээ та номын төгсгөлд хүрэхэд ихэнх нь утга учиртай болно.
Таны таамаглаж байгаачлан энэ нь энэ номонд тусгайлан өгсөн Python кодыг агуулсан `thinkpython.py` файлыг татаж авдаг.
Сүүлийн мөр нь энэ кодыг "импортлох" бөгөөд энэ нь бид кодыг дэвтэрт ашиглах боломжтой гэсэн үг юм.

Бусад бүлгүүдэд та номын диаграммыг үүсгэхэд ашигладаг `diagram.py` кодыг татаж авах кодыг болон яст мэлхийн графикийг бүтээхэд хэд хэдэн бүлэгт ашигладаг `jupyturtle.py` кодыг харах болно.
