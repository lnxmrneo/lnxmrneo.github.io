(section_encapsulation)=
## Багцлах ба ерөнхийлөх

Өмнөх хэсгийн дөрвөлжин зургийн кодыг аваад `square` хэмээх функцэд оруулъя.

````python
def square():
    for i in range(4):
        forward(50)
        left(90)
````

Одоо бид функцийг ингэж дуудаж болно.

````python
make_turtle()
square()
````

Функцэд кодын хэсгийг ороохыг **encapsulation** гэж нэрлэдэг.
Капсулжуулалтын нэг давуу тал нь кодонд нэр хавсаргах бөгөөд энэ нь нэг төрлийн баримт бичиг болдог. Өөр нэг давуу тал нь хэрэв та кодыг дахин ашиглавал үндсэн хэсгийг хуулж буулгахаас илүү функцийг хоёр удаа дуудах нь илүү товч юм!

Одоогийн хувилбарт квадратын хэмжээ үргэлж `50` байна.
Хэрэв бид өөр өөр хэмжээтэй квадратуудыг зурахыг хүсвэл талуудын уртыг параметр болгон авч болно.

````python
def square(length):
    for i in range(4):
        forward(length)
        left(90)
````

Одоо бид янз бүрийн хэмжээтэй дөрвөлжин зурж болно.

````python
make_turtle()
square(30)
square(60)
````

Функцэд параметр нэмэхийг **ерөнхийлэх** гэж нэрлэдэг, учир нь энэ нь функцийг илүү ерөнхий болгодог: өмнөх хувилбарт квадрат үргэлж ижил хэмжээтэй байна; Энэ хувилбарт энэ нь ямар ч хэмжээтэй байж болно.

Хэрэв бид өөр параметр нэмбэл бид үүнийг илүү ерөнхий болгож чадна.
Дараах функц нь өгөгдсөн талуудтай ердийн олон өнцөгтүүдийг зурдаг.

````python
def polygon(n, length):
    angle = 360 / n
    for i in range(n):
        forward(length)
        left(angle)
````

`n` талуудтай ердийн олон өнцөгтийн хажуу талуудын хоорондох өнцөг нь `360 / n` градус байна. 

Дараах жишээнд `30` талын урттай `7` талт олон өнцөгт зурсан.

````python
make_turtle()
polygon(7, 30)
````

Функц нь хэд хэдэн тоон аргументтай бол тэдгээр нь юу болох, ямар дарааллаар байх ёстойг мартах нь амархан байдаг. 
Аргументуудын жагсаалтад параметрүүдийн нэрийг оруулах нь зүйтэй.

````python
make_turtle()
polygon(n=7, length=30)
````

Эдгээр нь параметрийн нэрийг агуулсан байдаг тул заримдаа "нэрлэсэн аргументууд" гэж нэрлэдэг.
Гэхдээ Python-д тэдгээрийг ихэвчлэн **түлхүүр үгийн аргументууд** гэж нэрлэдэг (`for`, `def` зэрэг Python түлхүүр үгстэй андуурч болохгүй).

`=` онооны операторын энэхүү хэрэглээ нь аргументууд болон параметрүүд хэрхэн ажилладаг тухай сануулга юм -- та функцийг дуудах үед аргументуудыг параметрүүдэд хуваарилдаг.

## Тойргийг ойролцоогоор дүрслэх

Одоо бид тойрог зурахыг хүсч байна гэж бодъё.
Ойролцоогоор олон талтай олон өнцөгт зурснаар бид үүнийг хийж чадна, тиймээс тал бүр нь хангалттай жижиг тул харахад хэцүү байдаг.
Тойрог ойролцоох `30` талт олон өнцөгт зурахын тулд `polygon` ашигладаг функц энд байна.

````python
import math

def circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    polygon(n, length)
````

`circle` нь тойргийн радиусыг параметр болгон авдаг.
Энэ нь өгөгдсөн радиустай тойргийн тойрог болох `circumference`-г тооцоолдог.
`n` нь талуудын тоо тул `circumference / n` нь тал бүрийн урт юм.

Энэ функцийг ажиллуулахад удаан хугацаа шаардагдаж магадгүй.
Бид `make_turtle`-г `delay` хэмээх түлхүүр үгийн аргументын тусламжтайгаар дуудаж, яст мэлхий алхам бүрийн дараа хүлээх цагийг секундээр тогтоосноор үүнийг хурдасгаж чадна.
Өгөгдмөл утга нь `0.2` секунд бөгөөд хэрэв бид үүнийг `0.02` болгож тохируулбал 10 дахин хурдан ажиллана.

````python
make_turtle(delay=0.02)
circle(30)
````

Энэхүү шийдлийн хязгаарлалт нь `n` нь тогтмол бөгөөд энэ нь гэсэн үг юм
Энэ нь маш том тойргийн хувьд талууд нь хэтэрхий урт, жижиг хэмжээтэй байдаг
тойрог, бид маш богино талыг зурахад цаг алддаг.
Нэг сонголт бол `n`-г параметр болгон авч функцийг ерөнхийд нь харуулах явдал юм.
Гэхдээ одоохондоо үүнийг энгийн болгоё.

## Кодын бүтцийг сайжруулах

Одоо `circle` функцийн илүү ерөнхий хувилбар болох `arc` функцийг бичье. Энэ нь хоёр дахь параметр болох `angle`-ийг хүлээн авч, өгсөн өнцөгт харгалзах тойргийн нум зурна.
Жишээлбэл, `angle` нь `360` градус байвал бүтэн тойрог, `angle` нь `180` градус байвал хагас тойрог зурна.

Олон талт олон өнцөгт тойргийг сайн ойролцоолдог тул `circle`-ийг бичихдээ `polygon`-ийг дахин ашиглаж болсон. Харин `arc`-ийг бичихэд `polygon`-ийг шууд ашиглаж болохгүй.

Тиймээс `polygon`-ийн илүү ерөнхий хувилбар болох `polyline`-ийг үүсгэнэ.

````python
def polyline(n, length, angle):
    for i in range(n):
        forward(length)
        left(angle)
````

`polyline` нь зурах шугамын тоо, `n`, сегментийн урт, `length`, тэдгээрийн хоорондох өнцөг `angle` зэргийг параметр болгон авдаг.

Одоо бид `polyline`-г ашиглахын тулд `polygon`-г дахин бичиж болно.

````python
def polygon(n, length):
    angle = 360.0 / n
    polyline(n, length, angle)
````

Мөн бид `polyline` ашиглан `arc` бичих боломжтой.

````python
def arc(radius, angle):
    arc_length = 2 * math.pi * radius * angle / 360
    n = 30
    length = arc_length / n
    step_angle = angle / n
    polyline(n, length, step_angle)
````

`arc` нь `circle`-тай төстэй боловч тойргийн тойргийн хэсэг болох `arc_length`-г тооцоолдог.

Эцэст нь бид `arc`-г ашиглахын тулд `circle`-г дахин бичиж болно.

````python
def circle(radius):
    arc(radius,  360)
````

Эдгээр функцүүд хүлээгдэж буйгаар ажиллаж байгаа эсэхийг шалгахын тулд бид тэдгээрийг эмгэн хумс шиг зурахад ашиглана.
`delay=0`-ийн тусламжтайгаар яст мэлхий аль болох хурдан гүйдэг.

````python
make_turtle(delay=0)
polygon(n=20, length=9)
arc(radius=70, angle=70)
circle(radius=10)
````

Энэ жишээн дээр бид ажлын кодоор эхэлж, өөр өөр функцээр дахин зохион байгуулсан.
Кодын зан төлөвийг өөрчлөхгүйгээр сайжруулдаг ийм өөрчлөлтийг **рефакторинг** гэж нэрлэдэг.

Хэрэв бид урьдчилан төлөвлөж байсан бол `polyline`-г эхлээд бичиж, дахин засварлахаас зайлсхийх байсан ч та төслийн эхэнд бүх функцийг төлөвлөхөд хангалттай мэдлэггүй байдаг.
Код бичиж эхэлмэгц та асуудлыг илүү сайн ойлгодог.
Заримдаа рефакторинг хийх нь таныг ямар нэг зүйл сурсаны шинж юм.

## Програм боловсруулах төлөвлөгөө

**хөгжлийн төлөвлөгөө** нь програм бичих үйл явц юм.
Энэ бүлэгт бидний ашигласан үйл явц бол "бүртгэл ба ерөнхий ойлголт" юм.
Энэ үйл явцын алхамууд нь:

2. Хөтөлбөрийг ажиллуулсны дараа түүний уялдаа холбоотой хэсгийг тодорхойл,
    хэсгийг функцэд багтааж, түүнд нэр өгнө үү.
    Дахин бичихээс (болон дахин алдаа олж засварлах) зайлсхийхийн тулд ажлын кодыг хуулж буулгана уу.

3. Тохирох параметрүүдийг нэмж функцийг ерөнхийлнө.

4. 1-3-р алхмуудыг хэд хэдэн ажлын функцтэй болтол давт.

5. Хөтөлбөрийг дахин засварлах замаар сайжруулах боломжийг эрэлхийл. Учир нь
    Жишээлбэл, хэрэв танд ижил төстэй код хэд хэдэн газар байгаа бол үүнийг анхаарч үзээрэй
    үүнийг зохих ерөнхий функц болгон хувиргах.

Энэ процесс нь зарим сул талуудтай -- бид дараа нь өөр хувилбаруудыг харах болно -- гэхдээ хэрэв та програмыг функцэд хэрхэн хувахаа урьдчилан мэдэхгүй бол энэ нь ашигтай байж болох юм.
Энэ арга нь таныг цааш явахдаа дизайн хийх боломжийг олгодог.

Функцийн дизайн нь хоёр хэсгээс бүрдэнэ.

* **интерфейс** гэдэг нь функцийг хэрхэн ашиглах, түүний нэр, авах параметр, функц юу хийх ёстой зэргийг багтаадаг.

* **хэрэгжүүлэх** гэдэг нь тухайн функц хийх ёстой зүйлээ хэрхэн хийж байгааг хэлнэ.

Жишээлбэл, `polygon` ашигладаг `circle`-ийн анхны хувилбарыг энд оруулав.

````python
def circle(radius):
    circumference = 2 * math.pi * radius
    n = 30
    length = circumference / n
    polygon(n, length)
````

Энд `arc` ашигладаг дахин засварлагдсан хувилбар байна.

````python
def circle(radius):
    arc(radius,  360)
````

Эдгээр хоёр функц нь ижил интерфэйстэй -- тэд ижил параметрүүдийг авч, ижил зүйлийг хийдэг - гэхдээ тэдгээр нь өөр өөр хэрэгжилттэй байдаг.

## Функцийн баримтжуулах мөр

**docstring** нь интерфэйсийг тайлбарлах функцийн эхэнд байрлах мөр юм ("doc" нь "баримт бичиг" гэсэн үгийн товчлол).
Энд жишээ байна:

````python
def polyline(n, length, angle):
    """Draws line segments with the given length and angle between them.
    
    n: integer number of line segments
    length: length of the line segments
    angle: angle between segments (in degrees)
    """    
    for i in range(n):
        forward(length)
        left(angle)
````

Уламжлал ёсоор, docstrings нь гурвалсан хашилттай мөрүүд бөгөөд **олон мөр** гэж нэрлэгддэг, учир нь гурвалсан ишлэл нь мөрийг нэгээс олон мөрийг хамрах боломжийг олгодог.

Баримт бичгийн мөр нь:

* Функц хэрхэн ажилладаг талаар дэлгэрэнгүй мэдээлэл авалгүйгээр юу хийдэгийг товч тайлбарлах,

* Параметр бүр нь функцийн үйл ажиллагаанд ямар нөлөө үзүүлж байгааг тайлбарла

* Хэрэв тодорхой биш бол параметр бүр ямар төрлийн байх ёстойг зааж өгнө үү.

Энэ төрлийн баримт бичгийг бичих нь интерфейсийн дизайны чухал хэсэг юм.
Сайн боловсруулсан интерфэйс нь тайлбарлахад хялбар байх ёстой; Хэрэв та аль нэг функцээ тайлбарлахад хэцүү байвал интерфэйсийг сайжруулж магадгүй юм.

## Алдаа олж засварлах

Интерфейс нь функц болон дуудагч хоёрын хоорондох гэрээтэй адил юм. The
дуудлага хийгч тодорхой аргументуудыг өгөхийг зөвшөөрч, функц нь зөвшөөрч байна
тодорхой ажил хийх.

Жишээлбэл, `polyline` гурван аргумент шаарддаг: `n` бүхэл тоо байх ёстой; `length` эерэг тоо байх ёстой; болон `angle` нь градусаар ойлгогдох тоо байх ёстой.

Эдгээр шаардлагууд нь функц ажиллаж эхлэхээс өмнө үнэн байх ёстой тул **урьдчилсан нөхцөл**гэж нэрлэдэг. Үүний эсрэгээр, функцийн төгсгөлд байгаа нөхцөлүүд нь**дараах нөхцөл** юм.
Дараах нөхцөлүүд нь функцийн төлөвлөсөн нөлөө (шугамын хэсгүүдийг зурах гэх мэт) болон аливаа гаж нөлөө (яст мэлхийг хөдөлгөх эсвэл өөр өөрчлөлт хийх гэх мэт) орно.

Урьдчилсан нөхцөл нь дуудлага хийгчийн үүрэг юм. Хэрэв дуудагч урьдчилсан нөхцөлийг зөрчиж, функц зөв ажиллахгүй бол алдаа нь функцэд биш, харин дуудлага хийгчид байна.

Урьдчилсан нөхцөл хангагдсан ба дараах нөхцөл хангагдаагүй бол алдаа нь функцэд байна. Хэрэв таны урьдчилсан болон дараах нөхцөлүүд тодорхой байвал алдаа олж засварлахэд тусалж чадна.

## Нэр томьёоны тайлбар

**интерфейсийн загварчлал:**
Функцийн хүлээн авах параметрүүд зэрэг интерфейсийг төлөвлөх үйл явц.

**зургийн талбар:**
Шугам, тойрог, тэгш өнцөгт болон бусад дүрсийг харуулах цонх.

**багцлах:**
Заавруудын дарааллыг функцийн тодорхойлолт болгон хувиргах үйл явц.

**ерөнхийлөх:**
Шаардлагагүйгээр хэт тодорхойлсон зүйлийг, тухайлбал тоог, хувьсагч эсвэл параметр зэрэг ерөнхий зүйлээр солих үйл явц.

**нэрлэсэн аргумент:**
Параметрийн нэрийг хамт бичсэн аргумент.

**кодын бүтцийг сайжруулах:**
Ажиллаж буй програмын функцийн интерфейс болон кодын бусад чанарыг сайжруулах зорилгоор бүтцийг өөрчлөх үйл явц.

**боловсруулах төлөвлөгөө:**
Програм бичихэд баримтлах үйл явц.

**баримтжуулах мөр:**
Функцийн интерфейсийг тайлбарлахаар тодорхойлолтынх нь эхэнд бичсэн тэмдэгт мөр.

**олон мөрт тэмдэгт мөр:**
Гурвалсан хашилтаар хүрээлсэн, кодын хэд хэдэн мөр дамнаж болох тэмдэгт мөр.

**урьдчилсан нөхцөл:**
Функц ажиллаж эхлэхээс өмнө түүнийг дуудагч хангах ёстой шаардлага.

**төгсгөлийн нөхцөл:**
Функц дуусахаасаа өмнө хангах ёстой шаардлага.


## Дасгалууд

Доорх дасгалын хувьд яст мэлхийн хэд хэдэн функцийг ашиглахыг хүсэж болох юм.

* `penup` яст мэлхий хөдлөх үед мөр үлдээхгүйн тулд түүний төсөөллийн үзгийг өргөдөг.

* `pendown` үзэгээ буцааж тавина.

Дараах функц нь яст мэлхийг мөр үлдээлгүй хөдөлгөхийн тулд `penup` болон `pendown` ашигладаг.

````python
from jupyturtle import penup, pendown

def jump(length):
    """Move forward length units without leaving a trail.
    
    Postcondition: Leaves the pen down.
    """
    penup()
    forward(length)
    pendown()
````

## Дасгал

Өгөгдсөн хажуугийн урт, өгөгдсөн дотоод өнцөг бүхий ромбыг зурах `rhombus` функцийг бич. Жишээлбэл, энд хажуугийн урт `50`, дотоод өнцөг нь `60` градустай ромбус байна.

````python
def rhombus(length, angle):
    for i in range(2):
        forward(length)
        left(angle)
        forward(length)
        left(180-angle)
````

Та функцээ шалгахын тулд дараах кодыг ашиглаж болно.

````python
make_turtle()
rhombus(50, 60)
````
