# Жагсаалт

Энэ бүлэгт Python-ийн хамгийн хэрэгтэй суурилагдсан төрлүүдийн нэг болох жагсаалтыг танилцуулж байна.
Та мөн объектуудын талаар илүү ихийг мэдэх болно, олон хувьсагч нь нэг объектод хамаарах үед юу тохиолдож болох талаар мэдэх болно.

Бүлгийн төгсгөлд байгаа дасгалууд дээр бид үгийн жагсаалт гаргаж, түүгээрээ палиндром, анаграм зэрэг тусгай үгсийг хайх болно.

## Жагсаалт бол дараалал

Тэмдэгт мөртэй адил **жагсаалт** нь утгуудын дараалал юм. Тэмдэгт мөрийн утгууд тэмдэгт байдаг бол жагсаалтын утгууд дурын төрөлтэй байж болно. Жагсаалт доторх утгуудыг **элемент** гэж нэрлэдэг.

Шинэ жагсаалт үүсгэх хэд хэдэн арга бий. Хамгийн энгийн нь элементүүдийг дөрвөлжин хаалтанд (`[` ба `]`) бичих юм. Жишээлбэл, дараах жагсаалт хоёр бүхэл тоо агуулна.

```python
numbers = [42, 123]
```

Мөн энд гурван мөрний жагсаалт байна.

```python
cheeses = ['Cheddar', 'Edam', 'Gouda']
```

Жагсаалтын элементүүд нь нэг төрлийн байх албагүй.
Дараах жагсаалт нь мөр, хөвөгч, бүхэл тоо, тэр ч байтугай өөр жагсаалтыг агуулдаг.

```python
t = ['spam', 2.0, 5, [10, 20]]
```

Өөр жагсаалт доторх жагсаалт нь **үүрлэсэн** байна.

Элемент агуулаагүй жагсаалтыг хоосон жагсаалт гэж нэрлэдэг; та үүсгэж болно
нэг нь хоосон хаалттай, `[]`.

```python
empty = []
```

`len` функц нь жагсаалтын уртыг буцаана.

```python
len(cheeses)
```

Хоосон жагсаалтын урт нь `0` байна.

```python
len(empty)
```

Дараах зурагт `cheeses`, `numbers` болон `empty`-ийн төлөвийн диаграммыг харуулав.

```python
from diagram import make_list, Binding, Value

list1 = make_list(cheeses, dy=-0.3, offsetx=0.17)
binding1 = Binding(Value('cheeses'), list1)

list2 = make_list(numbers, dy=-0.3, offsetx=0.17)
binding2 = Binding(Value('numbers'), list2)

list3 = make_list(empty, dy=-0.3, offsetx=0.1)
binding3 = Binding(Value('empty'), list3)
```

```python
from diagram import diagram, adjust, Bbox

width, height, x, y = [3.66, 1.58, 0.45, 1.2]
ax = diagram(width, height)
bbox1 = binding1.draw(ax, x, y)
bbox2 = binding2.draw(ax, x+2.25, y)
bbox3 = binding3.draw(ax, x+2.25, y-1.0)

bbox = Bbox.union([bbox1, bbox2, bbox3])
adjust(x, y, bbox)

import matplotlib.pyplot as plt
plt.savefig('state_diagram_list.png')
```

Жагсаалтыг гадна талд нь "жагсаалт" гэсэн үг, дотор нь жагсаалтын дугаарласан элементүүдтэй хайрцагнуудаар төлөөлдөг.

## Жагсаалтыг өөрчилж болдог

Жагсаалтын элементийг уншихын тулд хаалт операторыг ашиглаж болно.
Эхний элементийн индекс нь `0` юм.

```python
cheeses[0]
```

Мөрүүдээс ялгаатай нь жагсаалт нь өөрчлөгдөх боломжтой. Хаалтны оператор гарч ирэх үед
даалгаврын зүүн талд, энэ нь жагсаалтын элементийг тодорхойлдог
үүнийг томилох болно.

```python
numbers[1] = 17
numbers
```

Өмнө нь `123` байсан `numbers` хоёр дахь элемент нь одоо `17` болсон.

Жагсаалтын индексүүд нь мөрийн индексүүдтэй адилхан ажилладаг:

- Аливаа бүхэл илэрхийллийг индекс болгон ашиглаж болно.

- Хэрэв та байхгүй элементийг унших, бичих гэж оролдвол олж авна
    `IndexError`.

- Хэрэв индекс сөрөг утгатай бол төгсгөлөөс хойш хойш тоолно
    жагсаалт.

`in` оператор нь жагсаалтууд дээр ажилладаг бөгөөд өгөгдсөн элемент жагсаалтын аль нэг хэсэгт байгаа эсэхийг шалгадаг.

```python
'Edam' in cheeses
```

```python
'Wensleydale' in cheeses
```

Жагсаалт нь өөр жагсаалт агуулж болох ч, үүрлэсэн жагсаалт нь нэг элемент гэж тооцогддог тул дараах жагсаалтад зөвхөн дөрвөн элемент байна.

```python
t = ['spam', 2.0, 5, [10, 20]]
len(t)
```

## Жагсаалт ба тэмдэгт мөр

Мөр нь тэмдэгтүүдийн дараалал, жагсаалт нь утгуудын дараалал юм.
гэхдээ тэмдэгтүүдийн жагсаалт нь тэмдэгт мөртэй адил биш юм. 
Мөрөөс тэмдэгтүүдийн жагсаалт руу хөрвүүлэхийн тулд та `list` функцийг ашиглаж болно.

```python
s = 'spam'
t = list(s)
t
```

`list` функц нь мөрийг тус тусад нь үсэг болгон хуваадаг.
Хэрэв та мөрийг үг болгон хуваахыг хүсвэл `split` аргыг ашиглаж болно.

```python
s = 'pining for the fjords'
t = s.split()
t
```

**Хязгаарлагч** гэж нэрлэгддэг нэмэлт аргумент нь ямар тэмдэгтүүдийг үгийн хил болгон ашиглахыг зааж өгдөг. Дараах жишээнд зураасыг зааглагч болгон ашиглаж байна.

```python
s = 'ex-parrot'
t = s.split('-')
t
```

Хэрэв танд мөрүүдийн жагсаалт байгаа бол тэдгээрийг `join` ашиглан нэг мөр болгон нэгтгэж болно.
`join` нь мөрийн арга тул та үүнийг хязгаарлагч дээр дуудаж, жагсаалтыг аргумент болгон дамжуулах хэрэгтэй.

```python
delimiter = ' '
t = ['pining', 'for', 'the', 'fjords']
s = delimiter.join(t)
s
```

Энэ тохиолдолд хязгаарлагч нь зайны тэмдэгт тул `join` зай тавина
үгсийн хооронд.
Мөрүүдийг хоосон зайгүйгээр холбохын тулд та `''` хоосон мөрийг хязгаарлагч болгон ашиглаж болно.

## Жагсаалтын элементүүдээр давтах

## Объект ба утга

Хэрэв бид эдгээр даалгаврын мэдэгдлийг ажиллуулбал:

```python
a = 'banana'
b = 'banana'
```

`a` болон `b` хоёулаа тэмдэгт мөрийг хэлдэг гэдгийг бид мэднэ, гэхдээ тэдгээр нь *ижил* мөрт хамаарах эсэхийг бид мэдэхгүй. 
Дараах зурагт үзүүлсэн хоёр боломжит төлөв байдаг.

```python
from diagram import Frame, Stack

s = 'banana'
bindings = [Binding(Value(name), Value(repr(s))) for name in 'ab']
frame1 = Frame(bindings, dy=-0.25)

binding1 = Binding(Value('a'), Value(repr(s)), dy=-0.11)
binding2 = Binding(Value('b'), draw_value=False, dy=0.11)
frame2 = Frame([binding1, binding2], dy=-0.25)

stack = Stack([frame1, frame2], dx=1.7, dy=0)
```

```python
width, height, x, y = [2.85, 0.76, 0.17, 0.51]
ax = diagram(width, height)
bbox = stack.draw(ax, x, y)
# adjust(x, y, bbox)
```

Зүүн талд байгаа диаграммд `a` болон `b` нь хоёр өөр объектыг хэлж байна.
ижил утга. Баруун талд байгаа диаграммд тэд ижил объектыг харуулдаг.
Хоёр хувьсагч нэг объектод хамаарах эсэхийг шалгахын тулд та `is` операторыг ашиглаж болно.

```python
a = 'banana'
b = 'banana'
a is b
```

Энэ жишээнд Python нь зөвхөн нэг string объектыг үүсгэсэн ба `a` хоёулаа
болон `b` үүнтэй холбогдоно.
Гэхдээ та хоёр жагсаалт үүсгэх үед хоёр объект авах болно.

```python
a = [1, 2, 3]
b = [1, 2, 3]
a is b
```

Тиймээс төлөвийн диаграм нь иймэрхүү харагдаж байна.

```python
t = [1, 2, 3]
binding1 = Binding(Value('a'), Value(repr(t)))
binding2 = Binding(Value('b'), Value(repr(t)))
frame = Frame([binding1, binding2], dy=-0.25)
```

```python
width, height, x, y = [1.16, 0.76, 0.21, 0.51]
ax = diagram(width, height)
bbox = frame.draw(ax, x, y)
# adjust(x, y, bbox)
```

Энэ тохиолдолд бид хоёр жагсаалт нь ижил элементтэй, гэхдээ **ижил биш**учир ижил объект биш учраас**эквивалент** гэж хэлэх болно. 
Хэрэв хоёр объект ижил байвал тэдгээр нь мөн адил тэнцүү байх боловч хэрэв тэд тэнцүү бол тэдгээр нь ижил байх албагүй.

## Жагсаалтыг аргумент болгон дамжуулах

Жагсаалтыг функц руу дамжуулахад функц нь тухайн функцийн лавлагааг авдаг
жагсаалт. Хэрэв функц жагсаалтад өөрчлөлт оруулбал залгагч өөрчлөлтийг харна. Учир нь
Жишээ нь, `pop_first` жагсаалтаас эхний элементийг устгахын тулд `pop` жагсаалтын аргыг ашигладаг.

```python
def pop_first(lst):
    return lst.pop(0)
```

Бид үүнийг ингэж ашиглаж болно.

```python
letters = ['a', 'b', 'c']
pop_first(letters)
```

Буцаах утга нь жагсаалтаас хасагдсан эхний элемент бөгөөд өөрчилсөн жагсаалтыг харуулах замаар харж болно.

```python
letters
```

Энэ жишээн дээр `lst` параметр ба `letters` хувьсагч нь ижил объектын нэр байх тул стек диаграмм дараах байдалтай байна.

```python
lst = make_list('abc', dy=-0.3, offsetx=0.1)
binding1 = Binding(Value('letters'), draw_value=False)
frame1 = Frame([binding1], name='__main__', loc='left')

binding2 = Binding(Value('lst'), draw_value=False, dx=0.61, dy=0.35)
frame2 = Frame([binding2], name='pop_first', loc='left', offsetx=0.08)

stack = Stack([frame1, frame2], dx=-0.3, dy=-0.5)
```

```python
width, height, x, y = [2.04, 1.24, 1.06, 0.85]
ax = diagram(width, height)
bbox1 = stack.draw(ax, x, y)
bbox2 = lst.draw(ax, x+0.5, y)
bbox = Bbox.union([bbox1, bbox2])
adjust(x, y, bbox)
```

Объект руу өгсөн лавлагааг функцэд аргумент болгон дамжуулах нь өөр нэр өгөх хэлбэрийг үүсгэдэг.

`remove` нь жагсаалтыг өөрчилж, `None`-г буцаадаг тул таны `t`-тай хийх дараагийн үйлдэл амжилтгүй болох магадлалтай.

Энэ алдааны мессежийг тайлбарлах шаардлагатай.

Хэрэв та үүнтэй төстэй алдааны мессежийг харвал та програмыг арагшаа харж, жагсаалтын аргыг буруу дуудсан эсэхээ шалгаарай.

## Нэр томьёоны тайлбар

**жагсаалт:**
Утгуудын дараалал агуулсан объект.

**элемент:**
Жагсаалт буюу бусад дараалал доторх утгуудын нэг.

**давхарласан жагсаалт:**
Өөр жагсаалтын элемент болсон жагсаалт.

**тусгаарлагч:**
Тэмдэгт мөрийг хаанаас хуваахыг заах тэмдэгт буюу тэмдэгт мөр.

**ижил утгатай:**
Утгууд нь адил байх.

**нэг объект байх:**
Яг нэг объектыг зааж байх. Энэ нь мөн ижил утгатай гэсэн үг.

**заалт:**
Хувьсагч болон түүний утгын хоорондох холбоо.

**олон нэрээр заагдах:**
Нэг объектыг нэгээс олон хувьсагч зааж байвал тэр объект олон нэрээр заагдаж байна гэсэн үг.

**атрибут:**
Объекттой холбоотой нэр бүхий утгуудын нэг.
