# tuple төрөл

Энэ бүлэгт Python-ы өөр нэг суурилагдсан төрөл болох tuple төрлийг танилцуулж, жагсаалт, dict төрөл, tuple төрлийг хэрхэн хамтад нь ашиглахыг үзнэ. Мөн tuple төрөл ашиглан утга олгох үйлдэл болон хувьсах тооны аргументтай функцэд хэрэгтэй багцлах, задлах операторуудтай танилцана.

Дасгалуудад tuple төрлийг жагсаалт, dict төрөлтэй хослуулан үгийн тааврууд бодож, үр ашигтай алгоритмууд хэрэгжүүлнэ.

Нэг тайлбар: англиар «tuple» гэдэг үгийг хоёр янзаар дууддаг. Зарим хүн «supple» гэдэг үгтэй төстэйгөөр «tuh-ple» гэж дууддаг. Харин програмчлалын хүрээнд ихэнх хүн «quadruple» гэдэг үгийн төгсгөлтэй төстэйгөөр «too-ple» гэж дууддаг.

## tuple төрөл нь жагсаалттай төстэй

tuple төрөл бол утгуудын дараалал юм. Утгууд нь дурын төрөлтэй байж болох бөгөөд бүхэл тоогоор индексжүүлдэг тул жагсаалттай их төстэй. Гол ялгаа нь tuple төрлийг өөрчилж болдоггүй.

tuple төрөл үүсгэхдээ утгуудыг таслалаар тусгаарлан бичиж болно.

```python
t = 'l', 'u', 'p', 'i', 'n'
type(t)
```

Заавал биш ч гэсэн хаалтанд tuple төрлүүдийг бичих нь элбэг.

```python
t = ('l', 'u', 'p', 'i', 'n')
type(t)
```

Ганц элементтэй tuple төрөл үүсгэхийн тулд та эцсийн таслалыг оруулах ёстой.

```python
t1 = 'p',
type(t1)
```

Хаалтанд байгаа ганц утга нь хэлхээ биш юм.

```python
t2 = ('p')
type(t2)
```

tuple төрөл үүсгэх өөр нэг арга бол суурилуулсан `tuple` функц юм. Үгүй
аргумент, энэ нь хоосон tuple төрөл үүсгэдэг.

```python
t = tuple()
t
```

Хэрэв аргумент нь дараалал (мөр, жагсаалт эсвэл хэлхээ) байвал үр дүн нь a
дарааллын элементүүдтэй tuple төрөл.

```python
t = tuple('lupin')
t
```

`tuple` нь суурилагдсан функцийн нэр учраас та үүнийг хувьсагчийн нэр болгон ашиглахаас зайлсхийх хэрэгтэй.

Жагсаалтын ихэнх операторууд мөн tuple төрлүүдтэй ажилладаг.
Жишээлбэл, хаалт оператор нь элементийг индексжүүлдэг.

```python
t[0]
```

Мөн зүсмэлийн оператор нь хэд хэдэн элементийг сонгоно.

```python
t[1:3]
```

`+` оператор нь tuple төрлүүдийг нэгтгэдэг.

```python
tuple('lup') + ('i', 'n')
```

Мөн `*` оператор нь өгөгдсөн тооны давтамжийг хуулбарладаг.

```python
tuple('spam') * 2 
```

`sorted` функц нь tuple төрлүүдтэй ажилладаг боловч үр дүн нь жагсаалт биш харин жагсаалт юм.

```python
sorted(t)
```

`reversed` функц нь мөн tuple төрлүүдтэй ажилладаг.

```python
reversed(t)
```

Үр дүн нь `reversed` объект болох бөгөөд бид үүнийг жагсаалт эсвэл tuple төрөл болгон хувиргаж болно.

```python
tuple(reversed(t))
```

Өнөөг хүртэлх жишээн дээр тулгуурлан, tuple төрөлs нь жагсаалттай адилхан юм шиг санагдаж магадгүй юм.

## tuple төрөл ашиглан утга олгох

Та даалгаврын зүүн талд хувьсагчийн багц, баруун талд нь утгын багц тавьж болно.

```python
a, b = 1, 2
```

Хувьсагчдад утгуудыг зүүнээс баруун тийш хуваарилдаг -- энэ жишээнд `a` нь `1` утгыг, `b` нь `2` утгыг авдаг.
Бид үр дүнг дараах байдлаар харуулах боломжтой.

```python
a, b
```

Ерөнхийдөө, хэрэв даалгаврын зүүн тал нь tuple төрөл бол баруун тал нь мөр, жагсаалт эсвэл tuple төрөл гэсэн ямар ч төрлийн дараалал байж болно. 
Жишээлбэл, имэйл хаягийг хэрэглэгчийн нэр болон домайн болгон хуваахын тулд та дараахийг бичиж болно:

```python
email = 'monty@python.org'
username, domain = email.split('@')
```

`split`-аас буцаах утга нь хоёр элемент бүхий жагсаалт бөгөөд эхний элементийг `username`, хоёр дахь нь `domain`-д оноогдсон.

```python
username, domain
```

Зүүн талд байгаа хувьсагчийн тоо, утгын тоо
зөв байх ёстой -- тэгэхгүй бол та `ValueError` авна.

```python
%%expect ValueError
a, b = 1, 2, 3
```

Хэрэв та хоёр хувьсагчийн утгыг солихыг хүсвэл tuple төрөл оноолт хэрэгтэй.
Уламжлалт даалгаврын хувьд та дараахтай адил түр зуурын хувьсагчийг ашиглах хэрэгтэй:

```python
temp = a
a = b
b = temp
```

Энэ нь ажилладаг, гэхдээ tuple төрөл хуваарилалтаар бид түр зуурын хувьсагчгүйгээр ижил зүйлийг хийж чадна.

```python
a, b = b, a
```

Баруун талд байгаа бүх илэрхийлэл нь аливаа даалгаврын өмнө үнэлэгдсэн тул энэ нь ажилладаг.

Бид мөн `for` заавард tuple төрөл оноолтыг ашиглаж болно.
Жишээлбэл, dict төрөлд байгаа зүйлсийг эргүүлэхийн тулд бид `items` аргыг ашиглаж болно.

```python
d = {'one': 1, 'two': 2}

for item in d.items():
    key, value = item
    print(key, '->', value)
```

Давталтыг давах бүрд `item`-д түлхүүр болон харгалзах утгыг агуулсан tuple төрөл оноогддог.

Бид энэ давталтыг дараах байдлаар илүү товчоор бичиж болно.

```python
for key, value in d.items():
    print(key, '->', value)
```

Давталтг дамжих бүрт түлхүүр болон холбогдох утгыг `key` болон `value`-д шууд оноодог.

## tuple төрлийг буцаах утга болгон ашиглах

Хатуухан хэлэхэд функц нь зөвхөн нэг утгыг буцаана, гэхдээ хэрэв
утга нь tuple төрөл, үр нөлөө нь олон утгыг буцаахтай ижил байна.
Жишээлбэл, хэрэв та хоёр бүхэл тоог хувааж, хуваахыг хүсвэл
үлдсэн тохиолдолд `x//y`, дараа нь `x%y`-г тооцоолох нь үр ашиггүй юм. Энэ нь
хоёуланг нь зэрэг тооцоолох нь дээр.

Суурилуулсан функц `divmod` нь хоёр аргумент авч, tuple төрөл буцаана
хоёр утгын хуваалт ба үлдэгдэл.

```python
divmod(7, 3)
```

Бид tuple төрөл-ийн элементүүдийг хоёр хувьсагчаар хадгалахын тулд tuple төрөл оноолтыг ашиглаж болно.

```python
quotient, remainder = divmod(7, 3)
quotient
```

```python
remainder
```

Толбо буцаадаг функцийн жишээ энд байна.

```python
def min_max(t):
    return min(t), max(t)
```

`max` болон `min` нь дарааллын хамгийн том ба хамгийн жижиг элементүүдийг олох суурилагдсан функцүүд юм. 
`min_max` хоёуланг нь тооцоолж, хоёр утгын багцыг буцаана.

```python
min_max([2, 4, 1, 3])
```

Бид үр дүнг дараах байдлаар хувьсагчдад оноож болно:

```python
low, high = min_max([2, 4, 1, 3])
low, high
```

## Аргументуудыг багцлах

Функцүүд нь хувьсах тооны аргумент авч болно. 
`*` оператороор эхэлсэн параметрийн нэр **аргументуудыг багц болгон цуглуулдаг.
Жишээлбэл, дараах функц нь дурын тооны аргументыг авч, тэдгээрийн арифметик дундажийг, өөрөөр хэлбэл тэдгээрийн нийлбэрийг аргументуудын тоонд хуваана.

```python
def mean(*args):
    return sum(args) / len(args)
```

Параметр нь таны дуртай ямар ч нэртэй байж болно, гэхдээ `args` уламжлалт байна.
Бид функцийг дараах байдлаар дуудаж болно.

```python
mean(1, 2, 3)
```

Хэрэв танд утгуудын дэс дараалал байгаа бөгөөд тэдгээрийг олон аргумент болгон функцэд шилжүүлэхийг хүсвэл `*` операторыг ашиглан багцыг **задлах** боломжтой.
Жишээлбэл, `divmod` нь яг хоёр аргумент авдаг -- хэрэв та tuple төрөл-ийг параметр болгон дамжуулбал алдаа гарна.

```python
%%expect TypeError
t = (7, 3)
divmod(t)
```

Хэдийгээр энэ нь хоёр элементийг агуулж байгаа ч энэ нь нэг аргумент гэж тооцогддог.
Гэхдээ хэрэв та tuple төрлийг задлах юм бол үүнийг хоёр аргумент гэж үзнэ.

```python
divmod(*t)
```

Хэрэв та одоо байгаа функцийн зан төлөвийг тохируулахыг хүсвэл савлах, задлах нь ашигтай байж болно.
Жишээлбэл, энэ функц нь ямар ч тооны аргументыг авч, хамгийн бага ба хамгийн ихийг хасч, үлдсэн хэсгийн дундажийг тооцоолно.

```python
def trimmed_mean(*args):
    low, high = min_max(args)
    trimmed = list(args)
    trimmed.remove(low)
    trimmed.remove(high)
    return mean(*trimmed)
```

Эхлээд `min_max`-г ашиглан хамгийн бага ба хамгийн өндөр элементүүдийг олно.
Дараа нь `args`-г жагсаалт болгон хувиргаснаар `remove` аргыг ашиглах боломжтой болно.
Эцэст нь энэ нь жагсаалтыг задлах бөгөөд ингэснээр элементүүдийг `mean` руу нэг жагсаалтаар бус тусдаа аргумент болгон дамжуулдаг.

Үр нөлөөг харуулсан жишээ энд байна.

```python
mean(1, 2, 3, 10)
```

```python
trimmed_mean(1, 2, 3, 10)
```

Усанд шумбах, гимнастик гэх мэт субьектив шүүлт бүхий зарим спортод оноо нь бусдаас зөрүүтэй шүүгчийн нөлөөг багасгахын тулд ийм "тайрсан" дундажийг ашигладаг.

## zip функц

tuple төрөлs нь хоёр дарааллын элементүүдийг давтах, харгалзах элементүүд дээр үйлдлүүдийг гүйцэтгэхэд хэрэгтэй.
Жишээ нь, хоёр баг долоон тоглолтын цуврал тоглосон гэж бодъё, бид тэдний оноог хоёр жагсаалтад, баг бүрт нэг нэгээр нь бичдэг гэж бодъё.

```python
scores1 = [1, 2, 4, 5, 1, 5, 2]
scores2 = [5, 5, 2, 2, 5, 2, 3]
```

Баг бүр хэдэн тоглолтонд хожсоныг харцгаая.
Бид `zip`-г ашиглах бөгөөд энэ нь хоёр ба түүнээс дээш дараалал авч, цахилгаан товчны шүд шиг дарааллын элементүүдийг хослуулсан тул **zip объект**-г буцаадаг суурилуулсан функц юм.

```python
zip(scores1, scores2)
```

Бид zip объектыг ашиглан дарааллын утгуудыг хосоор нь дамжуулж болно.

```python
for pair in zip(scores1, scores2):
     print(pair)
```

Давталтг дамжих бүрт `pair`-д олон тооны оноо оноогддог.
Тиймээс бид хувьсагчдад оноо оноож, эхний багийн ялалтыг дараах байдлаар тоолж болно.

```python
wins = 0
for team1, team2 in zip(scores1, scores2):
    if team1 > team2:
        wins += 1
        
wins
```

Харамсалтай нь нэгдүгээр баг гуравхан хожиж, цувралд хожигдсон.

Хэрэв танд хоёр жагсаалт байгаа бөгөөд та хосын жагсаалтыг хүсч байвал `zip` болон `list`-г ашиглаж болно.

```python
t = list(zip(scores1, scores2))
t
```

Үр дүн нь tuple төрлүүдийн жагсаалт бөгөөд бид сүүлийн тоглолтын үр дүнг дараах байдлаар авч болно.

```python
t[-1]
```

Хэрэв танд товчлуурын жагсаалт болон утгын жагсаалт байгаа бол та `zip`, `dict` зэргийг ашиглан dict төрөл хийж болно.
Жишээлбэл, үсэг бүрээс цагаан толгойн дээрх байрлалыг харуулсан dict төрлийг хэрхэн хийх талаар эндээс үзнэ үү.

```python
letters = 'abcdefghijklmnopqrstuvwxyz'
numbers = range(len(letters))
letter_map = dict(zip(letters, numbers))
```

Одоо бид үсэг хайж, цагаан толгойн үсгийн индексийг авах боломжтой.

```python
letter_map['a'], letter_map['z']
```

Энэ зураглалд `a` индекс нь `0`, `z` индекс нь `25` байна.

Хэрэв та дарааллын элементүүд болон тэдгээрийн индексүүдийг хооронд нь эргүүлэх шаардлагатай бол `enumerate` суулгасан функцийг ашиглаж болно.

```python
enumerate('abc')
```

Үр дүн нь хосуудын дарааллаар дамжих **объектыг тоолох** бөгөөд хос тус бүр нь индекс (0-ээс эхлэн) болон өгөгдсөн дарааллын элементийг агуулна.

```python
for index, element in enumerate('abc'):
    print(index, element)
```

## Харьцуулах ба эрэмбэлэх

Харьцангуй операторууд нь tuple төрөл болон бусад дараалалтай ажилладаг.
Жишээлбэл, хэрэв та `<` операторыг tuple төрөлтэй ашигладаг бол дараалал бүрийн эхний элементийг харьцуулж эхэлнэ.
Хэрэв тэдгээр нь тэнцүү бол дараагийн хос элемент рүү шилжих гэх мэтээр ялгаатай хосыг олох хүртэл үргэлжилнэ.

```python
(0, 1, 2) < (0, 3, 4)
```

Дараагийн элементүүд нь үнэхээр том байсан ч тооцогдохгүй.

```python
(0, 1, 2000000) < (0, 3, 4)
```

Хавтангуудыг харьцуулах ийм арга нь хэлхээний жагсаалтыг эрэмбэлэх эсвэл хамгийн бага эсвэл дээд хэмжээг олоход хэрэгтэй.
Жишээ болгон үгэнд хамгийн түгээмэл үсгийг олъё.
Өмнөх бүлэгт бид `value_counts` бичсэн бөгөөд энэ нь тэмдэгт мөр авч, үсэг бүрээс хэдэн удаа гарч ирэхийг харуулсан dict төрлийг буцаадаг.

```python
def value_counts(string):
    counter = {}
    for letter in string:
        if letter not in counter:
            counter[letter] = 1
        else:
            counter[letter] += 1
    return counter
```

`banana` мөрийн үр дүн энд байна.

```python
counter = value_counts('banana')
counter
```

Гуравхан зүйлээр бид хамгийн их тохиолддог үсэг нь гурван удаа гарч ирдэг `a` гэдгийг хялбархан харж болно.
Гэхдээ илүү олон зүйл байсан бол тэдгээрийг автоматаар ангилах нь ашигтай байх болно.

Бид `counter`-аас ийм зүйлсийг авч болно.

Та утгыг хайж, харгалзах түлхүүрийг авахын тулд dict төрлийг эргүүлэхийг хүсч байна гэж бодъё.
Жишээлбэл, хэрэв танд үг бүрээс хэдэн удаа гарч ирэхийг заадаг үг тоолуур байгаа бол бүхэл тооноос хэдэн удаа гарч ирэх үг хүртэл буулгадаг dict төрөл хийж болно.

Гэхдээ нэг асуудал бий - dict төрлийн түлхүүрүүд нь өвөрмөц байх ёстой, гэхдээ утга нь тийм биш юм. Жишээлбэл, үгийн тоолуурт ижил тоотой олон үг байж болно.

Тиймээс dict төрлийг хөрвүүлэх нэг арга бол шинэ dict төрөл үүсгэх явдал бөгөөд утгууд нь эх хувилбарын түлхүүрүүдийн жагсаалт юм.
Жишээ болгон `parrot` үсгүүдийг тоолж үзье.

```python
d =  value_counts('parrot')
d
```

Хэрэв бид энэ dict төрлийг эргүүлбэл `{1: ['p', 'a', 'o', 't'], 2: ['r']}` үр дүн гарах ёстой бөгөөд энэ нь нэг удаа гарч ирэх үсэг нь `p`, `a`, `o`, `t`, хоёр удаа гарч байгаа үсэг нь `2` болохыг харуулж байна.

Дараах функц нь dict төрлийг авч, түүний урвууг шинэ dict төрөл болгон буцаана.

```python
def invert_dict(d):
    new = {}
    for key, value in d.items():
        if value not in new:
            new[value] = [key]
        else:
            new[value].append(key)
    return new
```

`for` заавар нь `d` доторх түлхүүрүүд болон утгуудаар дамждаг.
Хэрэв утга нь шинэ dict төрөлд хараахан ороогүй бол энэ нь нэг элемент агуулсан жагсаалтад нэмэгдэж, холбогдох болно.
Үгүй бол энэ нь одоо байгаа жагсаалтад хавсаргасан болно.

Бид үүнийг дараах байдлаар шалгаж болно.

```python
invert_dict(d)
```
