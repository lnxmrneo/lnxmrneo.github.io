# Буцаах утга

Өмнөх бүлгүүдэд бид `abs`, `round` зэрэг суурилуулсан функцүүд болон математикийн модуль дахь `sqrt`, `pow` зэрэг функцүүдийг ашигласан.
Эдгээр функцүүдийн аль нэгийг дуудах үед энэ нь хувьсагчид оноож болох эсвэл илэрхийллийн хэсэг болгон ашиглаж болох утгыг буцаана.

Бидний өнөөг хүртэл бичсэн функцүүд нь өөр.
Зарим нь утгыг харуулахын тулд `print` функцийг ашигладаг бол зарим нь дүрс зурахдаа `Turtle` функцийг ашигладаг.
Гэхдээ тэдгээр нь бидний хувьсагчдад оноож өгсөн эсвэл илэрхийлэлд ашигладаг утгыг буцаадаггүй.

Энэ бүлэгт бид утгыг буцаадаг функцүүдийг хэрхэн бичихийг үзэх болно.

## Зарим функц утга буцаадаг

`math.sqrt` гэх мэт функцийг дуудах үед үр дүнг **буцаах утга** гэж нэрлэдэг.
Хэрэв функцийн дуудлага нүдний төгсгөлд гарч ирвэл Jupyter буцаах утгыг шууд харуулна.

```python
import math

math.sqrt(42 / math.pi)
```

Хэрэв та хувьсагчид буцаах утгыг оноох юм бол энэ нь харагдахгүй.

```python
radius = math.sqrt(42 / math.pi)
```

Гэхдээ та дараа нь харуулах боломжтой.

```python
radius
```

Эсвэл буцаах утгыг илэрхийллийн хэсэг болгон ашиглаж болно.

```python
radius + math.sqrt(42 / math.pi)
```

Утга буцаадаг функцийн жишээ энд байна.

```python
def circle_area(radius):
    area = math.pi * radius**2
    return area
```

`circle_area` нь `radius`-г параметр болгон авч тухайн радиустай тойргийн талбайг тооцоолно.

Сүүлийн мөр нь `area` утгыг буцаадаг `return` заавар юм.

Хэрэв бид функцийг ингэж дуудвал Jupyter буцаах утгыг харуулна.


```python
circle_area(radius)
```

Бид хувьсагч руу буцаах утгыг оноож болно.

```python
a = circle_area(radius)
```

Эсвэл илэрхийллийн хэсэг болгон ашиглах.

```python
circle_area(radius) + 2 * circle_area(radius / 2)
```

Дараа нь бид үр дүнг өгсөн хувьсагчийн утгыг харуулах боломжтой.

```python
a
```

Гэхдээ бид `area` руу нэвтрэх боломжгүй.

```python
%%expect NameError

area
```

`area` нь функц доторх локал хувьсагч тул функцийн гаднаас хандах боломжгүй.

## Зарим функц None буцаадаг

Хэрэв функц `return` зааваргүй бол `None`-г буцаадаг бөгөөд энэ нь `True` болон `False` зэрэг тусгай утга юм.
Жишээлбэл, 3-р бүлгийн `repeat` функцийг энд үзүүлэв.

```python
def repeat(word, n):
    print(word * n)
```

Хэрэв бид үүнийг ингэж нэрлэвэл энэ нь Monty Python дууны "Финланд" дууны эхний мөрийг харуулна.

```python
repeat('Finland, ', 3)
```

Энэ функц нь мөрийг харуулахын тулд `print` функцийг ашигладаг боловч утгыг буцаахын тулд `return` мэдэгдлийг ашигладаггүй.
Хэрэв бид үр дүнг хувьсагчид оноох юм бол энэ нь ямар ч байсан мөрийг харуулна.

```python
result = repeat('Finland, ', 3)
```

Хэрэв бид хувьсагчийн утгыг харуулбал бид юу ч авахгүй.

```python
result
```

`result` үнэндээ утгатай боловч Jupyter үүнийг харуулдаггүй.
Гэсэн хэдий ч бид үүнийг ингэж харуулж чадна.

```python
print(result)
```

`repeat`-аас буцаах утга нь `None` байна.

Одоо энд `repeat`-тэй төстэй функц байна, гэхдээ буцаах утгатай.

```python
def repeat_string(word, n):
    return word * n
```

Бид зөвхөн хувьсагч биш харин `return` хэллэгт илэрхийлэл ашиглаж болохыг анхаарна уу.

Энэ хувилбарын тусламжтайгаар бид үр дүнг хувьсагчид оноож болно.
Функц ажиллаж байх үед юу ч харуулахгүй.

```python
line = repeat_string('Spam, ', 4)
```

Гэхдээ дараа нь бид `line`-д өгсөн утгыг харуулах боломжтой.

```python
line
```

Үүнтэй адил функцийг **цэвэр функц** гэж нэрлэдэг, учир нь энэ нь утгыг буцаахаас өөр ямар нэгэн зүйлийг харуулахгүй эсвэл өөр ямар нэгэн нөлөө үзүүлэхгүй.

## Буцаах утга ба нөхцөлт заавар

Хэрэв Python `abs` өгөөгүй бол бид ингэж бичиж болно.

```python
def absolute_value(x):
    if x < 0:
        return -x
    else:
        return x
```

Хэрэв `x` сөрөг байвал эхний `return` заавар `-x`-г буцаадаг бөгөөд функц тэр даруй дуусна.
Үгүй бол хоёр дахь `return` заавар нь `x`-г буцааж, функц дуусна.
Тэгэхээр энэ функц зөв байна.

Гэсэн хэдий ч, хэрэв та `return` мэдэгдлийг болзолт байдалд оруулбал програмаар дамжин өнгөрөх бүх зам `return` заавартэй таарч байгаа эсэхийг шалгах хэрэгтэй.
Жишээлбэл, энд `absolute_value`-ийн буруу хувилбар байна.

```python
def absolute_value_wrong(x):
    if x < 0:
        return -x
    if x > 0:
        return x
```

Хэрэв бид энэ функцийг `0` аргумент болгон дуудвал юу болох вэ.

```python
absolute_value_wrong(0)
```

Бид юу ч авдаггүй! Асуудал энд байна: `x` `0` үед аль ч нөхцөл үнэн биш бөгөөд функц `return` мэдэгдлийг дарахгүйгээр дуусдаг бөгөөд энэ нь буцаах утга нь `None` гэсэн үг тул Jupyter юу ч харуулахгүй.

Өөр нэг жишээ болгон, төгсгөлд нь нэмэлт `return` заавар бүхий `absolute_value` хувилбарыг энд оруулав.

```python
def absolute_value_extra_return(x):
    if x < 0:
        return -x
    else:
        return x
    
    return 'This is dead code'
```

Хэрэв `x` сөрөг байвал эхний `return` заавар ажиллаж, функц дуусна.
Үгүй бол хоёр дахь `return` заавар ажиллаж, функц дуусна.
Аль ч тохиолдолд бид гурав дахь `return` заавард хэзээ ч хүрч чадахгүй -- тиймээс энэ нь хэзээ ч ажиллахгүй.

Хэзээ ч ажиллах боломжгүй кодыг **үхсэн код** гэж нэрлэдэг.
Ерөнхийдөө үхсэн код нь ямар ч хор хөнөөл учруулдаггүй, гэхдээ энэ нь ихэвчлэн буруу ойлголтыг илтгэдэг бөгөөд энэ нь програмыг ойлгохыг оролдож буй хэн нэгэнд төөрөгдүүлж магадгүй юм.

## Логик утга буцаах функцууд

Функцүүд нь `True` болон `False` логик утгуудыг буцааж чаддаг бөгөөд энэ нь ихэвчлэн функцэд нарийн төвөгтэй тестийг багтаахад тохиромжтой байдаг.
Жишээлбэл, `is_divisible` `x` нь `y`-д үлдэгдэлгүй хуваагдах эсэхийг шалгадаг.

```python
def is_divisible(x, y):
    if x % y == 0:
        return True
    else:
        return False
```

Функц дотор `==` операторын үр дүн нь логик байх тул бид дараахыг бичиж болно.
шууд буцаах замаар илүү товчоор ажиллах боломжтой.

```python
def is_divisible(x, y):
    return x % y == 0
```

## Утга буцаадаг рекурс

Одоо бид буцаах утгуудтай функцүүдийг бичиж, буцаах утгуудаар рекурсив функцүүдийг бичиж болох ба энэ чадвараараа бид чухал босгыг давсан -- бидэнд байгаа Python-ын дэд олонлог **Turing бүрэн** болсон бөгөөд энэ нь бид алгоритмаар дүрсэлж болох аливаа тооцооллыг хийх боломжтой гэсэн үг юм.

Буцаах утга бүхий рекурсийг харуулахын тулд бид хэд хэдэн рекурсив тодорхойлогдсон математик функцийг үнэлэх болно.
Рекурсив тодорхойлолт нь дугуй тодорхойлолттой төстэй бөгөөд уг тодорхойлолт нь тодорхойлогдож буй зүйлд хамаатай. Жинхэнэ дугуй тодорхойлолт нь тийм ч ашигтай биш юм:

> vorpal: Ямар нэг зүйлийг тодорхойлоход хэрэглэгддэг нэр үг.

Хэрэв та dict төрөлд энэ тодорхойлолтыг харсан бол та бухимдаж магадгүй юм. 
Нөгөө талаас, хэрэв та $!$ тэмдгээр тэмдэглэсэн хүчин зүйлийн функцийн тодорхойлолтыг хайвал дараах зүйлийг олж авах боломжтой. 

$$\begin{aligned}
0! &= 1 \\
n! &= n~(n-1)!
\end{aligned}$$ 

Энэ тодорхойлолт нь 0-ийн факториал нь 1 бөгөөд бусад утгын $n$ нь $n$-г $n-1$-ийн факториалаар үржүүлсэн байна.

Хэрэв та ямар нэг зүйлийн рекурсив тодорхойлолтыг бичиж чадвал түүнийг үнэлэхийн тулд Python програм бичиж болно. 
Өсөн нэмэгдэж буй хөгжүүлэлтийн дараа бид `n`-г параметр болгон авч `0`-г буцаадаг функцээс эхэлнэ.

```python
def factorial(n):
    return 0
```

Одоо тодорхойлолтын эхний хэсгийг нэмье -- хэрвээ аргумент 0 болвол 1-ийг буцаахад л хангалттай.

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return 0
```

Одоо хоёр дахь хэсгийг бөглөцгөөе -- хэрвээ `n` нь `0` биш бол бид рекурсив хийх ёстой.
дуудаж $n-1$-ийн факториалыг олоод үр дүнг $n$-аар үржүүлнэ.

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        recurse = factorial(n-1)
        return n * recurse
```

Энэ програмын гүйцэтгэлийн урсгал нь 5-р бүлэгт байгаа `countdown`-ийн урсгалтай төстэй.
Хэрэв бид `factorial` утгыг `3` гэж нэрлэвэл:

`3` нь `0` биш тул бид хоёр дахь салбарыг авч факториалыг тооцдог.
`n-1`\...

> `2` нь `0` биш тул бид хоёр дахь салбарыг авч тооцоолно.
> `n-1`-ийн факториал\...
>
> > `1` нь `0` биш тул бид хоёр дахь салбарыг авч тооцоолно.
>> `n-1` факториал\...
> >
> > > `0` нь `0`-тэй тэнцүү тул бид эхний салбарыг аваад `1`-г ямар ч хамаагүй буцаана.
> > > өөр рекурсив дуудлага хийх.
> >
> > Буцаах утга болох `1` нь `n` буюу `1`-аар үржүүлж,
>> үр дүн буцаж байна.
>
> Буцаах утга болох `1`-г `n` буюу `2`-аар үржүүлж үр дүн гарна.
> буцаасан.

Буцаах утгыг `2` `n`-аар үржүүлж, `3` үр дүн нь,
`6` нь бүхэлд нь эхлүүлсэн функцийн дуудлагын буцаах утга болно
үйл явц.

Дараах зурагт функц дуудлагын энэ дарааллын стек диаграммыг харуулав.

```python
from diagram import Frame, Stack, make_binding

main = Frame([], name='__main__', loc='left')
frames = [main]

ns = 3, 2, 1
recurses = 2, 1, 1
results = 6, 2, 1

for n, recurse, result in zip(ns, recurses, results):
    binding1 = make_binding('n', n)
    binding2 = make_binding('recurse', recurse)
    frame = Frame([binding1, binding2], 
                  name='factorial', value=result,
                  loc='left', dx=1.2)
    frames.append(frame)
    
binding1 = make_binding('n', n)
frame = Frame([binding1], name='factorial', value=1, 
              shim=1.2, loc='left', dx=1.4)
frames.append(frame)

stack = Stack(frames, dy=-0.45)
```

Буцаах утгуудыг стек дээр шилжүүлж байгааг харуулав.
Фрейм бүрт буцаах утга нь `n` ба `recurse`-ийн үржвэр юм.

Сүүлийн фреймд `recurse` локал хувьсагч байхгүй, учир нь түүнийг үүсгэсэн салбар ажиллахгүй.

Мэдээж бичиж дуусаагүй байхад функц зөв ажиллаж байна гэж үзэх нь хачирхалтай, гэхдээ ийм учраас итгэлийн үсрэлт гэж нэрлэдэг!

## Фибоначчи

`factorial`-ийн дараа рекурсив функцийн хамгийн түгээмэл жишээ бол `fibonacci` бөгөөд дараах тодорхойлолттой: 

$$\begin{aligned}
\mathrm{fibonacci}(0) &= 0 \\
\mathrm{fibonacci}(1) &= 1 \\
\mathrm{fibonacci}(n) &= \mathrm{fibonacci}(n-1) + \mathrm{fibonacci}(n-2)
\end{aligned}$$ 

Python хэл рүү орчуулбал дараах байдалтай байна.

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif  n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2)
```

Хэрэв та $n$-ийн жижиг утгын хувьд ч гэсэн энд гүйцэтгэлийн урсгалыг дагахыг оролдвол таны толгой дэлбэрнэ.
Гэхдээ итгэлийн үсрэлтийн дагуу хэрэв та хоёр рекурсив дуудлага зөв ажилладаг гэж үзвэл сүүлчийн буцах заавар зөв гэдэгт итгэлтэй байж болно.

Одоо энд `factorial`-ийн алдаа шалгах хувилбар байна.

```python
def factorial(n):
    if not isinstance(n, int):
        print('factorial is only defined for integers.')
        return None
    elif n < 0:
        print('factorial is not defined for negative numbers.')
        return None
    elif n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

Эхлээд `n` бүхэл тоо эсэхийг шалгана.
Үгүй бол алдааны мессежийг харуулах ба `None`-г буцаана.

Хэрэв функц ажиллаж байгаа юм шиг байвал буцаах утгыг зөв ашиглаж байгаа эсвэл огт ашиглаж байгаа эсэхийг шалгахын тулд функцийн дуудлагыг харна уу!

Функцийн эхэн ба төгсгөлд `print` мэдэгдлийг нэмэх нь гүйцэтгэлийн урсгалыг илүү харагдахуйц болгоход тусална.
Жишээлбэл, энд хэвлэх заавар бүхий `factorial` хувилбар байна:

```python
def factorial(n):
    space = ' ' * (4 * n)
    print(space, 'factorial', n)
    if n == 0:
        print(space, 'returning 1')
        return 1
    else:
        recurse = factorial(n-1)
        result = n * recurse
        print(space, 'returning', result)
        return result
```

`space` нь догол мөрийг удирддаг зай тэмдэгтүүдийн мөр юм
гаралт. `factorial(4)`-ийн үр дүн энд байна:

```python
factorial(3)
```

Хэрэв та гүйцэтгэлийн урсгалын талаар эргэлзэж байвал ийм төрлийн гаралт нь тустай байж болно.
