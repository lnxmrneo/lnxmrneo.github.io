## Төлвийн диаграмм

Хувьсагчдыг цаасан дээр дүрслэх түгээмэл арга нь хувьсагчийн нэрийг бичиж, түүний зааж буй утга руу чиглэсэн сумаар холбох явдал юм. 

````python
import math
from graphviz import Digraph
from IPython.display import display

# Хувьсагчид утга олгох
message = 'And now for something completely different'
n = 17
pi = math.pi

# Диаграмм үүсгэх
dot = Digraph(format='png')
dot.attr(rankdir='LR')
dot.attr('node', shape='box', style='rounded')

# Хувьсагч болон утгууд
variables = {
    'message': message,
    'n': n,
    'pi': pi
}

for i, (name, value) in enumerate(variables.items()):
    dot.node(f'var{i}', name, shape='plaintext')
    dot.node(f'val{i}', repr(value))
    dot.edge(f'var{i}', f'val{i}')

# Диаграммыг харуулах
display(dot)
````

````python
import math
import matplotlib.pyplot as plt

# Хувьсагчид
message = 'And now for something completely different'
n = 17
pi = math.pi

# Зургийн хэмжээ
width, height = 7.5, 2.5
fig, ax = plt.subplots(figsize=(width, height))

# Координатын тохиргоо
ax.set_xlim(0, 7.5)
ax.set_ylim(0, 2.5)
ax.axis('off')

# Хувьсагчдын төлөв
variables = [
    ("n", str(n)),
    ("pi", str(pi)),
    ("message", repr(message))
]

# Диаграмм зурах
for i, (name, value) in enumerate(variables):
    y = 2.0 - i * 0.7

    ax.text(0.5, y, name,
            fontsize=12, va='center',
            family='monospace')

    ax.annotate(
        "",
        xy=(2.0, y),
        xytext=(1.4, y),
        arrowprops=dict(arrowstyle="->")
    )

    ax.text(2.2, y, value,
            fontsize=11, va='center',
            family='monospace')

plt.show()
````

Ийм төрлийн дүрслэлийг төлөвийн диаграмм (State Diagram) гэж нэрлэдэг. Учир нь уг диаграмм нь хувьсагч бүр ямар төлөвт байгааг харуулдаг.
Бид энэ номын турш Python хэл хувьсагчид болон тэдгээрийн утгуудыг хэрхэн хадгалдгийг загварчлан дүрслэхдээ төлөвийн диаграммыг ашиглах болно.

## Алдаа олж засварлах

Програмд гурван төрлийн алдаа гарч болно: синтакс алдаа, ажиллах үеийн алдаа, семантик алдаа.
Тэднийг илүү хурдан олохын тулд тэдгээрийг хооронд нь ялгах нь ашигтай байдаг.

* **Syntax error**: "Syntax" гэдэг нь програмын бүтэц, тухайн бүтцийн талаарх дүрмийг хэлнэ. Хэрэв таны програмын аль ч хэсэгт синтакс алдаа байвал Python програмыг ажиллуулдаггүй. Энэ нь алдааны мессежийг шууд харуулна.

* **Ажиллах үеийн алдаа**: Хэрэв таны программд синтаксийн алдаа байхгүй бол програм ажиллаж эхлэх боломжтой. Гэхдээ ямар нэг зүйл буруу болвол Python алдааны мессежийг харуулж зогсох болно. Энэ төрлийн алдааг ажиллуулах үеийн алдаа гэж нэрлэдэг. Энэ нь ямар нэг онцгой зүйл тохиолдсоныг илтгэдэг тул **онцгой нөхцөл** гэж бас нэрлэдэг.

* **Утгын алдаа**: Гурав дахь төрлийн алдаа нь "семантик" бөгөөд утга учиртай холбоотой гэсэн үг. Хэрэв таны програмд ​​семантик алдаа гарсан бол энэ нь алдааны мэдэгдэл үүсгэхгүйгээр ажилладаг боловч таны бодож байсан зүйлийг хийхгүй. Хөтөлбөрийн гаралтыг харж, юу хийж байгааг ойлгохыг оролдох замаар ухарч ажиллах шаардлагатай тул утгын алдааг тодорхойлох нь төвөгтэй байж болно.

**хувьсагч:**

**төлөвийн диаграмм:**
Хувьсагчид болон тэдгээрийн зааж буй утгуудыг харуулсан зураглал.

**түлхүүр үг:**
Програмын бүтцийг тодорхойлоход хэрэглэдэг тусгай үг.

**импортлох заавар:**
Модулийн файлыг уншиж, модулийн объект үүсгэдэг заавар.

**модуль:**
Функцийн тодорхойлолт болон заримдаа бусад зааврыг агуулсан Python код бүхий файл.

**цэгэн оператор:**
Модулийн нэрийн араас цэг, дараа нь функцийн нэрийг бичиж, тухайн модулийн функцэд хандахад ашигладаг `.` оператор.

**илэрхийллийн утгыг тооцоолох:**
Утгыг гаргахын тулд илэрхийлэл дэх үйлдлүүдийг гүйцэтгэх.

**заавар:**
Команд буюу үйлдлийг илэрхийлэх нэг эсвэл хэд хэдэн мөр код.

**гүйцэтгэх:**
Зааврыг ажиллуулж, заасан үйлдлийг хийх.

**аргумент:**
Функцийг дуудах үед түүнд дамжуулдаг утга. Аргумент бүрийг функцийн харгалзах параметрт олгоно.

**кодын тайлбар:**
Програмын талаар мэдээлэл өгөх боловч түүний гүйцэтгэлд нөлөөлөхгүй текст.

**ажиллах үеийн алдаа:**
Програм алдааны мэдэгдэл харуулаад зогсоход хүргэдэг алдаа.

**онцгой нөхцөл:**
Програм ажиллаж байх үед илэрсэн алдаа.

**утгын алдаа:**
Алдааны мэдэгдэл харуулахгүй боловч програм буруу үйлдэл хийхэд хүргэдэг алдаа.
