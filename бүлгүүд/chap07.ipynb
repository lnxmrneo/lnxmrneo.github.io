# Давталт ба хайлт

Хичнээн хэцүү болохыг ойлгохын тулд энэ бүлэгт бид дор хаяж нэг "e"-тэй англи үгсийн хэсгийг тооцоолох болно.

Үүний тулд бид `for` хэллэгийг ашиглан мөр дэх үсэг болон файл дахь үгсийг хооронд нь эргүүлж, "e" агуулсан үгсийн тоог тоолохын тулд давталтын хувьсагчдыг шинэчилнэ.
Бид `in` операторыг ашиглан үгэнд үсэг гарч байгаа эсэхийг шалгах бөгөөд та "шугаман хайлт" хэмээх програмчлалын загварыг сурах болно.

## Давталт ба тэмдэгт мөр

3-р бүлэгт бид тоонуудын дарааллыг харуулахын тулд `range` функцийг ашигладаг `for` давталт харсан.

```python
for i in range(3):
    print(i, end=' ')
```

Энэ хувилбар нь `end` түлхүүр үгийн аргументыг ашигладаг тул `print` функц нь шинэ мөрийн оронд тоо бүрийн ард хоосон зай тавьдаг.

Бид мөн `for` давталт ашиглан үсгүүдийг мөрөнд харуулах боломжтой.

```python
for letter in 'Gadsby':
    print(letter, end=' ')
```

Би хувьсагчийн нэрийг `i`-аас `letter` болгон өөрчилсөнийг анхаарна уу, энэ нь түүний хамаарах утгын талаар илүү их мэдээлэл өгдөг.
`for` давталтад тодорхойлсон хувьсагчийг **хувьсагч** гэж нэрлэдэг.

Одоо бид үгэнд байгаа үсгүүдийг хооронд нь эргүүлэх боломжтой болсон тул "e" үсэг байгаа эсэхийг шалгаж болно.

```python
for letter in "Gadsby":
    if letter == 'E' or letter == 'e':
        print('This word has an "e"')
```

Үргэлжлүүлэхээсээ өмнө тэр давталтыг функцэд багтааж үзье.

```python
def has_e():
    for letter in "Gadsby":
        if letter == 'E' or letter == 'e':
            print('This word has an "e"')
```

Хэрэв үгэнд "e" байвал `True`, бусад тохиолдолд `False` буцаах цэвэр функц болгоцгооё.

```python
def has_e():
    for letter in "Gadsby":
        if letter == 'E' or letter == 'e':
            return True
    return False
```

Үгийг параметр болгон авахын тулд бид үүнийг ерөнхийд нь хэлж болно.

```python
def has_e(word):
    for letter in word:
        if letter == 'E' or letter == 'e':
            return True
    return False
```

Одоо бид үүнийг дараах байдлаар шалгаж болно.

```python
has_e('Gadsby')
```

```python
has_e('Emma')
```

## Хайлт

`has_e`-ийн энэ хялбар хувилбар дээр тулгуурлан `uses_any` хэмээх үсгийн мөр болох хоёр дахь параметрийг авдаг илүү ерөнхий функцийг бичье.
Хэрэв үгэнд аль нэг үсгийг ашигласан бол `True`-г, өөрөөр хэлбэл `False`-г буцаана.

```python
def uses_any(word, letters):
    for letter in word.lower():
        if letter in letters.lower():
            return True
    return False
```

Үр дүн нь `True` болох жишээ энд байна.

```python
uses_any('banana', 'aeiou')
```

Мөн өөр газар нь `False`.

```python
uses_any('apple', 'xyz')
```

`uses_only` `word` болон `letters`-г жижиг үсгээр хөрвүүлдэг тул ямар ч хослолд ажилладаг.

```python
uses_any('Banana', 'AEIOU')
```

`uses_any`-ийн бүтэц нь `has_e`-тай төстэй.
Энэ нь `word` дахь үсгүүдийг нэг нэгээр нь шалгадаг.
Хэрэв `letters`-д харагдах нэгийг олвол тэр даруй `True`-г буцаана.
Хэрэв ямар ч ололгүйгээр давталтыг бүхэлд нь даввал `False`-г буцаана.

Энэ загварыг **шугаман хайлт** гэж нэрлэдэг.
Энэ бүлгийн төгсгөлд байгаа дасгалууд дээр та энэ загварыг ашигладаг илүү олон функцийг бичих болно.

## Doctest шалгалт

[Бүлэг 4](section_docstring)-д бид функцийг баримтжуулах, өөрөөр хэлбэл энэ нь юу хийдгийг тайлбарлахын тулд docstring ашигласан.
Мөн функцийг *турших*-ын тулд docstring ашиглах боломжтой.
Энд `uses_any`-ийн тест агуулсан баримт бичиг бүхий хувилбар байна.

```python
def uses_any(word, letters):
    """Checks if a word uses any of a list of letters.
    
    >>> uses_any('banana', 'aeiou')
    True
    >>> uses_any('apple', 'xyz')
    False
    """
    for letter in word.lower():
        if letter in letters.lower():
            return True
    return False
```

Туршилт бүр нь `>>>`-ээр эхэлдэг бөгөөд энэ нь зарим Python орчинд хэрэглэгч хаана код бичиж болохыг зааж өгөх сануулга болгон ашигладаг.
Doctest-д мөрийн дараа илэрхийлэл, ихэвчлэн функцийн дуудлага ирдэг.
Дараах мөрөнд функц зөв ажиллаж байвал илэрхийлэлд байх ёстой утгыг заана.

Эхний жишээнд `'banana'` нь `'a'` ашигладаг тул үр дүн нь `True` байх ёстой.
Хоёрдахь жишээнд `'apple'` нь `'xyz'`-ийн алийг нь ч ашигладаггүй тул үр дүн нь `False` байх ёстой.

Эдгээр туршилтыг явуулахын тулд бид `doctest` модулийг импортлон `run_docstring_examples` нэртэй функцийг ажиллуулах ёстой.
Энэ функцийг ашиглахад хялбар болгох үүднээс функцийн объектыг аргумент болгон авдаг дараах функцийг бичсэн.

```python
from doctest import run_docstring_examples

def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)
```

Бид `globals` болон `__name__`-ийн талаар хараахан сураагүй байна -- та тэдгээрийг үл тоомсорлож болно.
Одоо бид `uses_any`-г ингэж шалгаж болно.

```python
run_doctests(uses_any)
```

`run_doctests` нь баримт бичгийн мөр дэх илэрхийллүүдийг олж, үнэлдэг.
Хэрэв үр дүн нь хүлээгдэж буй утга байвал тест **тэнцэнэ**.
Тэгэхгүй бол **бүтэлгүйтнэ**.

Хэрэв бүх шалгалт тэнцсэн бол `run_doctests` гаралт байхгүй -- энэ тохиолдолд ямар ч мэдээ сайн мэдээ биш болно.
Туршилт амжилтгүй болоход юу болохыг харахын тулд `uses_any`-ийн буруу хувилбарыг эндээс үзнэ үү.

```python
def uses_any_incorrect(word, letters):
    """Checks if a word uses any of a list of letters.
    
    >>> uses_any_incorrect('banana', 'aeiou')
    True
    >>> uses_any_incorrect('apple', 'xyz')
    False
    """
    for letter in word.lower():
        if letter in letters.lower():
            return True
        else:
            return False     # INCORRECT!
```

Бид үүнийг туршиж үзэхэд юу болохыг энд харуулав.

```python
run_doctests(uses_any_incorrect)
```

Гаралт нь бүтэлгүйтсэн жишээ, функцийн гаргахаар хүлээгдэж буй утга, функцийн бодитой гаргасан утгыг агуулна.

Хэрэв та энэ шалгалт яагаад бүтэлгүйтсэнийг сайн мэдэхгүй байгаа бол дасгал болгон алдаа олж засварлах боломжтой.

## Нэр томьёоны тайлбар

**давталтын хувьсагч:**
`for` давталтын толгой мөрөнд тодорхойлсон хувьсагч.

**файлын объект:**
Нээлттэй файлыг төлөөлж, аль хэсгийг нь уншсан эсвэл бичсэнийг хянадаг объект.

**метод:**
Объекттой холбоотой, цэгэн оператор ашиглан дууддаг функц.

**шинэчлэх:**
Шинэ хувьсагч үүсгэхийн оронд байгаа хувьсагчид шинэ утга олгох.

**анхны утга олгох:**
Шинэ хувьсагч үүсгэж, түүнд утга олгох.

**нэмэгдүүлэх:**
Хувьсагчийн утгыг өсгөх.

**бууруулах:**
Хувьсагчийн утгыг багасгах.

**тоолуур:**

```python
def uses_any(word, letters):
    for letter in word.lower():
        if letter in letters.lower():
            return True
    return False
```

```python
def uses_any_incorrect(word, letters):
    for letter in word.lower():
        if letter in letters.lower():
            return True
        else:
            return False     # INCORRECT!
```
