## in оператор

`in` оператор нь dict төрөл дээр мөн ажилладаг; Энэ нь dict төрөлд ямар нэг зүйл *түлхүүр* болж байгаа эсэхийг хэлж өгнө.

```python
'one' in numbers
```

`in` оператор нь ямар нэг зүйл утга болж байгаа эсэхийг * шалгадаггүй*.

```python
1 in numbers
```

Ямар нэг зүйл dict төрөлд утга болж байгаа эсэхийг харахын тулд утгуудын дарааллыг буцаадаг `values` аргыг ашиглаж, `in` операторыг ашиглаж болно.

```python
1 in numbers.values()
```

Python dict төрөлд байгаа зүйлүүд нь **хэш хүснэгт**-д хадгалагддаг бөгөөд энэ нь гайхалтай шинж чанартай өгөгдлийг цэгцлэх арга юм: `in` оператор dict төрөлд хичнээн зүйл байсан ч ижил хэмжээний цаг зарцуулдаг.
Энэ нь гайхалтай үр дүнтэй алгоритмуудыг бичих боломжийг олгодог.

```python
download('https://raw.githubusercontent.com/AllenDowney/ThinkPython/v3/words.txt');
```

Үзүүлэн харуулахын тулд бид `stressed` болон `desserts` гэх мэт нэг нь нөгөөгийнхөө эсрэг утгатай үгийн хосыг олох хоёр алгоритмыг харьцуулах болно.
Бид үгийн жагсаалтыг уншиж эхэлнэ.

```python
word_list = open('words.txt').read().split()
len(word_list)
```

Өмнөх бүлгийн `reverse_word`-г энд оруулав.

```python
def reverse_word(word):
    return ''.join(reversed(word))
```

Дараах функц нь жагсаалтад байгаа үгсийн дундуур давтагдана.
Тус бүрийн хувьд энэ нь үсгүүдийг эргүүлж, дараа нь үгийн жагсаалтад урвуу үг байгаа эсэхийг шалгадаг.

```python
def too_slow():
    count = 0
    for word in word_list:
        if reverse_word(word) in word_list:
            count += 1
    return count
```

Энэ функцийг ажиллуулахад нэг минутаас илүү хугацаа шаардагдана.
Асуудал нь `in` оператор жагсаалтад байгаа үгсийг эхнээс нь эхлэн нэг нэгээр нь шалгадагт оршино.
Хэрэв тэр хайж буй зүйлээ олохгүй бол - энэ нь ихэвчлэн тохиолддог - эцсээ хүртэл хайх хэрэгтэй.

```python
# %time too_slow()
```

Мөн `in` оператор нь давталтын дотор байдаг тул үг бүрт нэг удаа ажилладаг.
Жагсаалтад 100,000 гаруй үг байгаа бөгөөд бид тус бүрээр нь 100,000 гаруй үг шалгадаг тул харьцуулсан нийт тоо нь үгийн квадрат буюу ойролцоогоор 13 тэрбум юм.

```python
len(word_list)**2
```

Бид dict төрлийн тусламжтайгаар энэ функцийг илүү хурдан болгож чадна.
Дараах давталт нь үгсийг түлхүүр болгон агуулсан dict төрлийг үүсгэдэг.

```python
word_dict = {}
for word in word_list:
    word_dict[word] = 1
```

`word_dict` дээрх утгууд нь бүгд `1` боловч тэдгээр нь юу ч байж болно, учир нь бид тэдгээрийг хэзээ ч хайхгүй -- бид зөвхөн түлхүүр байгаа эсэхийг шалгахын тулд энэ dict төрлийг ашиглах болно.

Одоо `word_list`-г `word_dict`-аар сольсон өмнөх функцийн хувилбар энд байна.

```python
def much_faster():
    count = 0
    for word in word_dict:
        if reverse_word(word) in word_dict:
            count += 1
    return count
```

Энэ функц нь секундын зууны нэгээс бага хугацаа шаардагдах тул өмнөх хувилбараас 10,000 дахин хурдан юм.

```python
%time much_faster()
```

Ерөнхийдөө жагсаалтын элементийг олоход шаардагдах хугацаа нь жагсаалтын урттай пропорциональ байна.
dict төрлийн түлхүүрийг олоход шаардагдах хугацаа нь зүйлийн тооноос үл хамааран бараг тогтмол байдаг.

## Тоолууруудын цуглуулга

Танд мөр өгөгдсөн бөгөөд үсэг бүр хэдэн удаа гарч байгааг тоолохыг хүсч байна гэж бодъё.
dict төрөл бол энэ ажилд тохиромжтой хэрэгсэл юм.
Бид хоосон dict төрлөөс эхэлнэ.

```python
counter = {}
```

Мөр дэх үсгүүдийг давталтаар бид `a` үсгийг анх удаа харлаа гэж бодъё.
Бид үүнийг dict төрөлд ингэж нэмж болно.

```python
counter['a'] = 1
```

`1` утга нь бид захидлыг нэг удаа харсан гэдгийг харуулж байна.
Дараа нь бид ижил үсгийг дахин харвал тоологчийг ингэж өсгөж болно.

```python
counter['a'] += 1
```

Одоо `a`-тай холбоотой утга нь `2` байна, учир нь бид энэ үсгийг хоёр удаа харсан.

```python
counter
```

Дараах функц нь үсэг бүрийг мөрөнд хэдэн удаа гарч ирэхийг тоолохын тулд эдгээр функцүүдийг ашигладаг.

```python
def value_counts(string):
    counter = {}
    for letter in string:
        if letter not in counter:
            counter[letter] = 1
        else:
            counter[letter] += 1
    return counter
```

dict төрөлд `letter` байхгүй бол давталт болгонд бид `letter` түлхүүр, `1` утгатай шинэ зүйл үүсгэдэг.
Хэрэв `letter` dict төрөлд байгаа бол бид `letter`-тай холбоотой утгыг нэмэгдүүлнэ.

Энд нэг жишээ байна.

```python
counter = value_counts('brontosaurus')
counter
```

`counter`-д байгаа зүйлүүд нь `b` үсэг нэг удаа, `r` хоёр удаа гарч ирдэг гэх мэтийг харуулж байна.

## Жагсаалтад утга цуглуулах

Програмчлалын олон даалгаврын хувьд нэг жагсаалт эсвэл dict төрлийг бүтээхдээ өөр нэг жагсаалт үүсгэх нь ашигтай байдаг.
Жишээ болгон бид `word_dict`-д байгаа үгсийг давтаж, палиндромуудын жагсаалтыг гаргана, өөрөөр хэлбэл "үд", "эргэлт" гэх мэт урагш болон хойшоо адилхан бичигдсэн үгс.

Өмнөх бүлэгт дасгалуудын нэг нь үг палиндром мөн эсэхийг шалгах функц бичихийг хүссэн.
`reverse_word` ашигладаг шийдэл энд байна.

```python
def is_palindrome(word):
    """Check if a word is a palindrome."""
    return reverse_word(word) == word
```

Хэрэв бид `word_dict` дээрх үгсийг давталт хийвэл үүнтэй төстэй палиндромуудын тоог тоолж болно.

```python
count = 0

for word in word_dict:
    if is_palindrome(word):
        count +=1
        
count
```

Одоогийн байдлаар энэ загвар танил болсон.

* Давталтын өмнө `count`-г `0` болгож эхлүүлнэ.

* Давталтын дотор `word` нь палиндром бол бид `count`-г нэмэгдүүлнэ.

* Давталт дуусах үед `count` нийт палиндромуудын тоог агуулна.

Бид ижил төстэй загварыг ашиглан палиндромуудын жагсаалтыг гаргаж болно.

```python
palindromes = []

for word in word_dict:
    if is_palindrome(word):
        palindromes.append(word)

palindromes[:10]
```

Энэ нь дараах байдалтай байна.

* Давталтын өмнө `palindromes` хоосон жагсаалтаар эхлүүлнэ.

* Давталтнд `word` палиндром бол бид үүнийг `palindromes`-ийн төгсгөлд хавсаргана.

* Давталт дуусахад `palindromes` нь палиндромуудын жагсаалт болно.

Энэ давталтад `palindromes`-г **аккумлятор** болгон ашигладаг бөгөөд энэ нь тооцооллын явцад өгөгдөл цуглуулах буюу хуримтлуулах хувьсагч юм.

Одоо бид зөвхөн долоон ба түүнээс дээш үсэгтэй палиндром сонгохыг хүсч байна гэж бодъё.
Бид `palindromes`-г давтаж, зөвхөн урт палиндром агуулсан шинэ жагсаалт гаргах боломжтой.

```python
long_palindromes = []

for word in palindromes:
    if len(word) >= 7:
        long_palindromes.append(word)
        
long_palindromes
```

Жагсаалтыг ингэж эргүүлж, зарим элементийг сонгоод, заримыг нь орхигдуулахыг **шүүлт** гэж нэрлэдэг.

## Тооцоолсон үр дүнг хадгалах

Хэрэв та `fibonacci` функцийг [6-р бүлэг](Fibonacci_section)-аас ажиллуулсан бол таны оруулсан аргумент хэдий чинээ том байх тусам функц удаан ажиллах болно гэдгийг анзаарсан байх.

```python
def fibonacci(n):
    if n == 0:
        return 0
    
    if n == 1:
        return 1

    return fibonacci(n-1) + fibonacci(n-2)
```

Цаашилбал, ажиллах хугацаа хурдан нэмэгддэг.

Дуудлагын график нь фрейм бүрийг дуудаж буй функцүүдийн фреймүүдтэй холбосон шугамууд бүхий функцийн хүрээг харуулж байна.
Графикийн дээд талд, `n=4`-тай `fibonacci` нь ` n=3` болон `n=2`-тай `fibonacci`-г дууддаг.
Хариуд нь `n=3`-тай `fibonacci` нь `n=2` болон `n=1`-тай `fibonacci`-г дууддаг. гэх мэт.

`fibonacci(0)` болон `fibonacci(1)` хэдэн удаа дуудагдсаныг тоол. 
Энэ нь асуудлыг шийдэх үр ашиггүй шийдэл бөгөөд аргумент улам бүр дорддог.

Нэг шийдэл бол аль хэдийн тооцоолсон утгуудыг dict төрөлд хадгалах замаар бүртгэх явдал юм.
Урьд нь тооцоолсон, хожим ашиглахаар хадгалагдсан утгыг **memo** гэж нэрлэдэг.
Энд `fibonacci`-ийн "чээжлэгдсэн" хувилбар байна:

```python
known = {0:0, 1:1}

def fibonacci_memo(n):
    if n in known:
        return known[n]

    res = fibonacci_memo(n-1) + fibonacci_memo(n-2)
    known[n] = res
    return res
```

`known` бол бидний аль хэдийн мэддэг Фибоначчийн тоонуудыг бүртгэдэг dict төрөл юм.
Энэ нь хоёр зүйлээс эхэлдэг: `0` газрын зураг нь `0`, `1` нь `1`.

`fibonacci_memo` дуудагдах бүрт `known`-г шалгадаг.
Хэрэв үр дүн нь аль хэдийн байгаа бол тэр даруй буцаж болно.
Үгүй бол энэ нь шинэ утгыг тооцоолж, dict төрөлд нэмж, буцааж өгөх ёстой.

Хоёр функцийг харьцуулж үзвэл `fibonacci(40)` 30 секунд орчим ажиллана.
`fibonacci_memo(40)` нь 30 орчим микросекунд шаарддаг тул сая дахин хурдан болдог.
Энэ бүлгийн тэмдэглэлийн дэвтэрт та эдгээр хэмжилтүүд хаанаас гарсныг харах болно.

## Нэр томьёоны тайлбар

**dict төрөл:**
Түлхүүр–утгын хосуудыг агуулсан объект. Хос бүрийг dict төрлийн элемент гэж бас нэрлэдэг.

**элемент:**
dict төрөл дахь түлхүүр–утгын хосын өөр нэр.

**түлхүүр:**
dict төрөл дахь түлхүүр–утгын хосын эхний хэсэг болох объект.

**утга:**
dict төрөл дахь түлхүүр–утгын хосын хоёр дахь хэсэг болох объект. Энэ нь өмнө хэрэглэсэн «утга» гэдэг нэр томьёоноос илүү нарийн утгатай.

**харгалзуулалт:**
Нэг олонлогийн элемент бүрийг нөгөө олонлогийн элементтэй холбох харьцаа.

**хэш хүснэгт:**
Түлхүүрээр нь утгыг хурдан хайж олох боломжтойгоор зохион байгуулсан түлхүүр–утгын хосуудын цуглуулга.

**хэшлэх боломжтой:**
Бүхэл тоо, хөвөгч таслалтай тоо, тэмдэгт мөр зэрэг өөрчлөгддөггүй төрлүүд хэшлэгдэх боломжтой. Жагсаалт, dict төрөл зэрэг өөрчлөгддөг төрлүүд хэшлэгдэхгүй.

**хэш функц:**
Объект хүлээн авч, хэш хүснэгтэд түлхүүрийн байрлалыг тогтооход ашиглах бүхэл тоо тооцоолдог функц.

**хуримтлуулагч:**
Давталтын үр дүнг нэмж эсвэл цуглуулж хадгалахад ашигладаг хувьсагч.

**шүүх:**
Дарааллын элементүүдээр давтахдаа заримыг нь сонгох эсвэл хасах.

**дуудлагын граф:**
Програм ажиллахад үүссэн бүх фреймийг, дуудагчаас дуудагдсан функц рүү чиглэсэн сумтайгаар харуулсан диаграмм.

**хадгалсан үр дүн:**
Дараа нь шаардлагагүй дахин тооцоолол хийхээс зайлсхийхийн тулд хадгалсан тооцооллын утга.
