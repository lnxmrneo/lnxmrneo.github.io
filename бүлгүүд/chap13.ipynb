# Файл ба өгөгдлийн сан

Бидний харж байгаа ихэнх программууд нь богино хугацаанд ажиллаж, бүтээгдэхүүн гаргадаг гэдэг утгаараа **түр зуурын** байдаг ч дуусахад өгөгдөл нь алга болдог.
Түр зуурын программыг ажиллуулах бүрт энэ нь цэвэр хуудаснаас эхэлдэг.

Бусад програмууд нь **байнгын**: тэд удаан хугацаанд (эсвэл байнга) ажилладаг; тэд дор хаяж зарим өгөгдлөө урт хугацааны хадгалалтад хадгалдаг; мөн хэрэв тэд унтрааж, дахин эхлүүлэх юм бол тэд орхисон газраасаа үргэлжлүүлнэ.

Програмууд өөрсдийн мэдээллээ хадгалах энгийн арга бол текст файлуудыг уншиж бичих явдал юм.
Илүү уян хатан хувилбар бол өгөгдлийн санд өгөгдлийг хадгалах явдал юм.
Өгөгдлийн сан нь текст файлуудаас илүү уншиж, бичих боломжтой тусгайлсан файлууд бөгөөд тэдгээр нь нэмэлт боломжуудыг олгодог.

Энэ бүлэгт бид текст файлууд болон өгөгдлийн санг уншиж бичих программуудыг бичих бөгөөд дасгалын хувьд та зургийн цуглуулгаас давхардсан зургийг хайж олох программ бичих болно.

## f-тэмдэгт мөр

Програмд өгөгдөл хадгалах нэг арга бол үүнийг текст файлд бичих явдал юм.
Жишээлбэл, та тэмээ сонирхогч, ажиглалтын хугацаанд үзсэн тэмээнийхээ тоог тэмдэглэмээр байна гэж бодъё.
Нэг жил хагасын дараа та `23` тэмээ олж харлаа гэж бодъё.
Таны тэмээ толбоны номонд гарсан мэдээлэл иймэрхүү харагдаж магадгүй.

```python
num_years = 1.5
num_camels = 23
```

Энэ өгөгдлийг файлд бичихийн тулд та бидний [Бүлэг 8](бичих_файлууд) хэсэгт үзсэн `write` аргыг ашиглаж болно.
`write` аргумент нь мөр байх ёстой тул хэрэв бид файлд бусад утгуудыг оруулахыг хүсвэл тэдгээрийг мөр болгон хөрвүүлэх хэрэгтэй.
Үүнийг хийх хамгийн хялбар арга бол суурилуулсан `str` функц юм.

Энэ нь дараах байдалтай байна.

```python
writer = open('camel-spotting-book.txt', 'w')
writer.write(str(num_years))
writer.write(str(num_camels))
writer.close()
```

Энэ нь ажилладаг, гэхдээ та үүнийг тодорхой оруулаагүй л бол `write` нь хоосон зай эсвэл шинэ мөр нэмэхгүй.
Хэрэв бид файлыг дахин уншвал хоёр тоо хамт ажиллаж байгааг харж болно.

```python
open('camel-spotting-book.txt').read()
```

Хамгийн багадаа бид тоонуудын хооронд хоосон зай нэмэх хэрэгтэй.
Үүнийг хийж байх хооронд тайлбар бичвэр нэмье.

Мөр болон бусад утгуудын хослолыг бичихийн тулд бид нээх хашилтын өмнө `f` үсэгтэй, буржгар хаалтанд нэг буюу хэд хэдэн Python илэрхийлэл агуулсан мөр болох **f-string** ашиглаж болно.
Дараах f-мөр нь хувьсах нэр болох нэг илэрхийлэлийг агуулна.

```python
f'I have spotted {num_camels} camels'
```

Үр дүн нь илэрхийлэлийг үнэлж, үр дүнгээр сольсон мөр юм.
Нэгээс олон илэрхийлэл байж болно.

```python
f'In {num_years} years I have spotted {num_camels} camels'
```

Мөн илэрхийлэл нь оператор болон функцийн дуудлагыг агуулж болно.

```python
line = f'In {round(num_years * 12)} months I have spotted {num_camels} camels'
line
```

Тиймээс бид өгөгдлийг ийм текст файлд бичиж болно.

```python
writer = open('camel-spotting-book.txt', 'w')
writer.write(f'Years of observation: {num_years}\n')
writer.write(f'Camels spotted: {num_camels}\n')
writer.close()
```

Хоёр f-мөр нь `\n` дарааллаар төгсдөг бөгөөд энэ нь шинэ мөрийн тэмдэгт нэмнэ.

Бид файлыг дараах байдлаар буцааж уншиж болно.

```python
data = open('camel-spotting-book.txt').read()
print(data)
```

F тэмдэгтэнд буржгар хаалтанд байгаа илэрхийлэл нь тэмдэгт мөр болж хувирдаг тул та жагсаалт, dict төрөл болон бусад төрлийг оруулах боломжтой.

```python
t = [1, 2, 3]
d = {'one': 1}
f'Here is a list {t} and a dictionary {d}'
```

## YAML

Программууд файл уншиж, бичих нэг шалтгаан нь **тохируулгын өгөгдөл** хадгалах явдал бөгөөд энэ нь программ юу хийх, хэрхэн хийхийг зааж өгсөн мэдээлэл юм.

Жишээ нь, давхардсан зураг хайдаг программ дээр бид `config` нэртэй dict төрөлтэй байж болох бөгөөд энэ нь хайх директорийн нэр, үр дүнг хадгалах өөр директорийн нэр, зургийн файлуудыг тодорхойлоход ашиглах файлын өргөтгөлүүдийн жагсаалтыг агуулсан байж болно.

Энэ нь дараах байдалтай байж магадгүй юм.

```python
config = {
    'photo_dir': 'photos',
    'data_dir': 'photo_info',
    'extensions': ['jpg', 'jpeg'],
}
```

Энэ өгөгдлийг текст файлд бичихийн тулд бид өмнөх хэсгийн адил f-мөрүүдийг ашиглаж болно. Гэхдээ яг ийм зүйлд зориулагдсан `yaml` модулийг ашиглах нь илүү хялбар байдаг.

`yaml` модуль нь YAML файлуудтай ажиллах функцээр хангадаг бөгөөд эдгээр нь текст файлууд нь хүмүүст *болон* программуудад уншиж бичихэд хялбар байхаар форматлагдсан байдаг.

`config` dict төрлийг YAML файлд бичихийн тулд `dump` функцийг ашигладаг жишээ энд байна.

```python
# this cell installs the pyyaml package, which provides the yaml module

try:
    import yaml
except ImportError:
    !pip install pyyaml
```

```python
import yaml

config_filename = 'config.yaml'
writer = open(config_filename, 'w')
yaml.dump(config, writer)
writer.close()
```

Хэрэв бид файлын агуулгыг дахин уншвал YAML формат ямар байгааг харж болно.

```python
readback = open(config_filename).read()
print(readback)
```

Одоо бид YAML файлыг уншихын тулд `safe_load` ашиглаж болно.

```python
reader = open(config_filename)
config_readback = yaml.safe_load(reader)
config_readback
```

Үүний үр дүнд эхтэй ижил мэдээллийг агуулсан шинэ dict төрөл гарч ирэв, гэхдээ энэ нь ижил dict төрөл биш юм.

```python
config is config_readback
```

dict төрөл шиг объектыг мөр болгон хөрвүүлэхийг **сериалчлал** гэнэ.
Мөрийг буцааж объект руу хөрвүүлэхийг **deserization** гэж нэрлэдэг.
Хэрэв та объектыг цуваа болгоод дараа нь цуваа арилгавал үр дүн нь анхныхтай тэнцүү байх ёстой.

Энэ жишээн дээр түлхүүр нь зургийн файлд хүрэх зам бөгөөд утга нь дүрсийг дүрсэлсэн мөр юм.

```python
value = db[key]
value
```

## Файлуудын агуулга ижил эсэхийг шалгах

Одоо энэ бүлгийн зорилго руу буцаж орцгооё: ижил өгөгдөл агуулсан өөр өөр файлуудыг хайх.
Шалгах нэг арга бол хоёр файлын агуулгыг уншиж, харьцуулах явдал юм.

Хэрэв файлд зураг байгаа бол бид тэдгээрийг `rb` горимоор нээх ёстой бөгөөд `r` нь агуулгыг уншихыг хүсэж байгаа бол `b` нь **хоёртын горим** гэсэн үг юм.
Хоёртын горимд агуулгыг текст гэж ойлгодоггүй бөгөөд тэдгээрийг байтуудын дараалал гэж үздэг.

Зургийн файлыг нээж уншдаг жишээ энд байна.

```python
path1 = 'photos/jan-2023/photo1.jpg'
data1 = open(path1, "rb").read()
type(data1)
```

`read`-ийн үр дүн нь `bytes` объект бөгөөд нэрнээс нь харахад энэ нь байтуудын дарааллыг агуулдаг.

Ерөнхийдөө зургийн файлын агуулга нь хүн унших боломжгүй.
Гэхдээ бид хоёр дахь файлын агуулгыг уншвал `==` операторыг ашиглан харьцуулах боломжтой.

```python
path2 = 'photos/jan-2023/photo2.jpg'
data2 = open(path2, 'rb').read()
data1 == data2
```

Эдгээр хоёр файл нь тэнцүү биш юм.

Одоог хүртэл байгаа зүйлээ функцэд багтааж үзье.

```python
def same_contents(path1, path2):
    data1 = open(path1, "rb").read()
    data2 = open(path2, 'rb').read()
    return data1 == data2
```

Хэрэв бидэнд зөвхөн хоёр файл байгаа бол энэ функц нь сайн сонголт юм.
Гэхдээ бидэнд олон тооны файл байгаа бөгөөд тэдгээрийн аль нэг нь ижил өгөгдөл агуулсан эсэхийг мэдэхийг хүсч байна гэж бодъё.
Хос файл бүрийг харьцуулах нь үр ашиггүй байх болно.

Альтернатив хувилбар бол файлын агуулгыг авч, ихэвчлэн том бүхэл тоо болох **digest**-ийг тооцоолох **хэш функц** ашиглах явдал юм.
Хэрэв хоёр файл ижил өгөгдлийг агуулж байвал тэдгээр нь ижил мэдээлэлтэй байх болно.
Хэрэв хоёр файл өөр байвал *бараг үргэлж* өөр өөр мэдээлэлтэй байх болно.

`hashlib` модуль нь хэд хэдэн хэш функцээр хангадаг бөгөөд бидний ашиглахыг `md5` гэж нэрлэдэг.
Бид `HASH` объект үүсгэхийн тулд `hashlib.md5` ашиглан эхэлнэ.

```python
import hashlib

md5_hash = hashlib.md5()
type(md5_hash)
```

`HASH` объект нь файлын агуулгыг аргумент болгон авдаг `update` функцээр хангадаг.

```python
md5_hash.update(data1)
```

Одоо бид `hexdigest`-г ашиглан 16-р суурь дахь бүхэл тоог төлөөлсөн арван арвантэгт оронтой тоонуудын мөр болгон тоймыг авах боломжтой.

```python
digest = md5_hash.hexdigest()
digest
```

Дараах функц нь эдгээр алхмуудыг багтаасан болно.

```python
def md5_digest(filename):
    data = open(filename, "rb").read()
    md5_hash = hashlib.md5()
    md5_hash.update(data)
    digest = md5_hash.hexdigest()
    return digest
```

Хэрэв бид өөр файлын агуулгыг хэш болговол бид өөр мэдээлэл авах болно.

```python
filename2 = 'photos/feb-2023/photo2.jpg'
md5_digest(filename2)
```

Одоо бидэнд ижил төстэй файлуудыг олоход хэрэгтэй бараг бүх зүйл байна.
Сүүлийн алхам бол лавлахаас хайж, бүх зургийн файлуудыг олох явдал юм.

## Хавтаснуудаар дамжин үзэх

Дараах функц нь бидний хайхыг хүссэн лавлахыг аргумент болгон авдаг.
Энэ нь `listdir`-г ашиглан лавлахын агуулгыг давтдаг.
Энэ нь файлыг олох үед бүрэн замыг хэвлэдэг.
Энэ нь лавлахыг олох үед дэд директороос хайхын тулд өөрийгөө рекурсив байдлаар дууддаг.

```python
def walk(dirname):
    for name in os.listdir(dirname):
        path = os.path.join(dirname, name)

        if os.path.isfile(path):
            print(path)
        elif os.path.isdir(path):
            walk(path)
```

Бид үүнийг дараах байдлаар ашиглаж болно.

```python
walk('photos')
```

Үр дүнгийн дараалал нь үйлдлийн системийн дэлгэрэнгүй мэдээлэлээс хамаарна.

Энэ бол функцийн объектыг хоёр дахь параметр болгон авдаг `walk`-ийн илүү ерөнхий хувилбар юм.
Олсон файлынхаа замыг хэвлэхийн оронд энэ функцийг дуудаж, замыг параметр болгон дамжуулдаг.

```python
def walk(dirname, visit_func):
    for name in os.listdir(dirname):
        path = os.path.join(dirname, name)

        if os.path.isfile(path):
            visit_func(path)
        else:
            walk(path, visit_func)
```

`print`-г аргумент болгон дамжуулдаг жишээ энд байна, тиймээс `walk` `visit_func` руу залгахдаа олсон файлынхаа замыг хэвлэдэг.

```python
walk('photos', print)
```

Уг параметрийг `visit_func` гэж нэрлэсэн учир нь бид лавлахаар "алхаж" байхдаа файл бүрээр "зочилдог" гэсэн санааг илэрхийлдэг.

**цуваанаас сэргээх:**
