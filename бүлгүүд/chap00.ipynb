# Өмнөх үг



## Энэ ном хэнд зориулагдсан бэ?

Хэрэв та програмчлал сурч эхлэхийг хүсэж байгаа бол зөв газраа иржээ. Python бол анхлан суралцагчдад хамгийн тохиромжтой програмчлалын хэлнүүдийн нэг бөгөөд өнөө үед хөдөлмөрийн зах зээлд 
хамгийн их эрэлттэй ур чадваруудын нэг юм.

Түүнчлэн та програмчлал сурахад хамгийн тохиромжтой цаг үед ирсэн байна. Учир нь өнөөдөр програмчлал сурах нь урьд өмнөхөөсөө илүү хялбар болсон гэж хэлж болно. ChatGPT зэрэг хиймэл оюун ухаанд суурилсан виртуал туслахуудын ачаар та програмчлалыг заавал ганцаараа сурах шаардлагагүй болсон. Энэхүү номын туршид би эдгээр хэрэгслийг ашиглан суралцах үйл явцаа хэрхэн хурдасгаж, илүү үр дүнтэй болгох талаар зөвлөмж өгөх болно.

Энэ ном нь үндсэндээ өмнө нь огт програмчилж байгаагүй анхлан суралцагчид болон өөр програмчлалын хэл дээр тодорхой хэмжээний туршлагатай хүмүүст зориулагдсан. Харин та Python хэл дээр хангалттай туршлагатай бол номын эхний хэдэн бүлгийн агуулга хэт энгийн, удаан санагдаж магадгүй.

Програмчлал сурахад тулгардаг нэгэн онцлог бэрхшээл бол та нэгэн зэрэг хоёр өөр хэлийг эзэмших шаардлагатай болдог явдал юм. Үүний нэг нь програмчлалын хэл өөрөө, харин нөгөө нь програмын талаар тайлбарлах, хэлэлцэхэд ашигладаг мэргэжлийн нэр томьёо, ойлголтуудын хэл юм.

Хэрэв та зөвхөн програмчлалын хэлний бичиглэл, дүрмийг сурсан бол алдааны мэдэгдлийг ойлгож тайлбарлах, техникийн баримт бичиг унших, бусад программисттай мэргэжлийн түвшинд харилцах, эсвэл хиймэл оюун ухаанд суурилсан виртуал туслахуудыг үр дүнтэй ашиглах үед хүндрэлтэй тулгарч болзошгүй.

Тиймээс хэрэв та өмнө нь тодорхой хэмжээнд програмчилж байсан ч програмчлалын тухай ойлголт, нэр томьёо, мэргэжлийн хэллэгийг бүрэн эзэмшээгүй бол энэхүү ном танд хэрэг болно гэдэгт найдаж байна.


## Номын зорилго

Энэхүү номыг бичихдээ би мэргэжлийн нэр томьёо, ойлголтуудыг аль болох оновчтой, ойлгомжтой хэрэглэхийг хичээсэн. Шинэ нэр томьёо анх дурдагдах бүрд түүний утга, тодорхойлолтыг тайлбарласан. Мөн бүлэг бүрийн төгсгөлд тухайн бүлэгт шинээр танилцуулсан нэр томьёо, ойлголтуудыг нэгтгэн тайлбарласан нэр томьёоны тайлбар толь оруулсан.

Түүнчлэн номын агуулгыг аль болох товч, тодорхой, ойлгоход хялбар байдлаар боловсруулахыг зорьсон. Учир нь номыг уншиж ойлгоход бага оюуны ачаалал шаардагдах тусам програмчлалын бодлого бодох, код бичих, шинэ мэдлэг эзэмшихэд зарцуулах оюуны нөөц тань төдий чинээ нэмэгдэх болно.

Гэвч зөвхөн ном уншсанаар програмчилж сурна гэдэг боломжгүй. Програмчлалыг эзэмшихийн тулд заавал дадлага хийж, код бичиж турших шаардлагатай. Иймээс энэхүү номын бүлэг бүрийн төгсгөлд сурсан мэдлэгээ бататгах, практикт хэрэглэх боломжийг олгох дасгал, даалгавруудыг багтаасан.

Хэрэв та номыг анхааралтай уншиж, дасгал даалгавруудыг тогтмол гүйцэтгэвэл програмчлалын мэдлэг, ур чадвараа алхам алхмаар ахиулах болно.

Гэхдээ нэг зүйлийг урьдчилан анхааруулъя. Програмчлал сурах нь тийм ч амар зүйл биш. Туршлагатай программистууд хүртэл заримдаа хүндрэлтэй асуудалтай тулгарч, шийдлийг нь олох гэж багагүй бухимдах үе байдаг.

Иймээс энэхүү номын туршид би зөв ажилладаг програм бичих, програмын алдааг илрүүлэх, шалтгааныг нь тодорхойлох, засварлахад туслах арга барил, стратегиудыг санал болгох болно.

## Номыг судлах дараалал

Энэхүү номын бүлэг бүр өмнөх бүлгүүдэд эзэмшсэн мэдлэг, ойлголтод тулгуурлан боловсруулагдсан. Иймээс бүлгүүдийг өгөгдсөн дарааллаар нь судалж, дараагийн бүлэгт шилжихээсээ өмнө тухайн бүлгийн дасгал, даалгавруудыг гүйцэтгэж, сурсан мэдлэгээ бататгахыг зөвлөж байна.
Эхний зургаан бүлэгт арифметик үйлдэл, нөхцөлт илэрхийлэл, давталт зэрэг програмчлалын үндсэн элементүүдийг танилцуулна. Мөн програмчлалын хамгийн чухал ойлголтуудын нэг болох функц (function) болон функцийг үр дүнтэй ашиглах хүчирхэг арга болох рекурс (recursion)-ийн талаар авч үзнэ.

7 болон 8 дугаар бүлэгт үсэг, үг, өгүүлбэр зэрэг текстэн мэдээллийг илэрхийлэхэд ашигладаг тэмдэгт мөр (string) өгөгдлийн төрөл болон тэмдэгт мөртэй ажиллах алгоритмуудыг судална.

9–12 дугаар бүлэгт Python хэлний өгөгдлийн үндсэн бүтцүүд болох жагсаалт (list), толь (dictionary), кортеж (tuple)-ийг танилцуулна. Эдгээр нь үр ашигтай, оновчтой програм боловсруулахад өргөн ашиглагддаг чухал хэрэгслүүд юм.

Тухайлбал, 12 дугаар бүлэгт текстэн мэдээлэлд дүн шинжилгээ хийх болон санамсаргүй байдлаар шинэ текст үүсгэх алгоритмуудыг авч үзнэ. Эдгээртэй төстэй алгоритмын зарчмууд нь том хэлний загваруудын (Large Language Models — LLMs) үндэс суурийн нэг хэсэг болдог. Иймээс энэ бүлгийг судалснаар ChatGPT зэрэг хиймэл оюун ухаанд суурилсан хэрэгслүүд хэрхэн ажилладаг талаар анхан шатны ойлголттой болно.

13 дугаар бүлэгт өгөгдлийг урт хугацаанд хадгалах аргууд, тухайлбал файл (file) болон өгөгдлийн сан (database)-тай ажиллах үндсэн ойлголтуудыг судална. Энэ бүлгийн практик дасгалын хүрээнд файлын системээс хайлт хийж, давхардсан файлуудыг илрүүлэх програм боловсруулах боломжтой.

14–17 дугаар бүлэгт програм болон түүнд ашиглагдах өгөгдлийг зохион байгуулах арга барил болох объект хандалтат програмчлал (Object-Oriented Programming — OOP)-ыг танилцуулна.

Python хэлний олон сан (library) объект хандалтат програмчлалын зарчимд тулгуурлан бүтээгдсэн байдаг. Иймээс эдгээр бүлгийг судалснаар Python сангуудын бүтэц, зохион байгуулалтыг ойлгохоос гадна өөрийн гэсэн объектуудыг тодорхойлж, ашиглах чадвартай болно.
Энэхүү номын зорилго нь Python програмчлалын хэлний бүх боломж, онцлогийг бүрэн хамрахад оршихгүй. Харин хамгийн цөөн үндсэн ойлголтыг эзэмшсэнээр аль болох өргөн хүрээний програмчлалын асуудлыг шийдвэрлэх чадвар олгох агуулгад голлон анхаарсан.

Гэсэн хэдий ч Python хэл нь түгээмэл тохиолддог програмчлалын асуудлуудыг үр ашигтай шийдвэрлэхэд зориулагдсан олон боломж, хэрэгсэлтэй. 18 дугаар бүлэгт эдгээр боломжийн заримыг танилцуулна.
Эцэст нь, 19 дүгээр бүлэгт зохиогчийн зүгээс програмчлалын талаарх эцсийн бодол, дүгнэлт болон цаашид програмчлалын мэдлэг, ур чадвараа тасралтгүй хөгжүүлэхэд хэрэг болох зөвлөмжүүдийг хүргэнэ.


## Энэхүү номонд ямар шинэчлэлүүд орсон бэ? 

Энэхүү номонд орсон хамгийн том өөрчлөлтүүд нь Jupyter Notebook болон хиймэл оюун ухаанд суурилсан виртуал туслахууд (virtual assistants) хэмээх хоёр шинэ технологийн хөгжилтэй холбоотой юм.
Энэ номын бүлэг бүрийг Jupyter Notebook хэлбэрээр боловсруулсан. Jupyter Notebook нь энгийн тайлбар текст болон програмын кодыг нэг баримт бичигт нэгтгэн агуулдаг интерактив орчин юм.

Зохиогчийн хувьд энэ нь програмын код бичих, турших, мөн кодыг тайлбар тексттэй нь уялдуулан боловсруулах ажлыг илүү хялбар болгосон. Харин суралцагчдын хувьд програмын кодыг ажиллуулах, өөрчлөх, турших, дасгал даалгавар гүйцэтгэх зэрэг үйлдлийг нэг орчинд хийх боломжийг бүрдүүлж байна. Jupyter Notebook орчинтой хэрхэн ажиллах зааврыг номын нэгдүгээр бүлэгт багтаасан.

Өөр нэг томоохон шинэчлэл нь ChatGPT зэрэг хиймэл оюун ухаанд суурилсан виртуал туслахуудыг ашиглан суралцах үйл явцаа хэрхэн хурдасгах талаар зөвлөмжүүдийг нэмж оруулсан явдал юм.
Өнөөдөр эдгээр хэрэгслүүд нь програм хангамжийн инженерчлэлийн салбарт өргөн хэрэглэгддэг стандарт хэрэгслийн нэг болжээ. Миний бодлоор тэдгээр нь зөвхөн програмчлал сурах үйл явцыг төдийгүй бусад олон салбарын мэдлэг, ур чадвар эзэмших арга барилыг үндсээр нь өөрчлөх чухал хэрэгсэл болох юм.

Хиймэл оюун ухаанд суурилсан виртуал туслахууд хөгжсөн өнөө үед автоматжуулсан тестлэлт (automated testing) улам бүр чухал болж байна.

Иймээс энэхүү хэвлэлд Python хэлний тестлэлтэд өргөн ашиглагддаг doctest болон unittest хэрэгслүүдийг танилцуулж, тэдгээрийг практикт ашиглах чадвар эзэмшүүлэх хэд хэдэн дасгал, даалгаврыг нэмж оруулсан.

Jupyter Notebook орчинд шилжсэнээр дасгал, даалгавруудыг илүү системтэйгээр боловсруулж, турших боломж бүрдсэн. Үүний үр дүнд суралцагчдын сонирхлыг илүү татахуйц, сургалтын үр дүнг дээшлүүлэхэд чиглэсэн дасгалуудын оновчтой дарааллыг боловсруулах боломжтой болсон.

Түүнчлэн тэмдэгт мөр (string)-ийн талаарх агуулгыг өргөжүүлж, регуляр илэрхийлэл (regular expression) буюу тэмдэгт мөрийн хэв загварыг тодорхойлох, хайх, боловсруулах аргыг нэмж оруулсан.
Номын зарим бүлэгт Turtle Graphics буюу яст мэлхийн график ашигладаг. Python хэлний стандарт turtle модулийг ашиглаж байсан боловч харамсалтай нь уг модуль Jupyter Notebook орчинд шууд ажилладаггүй.

Иймээс Jupyter Notebook орчинд модулийг хэрэглэхэд илүү хялбар байх шинэ turtle модулиар сольсон.

Энэхүү ном нь танд таалагдана гэдэгт найдаж байна!


## Програмчлал сурч эхлэх

Python болон бусад ихэнх програмчлалын хэл дээр програм бичих, ажиллуулахад ашиглаж болох олон төрлийн хэрэгсэл байдаг. Эдгээр хэрэгслийг хөгжүүлэлтийн нэгдсэн орчин (Integrated Development Environment — IDE) гэж нэрлэдэг.

Ерөнхийдөө хөгжүүлэлтийн нэгдсэн орчныг дараах хоёр төрөлд ангилж болно.

Кодын файлд суурилсан хөгжүүлэлтийн орчин. Энэ төрлийн орчин нь програмын код агуулсан файлуудтай ажиллахад зориулагдсан бөгөөд код бичих, засварлах, ажиллуулах хэрэгслүүдээр хангадаг.

Notebook-д суурилсан хөгжүүлэлтийн орчин. Энэ төрлийн орчин нь тайлбар текст болон програмын кодыг нэг баримт бичигт нэгтгэн агуулдаг Notebook-той ажиллахад голлон чиглэдэг.

Програмчлалыг анхлан суралцаж буй хүмүүст би Jupyter зэрэг Notebook-д суурилсан хөгжүүлэлтийн орчноос эхлэхийг зөвлөж байна.
Энэхүү номд ашиглагдах Jupyter Notebook файлуудыг дараах цахим сангаас авах боломжтой.

Эдгээр Notebook файлуудыг ашиглах хоёр үндсэн арга бий.
Нэгдүгээр арга: Өөрийн компьютерт татаж авч ажиллуулах.
Та Notebook файлуудыг татаж аваад өөрийн компьютер дээр ажиллуулж болно. Энэ тохиолдолд Python болон Jupyter програмуудыг суулгах шаардлагатай.

Эдгээр програмыг суулгах нь техникийн хувьд тийм ч төвөгтэй биш боловч Python хэл сурч эхлэхийг хүсэж буй хүний хувьд програм суулгах, тохируулахад ихээхэн цаг зарцуулах нь хүндрэлтэй, заримдаа бухимдал төрүүлэх үйл явц болж болзошгүй.
Хоёрдугаар арга: Google Colab орчинд ажиллуулах.
Өөр нэг боломж нь Notebook файлуудыг Google Colab орчинд ажиллуулах явдал юм.

Colab нь Jupyter Notebook-д суурилсан, веб хөтөч дээр ажилладаг хөгжүүлэлтийн орчин бөгөөд компьютер дээрээ нэмэлт програм суулгах шаардлагагүй.

Энэхүү үйлчилгээг Google компани хөгжүүлж, ажиллуулдаг бөгөөд үнэ төлбөргүй ашиглах боломжтой.

Хэрэв та програмчлалыг дөнгөж сурч эхэлж байгаа бол Google Colab орчноос эхлэхийг би онцгойлон зөвлөж байна.
