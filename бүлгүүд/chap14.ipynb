# Класс ба функц

Та одоо кодоо функцээр, өгөгдлөө суурилагдсан төрлүүдээр зохион байгуулж чадна. Дараагийн алхам бол **объект хандалтат програмчлал** юм. Энэ нь программистын тодорхойлсон төрлүүдийг ашиглан код, өгөгдлийг хамтад нь зохион байгуулдаг.

Объект хандалтат програмчлал өргөн сэдэв тул бид аажмаар урагшилна. Энэ бүлэгт Python-ы түгээмэл хэв маягийг бүрэн дагаагүй, туршлагатай программистын бичихээс өөр боловч суралцаж эхлэхэд тохиромжтой кодоос эхэлнэ. Дараагийн хоёр бүлэгт нэмэлт боломжуудыг ашиглан Python-ы хэв маягт илүү нийцсэн код бичнэ.


## Программистын тодорхойлсон төрлүүд

Бид Python-ийн олон төрлийн суулгасан төрлийг ашигласан -- одоо бид шинэ төрлийг тодорхойлох болно.
Эхний жишээ болгон бид `Time` хэмээх өдрийн цагийг илэрхийлдэг төрлийг үүсгэх болно.
Программист тодорхойлсон төрлийг **класс** гэж бас нэрлэдэг.
Классын тодорхойлолт дараах байдалтай байна.

```python
class Time:
    """Represents a time of day."""
```

Гарчиг нь шинэ класс нь `Time` гэж нэрлэгддэг болохыг харуулж байна.
Их бие нь тухайн класс юунд зориулагдсан болохыг тайлбарладаг баримт бичиг юм.
Ангиудыг тодорхойлсноор **классын объект** үүсдэг.

Классын объект нь объект үүсгэх үйлдвэртэй адил юм.
`Time` объектыг үүсгэхийн тулд та `Time`-г функц гэж нэрлэдэг.

```python
lunch = Time()
```

Үр дүн нь `__main__.Time` төрөлтэй шинэ объект бөгөөд `__main__` нь `Time` тодорхойлогдсон модулийн нэр юм.

```python
type(lunch)
```

Объектыг хэвлэх үед Python нь ямар төрөл, санах ойд хаана хадгалагдаж байгааг хэлж өгдөг (`0x` угтвар нь дараах тоо арван арван тоот байна гэсэн үг).

```python
print(lunch)
```

Шинэ объект үүсгэхийг **instantiation**гэж нэрлэх ба уг объект нь классын**жишээ** юм.

## Атрибутууд

Объект нь **атрибутууд** гэж нэрлэгддэг хувьсагчдыг агуулж болох бөгөөд "a-TRIB-ute" гэх мэт хоёр дахь үе гэхээсээ илүү "AT-trib-ute" гэх мэт эхний үеийг онцолж дууддаг.
Бид цэгэн тэмдэглэгээ ашиглан шинж чанаруудыг үүсгэж болно.

```python
lunch.hour = 11
lunch.minute = 59
lunch.second = 1
```

Энэ жишээ нь `hour`, `minute`, `second` гэсэн атрибутуудыг үүсгэдэг бөгөөд эдгээр нь `11:59:01` цагийн цаг, минут, секундийг агуулдаг бөгөөд энэ нь миний хувьд өдрийн хоолны цаг юм.

Дараах диаграмм нь эдгээр даалгаврын дараах `lunch` төлөв болон түүний шинж чанаруудыг харуулж байна.

```python
from diagram import make_frame, make_binding

d1 = dict(hour=11, minute=59, second=1)
frame = make_frame(d1, name='Time', dy=-0.3, offsetx=0.48)
binding = make_binding('lunch', frame)
```

```python
from diagram import diagram, adjust

width, height, x, y = [1.77, 1.24, 0.25, 0.86]
ax = diagram(width, height)
bbox = binding.draw(ax, x, y)
#adjust(x, y, bbox)
```

`lunch` хувьсагч нь гурван шинж чанарыг агуулсан `Time` объектыг хэлдэг. 
Шинж чанар бүр нь бүхэл тоог илэрхийлдэг.
Обьект болон түүний шинж чанаруудыг харуулсан ийм төлөвийн диаграммыг **объект диаграм** гэж нэрлэдэг.

Та цэгийн оператор ашиглан шинж чанарын утгыг уншиж болно.

```python
lunch.hour
```

Та ямар ч илэрхийллийн хэсэг болгон шинж чанарыг ашиглаж болно.

```python
total_minutes = lunch.hour * 60 + lunch.minute
total_minutes
```

Мөн та f-мөр дэх илэрхийлэлд цэгийн операторыг ашиглаж болно.

```python
f'{lunch.hour}:{lunch.minute}:{lunch.second}'
```

Гэхдээ өмнөх жишээ нь стандарт форматтай биш гэдгийг анхаарна уу.
Үүнийг засахын тулд бид `minute` болон `second` шинж чанаруудыг тэгээр хэвлэх ёстой.
Бид үүнийг буржгар хаалтанд байгаа илэрхийлэлүүдийг **формат тодорхойлогч**-аар сунгах замаар хийж болно.
Дараах жишээнд `minute` ба `second`-г шаардлагатай бол хамгийн багадаа хоёр оронтой, эхний тэгээр харуулах ёстойг формат тодорхойлогч зааж байна.

```python
f'{lunch.hour}:{lunch.minute:02d}:{lunch.second:02d}'
```

Бид энэ f-мөрийг ашиглан цаг хугацааны объектын утгыг харуулах функцийг бичих болно.
Та объектыг ердийн аргаар аргумент болгон дамжуулж болно.
Жишээлбэл, дараах функц нь `Time` объектыг аргумент болгон авдаг.

```python
def print_time(time):
    s = f'{time.hour:02d}:{time.minute:02d}:{time.second:02d}'
    print(s)
```

Бид үүнийг дуудахдаа `lunch`-г аргумент болгон дамжуулж болно.

```python
print_time(lunch)
```

## Объектыг буцаах утга болгон ашиглах

Функцүүд нь объектуудыг буцаах боломжтой. Жишээлбэл, `make_time` нь `hour`, `minute`, `second` гэсэн параметрүүдийг авч, тэдгээрийг `Time` объектод атрибут болгон хадгалж, шинэ объектыг буцаана.

```python
def make_time(hour, minute, second):
    time = Time()
    time.hour = hour
    time.minute = minute
    time.second = second
    return time
```

Параметрүүд нь шинж чанаруудтай ижил нэртэй байгаа нь гайхмаар байж болох ч энэ нь иймэрхүү функцийг бичих нийтлэг арга юм.
`Time` объектыг үүсгэхийн тулд бид `make_time`-г хэрхэн ашиглахыг эндээс үзнэ үү.`

```python
time = make_time(11, 59, 1)
print_time(time)
```

## Объектыг өөрчилж болдог

Та `9:20 PM`-аас эхэлж `92` минут буюу нэг цаг `32` минут үргэлжлэх *Монти Питон ба Ариун Граил* гэх мэт киноны үзүүлбэрт оролцох гэж байна гэж бодъё.
Кино хэдэн цагт дуусах вэ?

Эхлээд бид эхлэх цагийг илэрхийлэх `Time` объектыг үүсгэнэ.

```python
start = make_time(9, 20, 0)
print_time(start)
```

Дуусах цагийг олохын тулд бид `Time` объектын шинж чанаруудыг өөрчилж, киноны үргэлжлэх хугацааг нэмж болно.

```python
start.hour += 1
start.minute += 32
print_time(start)
```

Кино 22:52 цагт дуусна.

Энэ тооцооллыг функцэд багтааж, киноны үргэлжлэх хугацааг `hours`, `minutes`, `seconds` гэсэн гурван параметрээр авч үзэхийн тулд ерөнхийд нь оруулъя.

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds
```

Үр нөлөөг харуулсан жишээ энд байна.

```python
start = make_time(9, 20, 0)
increment_time(start, 1, 32, 0)
print_time(start)
```

Дараах стек диаграмм нь `increment_time` объектыг өөрчлөхийн өмнөхөн програмын төлөвийг харуулж байна.

```python
from diagram import Frame, Binding, Value, Stack

d1 = dict(hour=9, minute=20, second=0)
obj1 = make_frame(d1, name='Time', dy=-0.25, offsetx=0.78)

binding1 = make_binding('start', frame, draw_value=False, dx=0.7)
frame1 = Frame([binding1], name='__main__', loc='left', offsetx=-0.2)

binding2 = Binding(Value('time'), draw_value=False, dx=0.7, dy=0.35)
binding3 = make_binding('hours', 1)
binding4 = make_binding('minutes',32)
binding5 = make_binding('seconds', 0)
frame2 = Frame([binding2, binding3, binding4, binding5], name='increment_time', 
               loc='left', dy=-0.25, offsetx=0.08)

stack = Stack([frame1, frame2], dx=-0.3, dy=-0.5)
```

```python
from diagram import Bbox

width, height, x, y = [3.4, 1.89, 1.75, 1.5]
ax = diagram(width, height)
bbox1 = stack.draw(ax, x, y)
bbox2 = obj1.draw(ax, x+0.23, y)
bbox = Bbox.union([bbox1, bbox2])
# adjust(x, y, bbox)
```

Функц доторх `time` нь `start`-н өөр нэр тул `time`-г өөрчлөх үед `start` өөрчлөгдөнө.

Энэ функц ажиллаж байгаа хэдий ч үүнийг ажиллуулсны дараа бидэнд *төгсгөл* цагийг илэрхийлдэг объектыг заадаг `start` нэртэй хувьсагч үлдэх ба эхлэх цагийг илэрхийлэх объект байхгүй болсон.
`start`-г хэвээр үлдээж, эцсийн цагийг илэрхийлэх шинэ объект хийсэн нь дээр.
Бид `start`-г хуулж, хуулбарыг өөрчилснөөр үүнийг хийж чадна.

## Хуулбарлах

`copy` модуль нь `copy` нэртэй аливаа объектыг хуулбарлах боломжтой функцээр хангадаг.
Ингэж оруулж ирж болно.

```python
from copy import copy
```

Энэ нь хэрхэн ажилладагийг харахын тулд кино эхлэх цагийг илэрхийлсэн `Time` шинэ объектоос эхэлцгээе.

```python
start = make_time(9, 20, 0)
```

Мөн хуулбар хий.

```python
end = copy(start)
```

Одоо `start` болон `end` ижил өгөгдлийг агуулж байна.

```python
print_time(start)
print_time(end)
```

Гэхдээ `is` оператор нь тэдгээр нь ижил объект биш гэдгийг баталж байна.

```python
start is end
```

`==` оператор юу хийдгийг харцгаая.

```python
start == end
```

Объектууд ижил өгөгдлийг агуулж байгаа тул та `==` `True` гарна гэж найдаж болно.
Гэхдээ программистаар тодорхойлсон ангиудын хувьд `==` операторын анхдагч үйлдэл нь `is` оператортой ижил байдаг бөгөөд энэ нь тэнцэх эсэхийг биш, таних эсэхийг шалгадаг.

## Цэвэр функцууд

Бид `copy`-г ашиглан параметрүүдийг нь өөрчилдөггүй цэвэр функцүүдийг бичиж болно.
Жишээлбэл, энд `Time` объект болон үргэлжлэх хугацааг цаг, минут, секундээр авдаг функц байна.
Энэ нь эх объектын хуулбарыг хийж, хуулбарыг өөрчлөхийн тулд `increment_time` ашигладаг бөгөөд буцааж өгдөг.

```python
def add_time(time, hours, minutes, seconds):
    total = copy(time)
    increment_time(total, hours, minutes, seconds)
    return total
```

Бид үүнийг хэрхэн ашиглахыг энд харуулав.

```python
end = add_time(start, 1, 32, 0)
print_time(end)
```

Буцаах утга нь киноны дуусах цагийг илэрхийлэх шинэ объект юм.
`start` өөрчлөгдөөгүй гэдгийг бид баталж чадна.

```python
print_time(start)
```

`add_time` нь **цэвэр функц** бөгөөд энэ нь аргумент болгон өөрт нь дамжуулсан объектуудын аль нэгийг нь өөрчлөхгүй бөгөөд цорын ганц нөлөө нь утгыг буцаах явдал юм.

Өөрчлөгчөөр хийж болох бүх зүйлийг мөн цэвэр функцээр хийж болно.
Үнэн хэрэгтээ зарим програмчлалын хэлүүд зөвхөн цэвэр функцийг зөвшөөрдөг.
Цэвэр функцийг ашигладаг програмууд нь өөрчлөгч ашигладаг програмуудаас бага алдаа гаргадаг.
Гэхдээ хувиргагч нь заримдаа тохиромжтой бөгөөд илүү үр дүнтэй байдаг.

Ерөнхийдөө би танд боломжийн үед цэвэр функц бичихийг санал болгож байна, зөвхөн давуу тал байгаа тохиолдолд өөрчлөгчийг ашиглахыг зөвлөж байна.
Энэ аргыг **функциональ програмчлалын хэв маяг** гэж нэрлэж болно.

## Туршилтын хувилбар үүсгэж, засварлах

Өмнөх жишээн дээр `increment_time` болон `add_time` ажиллаж байгаа мэт боловч хэрэв бид өөр жишээ оролдвол тэдгээр нь тийм ч зөв биш болохыг харах болно.

Та театрт ирээд кино `9:20` биш `9:40`-д эхэлж байгааг мэдлээ гэж бодъё.
Бид шинэчлэгдсэн дуусах цагийг тооцоолоход юу болохыг энд харуулав.

```python
start = make_time(9, 40, 0)
end = add_time(start, 1, 32, 0)
print_time(end)
```

Үр дүн нь зөв цагийн утга биш байна.
Учир нь `increment_time` функц секунд эсвэл минутын нийлбэр `60`-аас их болох тохиолдлыг тооцоогүй байна.

Дараах сайжруулсан хувилбар эхлээд `second` нь `60`-аас хэтэрсэн эсэхийг шалгаж, хэтэрсэн бол `minute`-ийг нэмэгдүүлнэ. Дараа нь `minute` нь `60`-аас хэтэрсэн эсэхийг шалгаж, хэтэрсэн бол `hour`-ийг нэмэгдүүлнэ.

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds

    if time.second >= 60:
        time.second -= 60
        time.minute += 1

    if time.minute >= 60:
        time.minute -= 60
        time.hour += 1
```

`increment_time`-г засах нь үүнийг ашигладаг `add_time`-г мөн засдаг.
Тэгэхээр одоо өмнөх жишээ зөв ажиллаж байна.

```python
end = add_time(start, 1, 32, 0)
print_time(end)
```

Гэхдээ аргументууд нь `60`-аас том байж болзошгүй тул энэ функц зөв биш хэвээр байна.
Жишээлбэл, бидэнд ажиллах хугацааг `1` цаг, `32` минут гэхээсээ илүү `92` минут гэж өгсөн гэж бодъё.
Бид `add_time`-г ингэж дуудаж магадгүй.

```python
end = add_time(start, 0, 92, 0)
print_time(end)
```

Үр дүн нь хүчинтэй хугацаа биш байна.
Тиймээс `divmod` функцийг ашиглан өөр аргыг туршиж үзье.
Бид `start`-ийн хуулбарыг хийж, `minutes` талбарыг нэмэгдүүлэх замаар өөрчлөх болно.

```python
end = copy(start)
end.minute = start.minute + 92
end.minute
```

Одоо `minute` `132` буюу `2` цаг, `12` минут болж байна.
Бид `divmod`-г ашиглан `60`-д хувааж, бүтэн цаг болон үлдсэн минутын тоог буцаана.

```python
carry, end.minute = divmod(end.minute, 60)
carry, end.minute
```

Одоо `minute` зөв, бид цагийг `hour`-д нэмж болно.

```python
end.hour += carry
print_time(end)
```

Үр дүн нь хүчинтэй хугацаа юм.
Бид `hour` болон `second`-тай ижил зүйлийг хийж, үйл явцыг бүхэлд нь функцэд багтааж болно.

```python
def increment_time(time, hours, minutes, seconds):
    time.hour += hours
    time.minute += minutes
    time.second += seconds
    
    carry, time.second = divmod(time.second, 60)
    carry, time.minute = divmod(time.minute + carry, 60)
    carry, time.hour = divmod(time.hour + carry, 60)
```

`increment_time`-ийн энэ хувилбарын тусламжтайгаар аргументууд `60`-аас хэтэрсэн ч `add_time` зөв ажилладаг.

```python
end = add_time(start, 0, 90, 120)
print_time(end)
```

Энэ хэсэгт миний **прототип ба нөхөөс** гэж нэрлэдэг программ боловсруулах төлөвлөгөөг харуулж байна.
Бид эхний жишээн дээр зөв ажилласан энгийн загвараас эхэлсэн.
Дараа нь бид үүнийг илүү хэцүү жишээн дээр туршиж үзсэн -- алдаа олсны дараа бид дугуйг цоорсон нөхөөс хийх гэх мэт алдааг засахын тулд програмыг өөрчилсөн.

Энэ арга нь ялангуяа та асуудлын талаар гүнзгий ойлголттой болоогүй бол үр дүнтэй байж болно.
Гэхдээ нэмэлт засварууд нь олон онцгой тохиолдлуудыг авч үздэг тул шаардлагагүй төвөгтэй кодыг үүсгэж болох бөгөөд найдвартай биш, учир нь танд байгаа эсэхийг мэдэхэд хэцүү байдаг.
бүх алдааг олсон.

## Эхлээд загвар гаргаж хөгжүүлэх

Альтернатив төлөвлөгөө нь **дизайн-анхны хөгжүүлэлт** бөгөөд энэ нь загварчлалын өмнө илүү их төлөвлөлтийг хамардаг. Анхны дизайн хийх явцад заримдаа асуудлыг өндөр түвшинд ойлгох нь програмчлалыг илүү хялбар болгодог.

Энэ тохиолдолд бид `Time` объектыг 60-ын суурь дахь гурван оронтой тоо гэж төсөөлж болох бөгөөд үүнийг сексийн жижиг гэж нэрлэдэг.
`second` шинж чанар нь "нэг багана", `minute` шинж чанар нь "жаран оны багана",
болон `hour` шинж чанар нь "гучин зургаан зуун багана" юм.
`increment_time`-г бичихдээ бид 60-р суурь дээр үр дүнтэй нэмэлтийг хийж байсан тул нэг баганаас нөгөө багана руу шилжүүлэх шаардлагатай болсон.

Энэхүү ажиглалт нь асуудлыг бүхэлд нь шийдвэрлэх өөр нэг хандлагыг санал болгож байна -- бид `Time` объектуудыг бүхэл тоо руу хөрвүүлж, Python хэрхэн бүхэл тоон арифметик хийхийг мэддэг давуу талыг ашиглах боломжтой.

`Time`-аас бүхэл тоо руу хөрвүүлдэг функц энд байна.

```python
def time_to_int(time):
    minutes = time.hour * 60 + time.minute
    seconds = minutes * 60 + time.second
    return seconds
```

Үр дүн нь тухайн өдрийн эхлэлээс хойш өнгөрсөн секундын тоо юм.
Жишээлбэл, `01:01:01` гэдэг нь өдрийн эхлэлээс хойш `1` цаг, `1` минут, `1` секунд өнгөрснийг илэрхийлнэ. Энэ нь `3600` секунд, `60` секунд, `1` секундын нийлбэр юм.

```python
time = make_time(1, 1, 1)
print_time(time)
time_to_int(time)
```

Энд `divmod` функцийг ашиглан бүхэл тоог `Time` объект руу хөрвүүлэх -- өөр чиглэлд явдаг функц байна.

```python
def int_to_time(seconds):
    minute, second = divmod(seconds, 60)
    hour, minute = divmod(minute, 60)
    return make_time(hour, minute, second)
```

Өмнөх жишээг буцааж `Time` болгон хөрвүүлснээр бид үүнийг шалгаж болно.

```python
time = int_to_time(3661)
print_time(time)
```

Эдгээр функцийг ашиглан бид `add_time`-ийн илүү товч хувилбарыг бичиж болно.

```python
def add_time(time, hours, minutes, seconds):
    duration = make_time(hours, minutes, seconds)
    seconds = time_to_int(time) + time_to_int(duration)
    return int_to_time(seconds)
```

Эхний мөр нь аргументуудыг `duration` нэртэй `Time` объект болгон хувиргадаг.
Хоёр дахь мөрөнд `time` болон `duration`-г секунд болгон хувиргаж, нэмдэг.
Гурав дахь мөр нь нийлбэрийг `Time` объект болгон хувиргаж, буцаана.

Энэ нь хэрхэн ажилладаг талаар эндээс үзнэ үү.

```python
start = make_time(9, 40, 0)
end = add_time(start, 1, 32, 0)
print_time(end)
```

Зарим талаараа 60-аас суурь 10 руу хөрвүүлэх нь илүү хэцүү байдаг
зүгээр л цаг үетэйгээ харьцаж байна. Суурь хөрвүүлэлт нь илүү хийсвэр юм; бидний зөн совин
Учир нь цаг хугацааны утгуудтэй харьцах нь илүү дээр юм.

Гэхдээ бид цагийг үндсэн 60 тоо гэж үзэж, `time_to_int` болон `int_to_time` хувиргах функцүүдийг бичихэд хүчин чармайлт гаргавал бид илүү богино, унших, алдаа олж засварлахэд хялбар, найдвартай программ авах болно.

Дараа нь функцүүдийг нэмэхэд илүү хялбар байдаг. Жишээлбэл, хоёр `Time` объектыг хасаад тэдгээрийн хоорондох үргэлжлэх хугацааг олно гэж төсөөлөөд үз дээ.
Гэнэн арга бол зээл авах замаар хасах үйлдлийг хэрэгжүүлэх явдал юм.
Хөрвүүлэх функцийг ашиглах нь илүү хялбар бөгөөд зөв байх магадлалтай.

## Нэр томьёоны тайлбар

**объект хандалтат програмчлал:**
Код болон өгөгдлийг объект ашиглан зохион байгуулах програмчлалын арга.

**класс:**
Программистын тодорхойлсон төрөл. Классын тодорхойлолт нь шинэ классын объект үүсгэнэ.

**классын объект:**
Классыг төлөөлөх объект; классын тодорхойлолтын үр дүн.

**классын жишээ үүсгэх:**
Тодорхой класст харьяалагдах объект үүсгэх үйл явц.

**классын жишээ:**
Тодорхой класст харьяалагдах объект.

**атрибут:**
Объекттой холбоотой хувьсагч. Үүнийг тухайн жишээний хувьсагч гэж бас нэрлэдэг.

**объектын диаграмм:**
Объект, түүний атрибутууд болон утгуудыг харуулсан зураглал.

**форматын заагч:**
f-тэмдэгт мөр дотор утгыг ямар хэлбэрээр тэмдэгт мөр болгохыг заадаг хэсэг.

**цэвэр функц:**
Параметрүүдээ өөрчлөхгүй, утга буцаахаас өөр нөлөө үзүүлэхгүй функц.

**функциональ програмчлалын хэв маяг:**
Боломжтой бүх тохиолдолд цэвэр функц ашиглах програмчлалын арга.

**туршилтын хувилбар үүсгэж засварлах:**
Эхний бүдүүвч хувилбараас эхлээд боломжуудыг аажмаар нэмж, алдааг засах арга.

**эхлээд загвар гаргаж хөгжүүлэх:**
Туршилтын хувилбар үүсгэн засварлах аргаас илүү нарийн төлөвлөлттэйгөөр програм боловсруулах арга.
