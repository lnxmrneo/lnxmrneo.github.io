## Бүхэл хуваалт ба үлдэгдэл

```python
minutes = 105
minutes / 60
```

Гэхдээ бид ихэвчлэн аравтын бутархайгаар цаг бичдэггүй. Шалны хуваалт

```python
remainder = minutes - hours * 60
remainder
```

Эсвэл та **модуль оператор**, `%`-г ашиглаж болох бөгөөд энэ нь хоёр тоог хувааж, үлдсэнийг буцаана.

```python
remainder = minutes % 60
remainder
```

## `else` салаа

`if` заавар нь `else` заалт гэж нэрлэгддэг хоёр дахь хэсэгтэй байж болно.
Синтакс нь дараах байдлаар харагдаж байна.

```python
if x % 2 == 0:
    print('x is even')
else:
    print('x is odd')
```

Хэрэв нөхцөл үнэн бол эхний догол мөр ажиллана; эс бөгөөс хоёр дахь доголтой заавар ажиллана.

Энэ жишээнд `x` тэгш байвал `x`-г `2`-д хуваахад үлдэгдэл нь `0` байх тул нөхцөл үнэн бөгөөд програм `x is even`-г харуулна.
Хэрэв `x` сондгой бол үлдсэн нь `1` байх тул нөхцөл
Энэ нь худал бөгөөд програм `x is odd`-г харуулна.

Нөхцөл нь үнэн эсвэл худал байх ёстой тул хувилбаруудын яг аль нэг нь ажиллах болно. 
Альтернатив хувилбаруудыг **салбар** гэж нэрлэдэг.

## Олон салаат нөхцөлт заавар

Заримдаа хоёроос илүү боломж байдаг бөгөөд бидэнд хоёроос илүү салбар хэрэгтэй байдаг.

## Давхарласан нөхцөлт заавар

Нэг нөхцөлийг нөгөөд нь оруулах боломжтой.
Өмнөх хэсэгт байгаа жишээг бид дараах байдлаар бичиж болох байсан.

```python
if x == y:
    print('x and y are equal')
else:
    if x < y:
        print('x is less than y')
    else:
        print('x is greater than y')
```

Гаднах `if` заавар нь хоёр салбартай. 
Эхний салбар нь энгийн мэдэгдлийг агуулдаг. Хоёрдахь салбар нь өөр `if` мэдэгдлийг агуулдаг бөгөөд энэ нь өөрийн гэсэн хоёр салбартай.
Эдгээр хоёр салаа нь хоёулаа энгийн өгүүлбэрүүд боловч нөхцөлт мэдэгдлүүд байж болох юм.

Хэдийгээр мэдэгдлийн догол мөр нь бүтцийг тодорхой болгож байгаа ч **үүрсэн нөхцөл**-ийг уншихад хэцүү байж болно.
Боломжтой бол тэднээс зайлсхийхийг зөвлөж байна.

Логик операторууд ихэвчлэн үүрлэсэн нөхцөлт мэдэгдлүүдийг хялбарчлах арга замыг өгдөг.
Энд үүрлэсэн нөхцөл бүхий жишээг үзүүлэв.

```python
if 0 < x:
    if x < 10:
        print('x is a positive single-digit number.')
```

`print` заавар нь хоёр нөхцөлийг давсан тохиолдолд л ажиллана, тиймээс бид `and` оператортой ижил үр дүнд хүрнэ.

```python
if 0 < x and x < 10:
    print('x is a positive single-digit number.')
```

Ийм нөхцөлд Python нь илүү товч сонголтоор хангадаг:

```python
if 0 < x < 10:
    print('x is a positive single-digit number.')
```

## Рекурс

Функц өөрийгөө дуудах нь хууль ёсны юм.
Энэ нь яагаад сайн зүйл болох нь тодорхойгүй байж болох ч энэ нь програмын хийж чадах хамгийн ид шидтэй зүйлсийн нэг болж хувирдаг.
Энд нэг жишээ байна.

```python
def countdown(n):
    if n <= 0:
        print('Blastoff!')
    else:
        print(n)
        countdown(n-1)
```

Хэрэв `n` 0 эсвэл сөрөг байвал `countdown` "Blastoff!" Үгүй бол тэр
`n` гаргаад дараа нь өөрийгөө дуудаж `n-1`-г аргумент болгон дамжуулдаг.

Бид энэ функцийг `3` аргументаар дуудах үед юу болохыг энд харуулав.

```python
countdown(3)
```

`countdown`-ийн гүйцэтгэл нь `n=3`-аас эхэлдэг бөгөөд `n` илүү их байдаг.
`0`-ээс `3`-г харуулж, дараа нь өөрийгөө дууддаг\...

> `countdown`-ийн гүйцэтгэл `n=2`-аас эхэлдэг ба `n`
> `0`-ээс их, `2`-г харуулж, дараа нь өөрийгөө дууддаг\...
>
> > `countdown`-ийн гүйцэтгэл `n=1`-аас эхэлдэг ба `n`
> > `0`-ээс их, `1`-г харуулж, дараа нь өөрийгөө дууддаг\...
> >
> > > `countdown`-ийн гүйцэтгэл нь `n=0`-аас эхэлдэг ба `n`
> > > `0`-аас ихгүй, "Blastoff!" ба буцаж ирдэг.
> >
> > `n=1` авсан `countdown` буцаж ирдэг.
>
> `n=2` авсан `countdown` буцаж ирдэг.

`n=3` авсан `countdown` буцаж ирдэг.

Өөрийгөө дууддаг функц нь **рекурсив** юм.
Өөр нэг жишээ болгон бид мөрийг `n` удаа хэвлэдэг функц бичиж болно.

```python
def print_n_times(string, n):
    if n > 0:
        print(string)
        print_n_times(string, n-1)
```

Хэрэв `n` эерэг байвал `print_n_times` нь `string`-ийн утгыг харуулж, дараа нь `string` болон `n-1`-г аргумент болгон дамжуулж өөрийгөө дууддаг.

Хэрэв `n` `0` эсвэл сөрөг байвал нөхцөл худал бөгөөд `print_n_times` юу ч хийхгүй.

Энэ нь хэрхэн ажилладаг талаар эндээс үзнэ үү.

```python
print_n_times("Spam ", 4)
```

Иймэрхүү энгийн жишээнүүдийн хувьд `for` ашиглах нь илүү хялбар байх магадлалтай
давталт. Гэхдээ бид `for` ашиглан бичихэд хэцүү жишээнүүдийг дараа нь харах болно
давталт, рекурсоор бичихэд хялбар тул эрт эхлэх нь зүйтэй.

## Рекурсив функцийн стекийн диаграмм

```python
from diagram import make_frame, Stack

frames = []
for n in [3,2,1,0]:
    d = dict(n=n)
    frame = make_frame(d, name='countdown', dy=-0.3, loc='left')
    frames.append(frame)

stack = Stack(frames, dy=-0.5)
```

## Нэр томьёоны тайлбар

**рекурс:**
Одоо гүйцэтгэгдэж буй функц өөрийгөө дуудах үйл явц.

**үлдэгдлийн оператор:**
Бүхэл тоон дээр үйлчилж, нэг тоог нөгөөд хуваасны үлдэгдлийг буцаадаг `%` оператор.

**логик илэрхийлэл:**
Утга нь `True` эсвэл `False` байдаг илэрхийлэл.

**харьцуулах оператор:**
Операндуудыг харьцуулдаг `==`, `!=`, `>`, `<`, `>=`, `<=` операторуудын нэг.

**логик оператор:**
Логик илэрхийлэлд үйлчилдэг `and`, `or`, `not` зэрэг оператор.

**нөхцөлт заавар:**
Тодорхой нөхцөлөөс хамаарч гүйцэтгэлийн дарааллыг удирддаг заавар.

**нөхцөл:**
Нөхцөлт зааврын аль салаа ажиллахыг шийдэх логик илэрхийлэл.

**блок:**
Өөр зааврын хэсэг болохыг нь илтгэхээр догол мөрөөр бичсэн нэг буюу хэд хэдэн заавар.

**салаа:**
Нөхцөлт зааварт сонгон гүйцэтгэх заавруудын дарааллын нэг.

**олон салаат нөхцөлт заавар:**
Хэд хэдэн сонголттой салаанаас бүрдэх нөхцөлт заавар.

**давхарласан нөхцөлт заавар:**
Өөр нөхцөлт зааврын аль нэг салаан дотор бичсэн нөхцөлт заавар.

**рекурсив:**
Өөрийгөө дууддаг функцийг рекурсив гэнэ.

**суурь тохиолдол:**
Рекурсив функцийн дахин рекурсив дуудлага хийдэггүй салаа.

**төгсгөлгүй рекурс:**
Суурь тохиолдолгүй эсвэл түүндээ хэзээ ч хүрдэггүй рекурс. Эцэстээ ажиллах үеийн алдаа үүсгэнэ.

**шинэ мөрийн тэмдэг:**
Тэмдэгт мөрийн хоёр хэсгийн хооронд мөр шилжүүлдэг тэмдэгт.

Дараах функцийг уншаад энэ нь юу хийж байгааг олж мэдэх боломжтой.
Дараа нь ажиллуулаад зөв хийсэн эсэхээ шалгаарай.
`length`, `angle` болон `factor` утгуудыг тохируулж, үр дүнд нь ямар нөлөө үзүүлэхийг харна уу.
Хэрэв та энэ хэрхэн ажилладагийг ойлгохгүй байгаа бол виртуал туслахаас асуугаарай.

```python
from jupyturtle import forward, left, right, back

def draw(length):
    angle = 50
    factor = 0.6
    
    if length > 5:
        forward(length)
        left(angle)
        draw(factor * length)
        right(2 * angle)
        draw(factor * length)
        left(angle)
        back(length)
```

```python
from jupyturtle import make_turtle

make_turtle(delay=0)
draw(50)
```
