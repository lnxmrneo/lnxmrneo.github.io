# Програмчлал бол сэтгэн бодох арга юм 
## Арифметик операторууд 
Арифметик оператор (Arithmetic Operator) гэдэг нь арифметик тооцоолол хийх үйлдлийг илэрхийлдэг тусгай тэмдэг юм.
Жишээлбэл, нэмэх тэмдэг + нь хоёр тоог нэмэх үйлдлийг гүйцэтгэнэ.


```sql
30 + 12
```

Хасах тэмдэг - нь хоёр тооны ялгаврыг олох буюу хасах үйлдлийг гүйцэтгэнэ.

```sql
43 - 1
```

Үржих тэмдэг * нь хоёр тоог үржүүлэх үйлдлийг гүйцэтгэнэ.

```sql
6 * 7
```

Харин налуу зураас / нь хуваах үйлдлийг гүйцэтгэнэ.

```sql
84 / 2
```

Хуваах үйлдлийн үр дүн 42 биш, харин 42.0 байгааг анзаараарай.
Учир нь Python хэлэнд тоон өгөгдлийн дараах хоёр үндсэн төрөл байдаг.
* Бүхэл тоо (Integer) - бутархай хэсэггүй бүхэл тоон утгыг илэрхийлнэ.
* Хөвөгч таслалтай тоо (Floating-point number) - бутархай хэсэгтэй тоон утгыг илэрхийлэхэд ашиглагдана.

Хэрэв хоёр бүхэл тоог нэмэх, хасах эсвэл үржүүлэх үйлдэл гүйцэтгэвэл үр дүн нь бүхэл тоо байна. Харин хоёр бүхэл тоог / оператороор хуваахад үр дүн нь хөвөгч таслалтай тоо болно.


```sql
84 // 2
```

Бүхэл тоогоор хуваахыг мөн "шалны хуваалт" гэж нэрлэдэг, учир нь энэ нь үргэлж доошоо ("шал" руу) доош бүхэлчилдэг.

```sql
85 // 2
```

Эцэст нь `**` оператор экспонентацийг гүйцэтгэдэг; өөрөөр хэлбэл, энэ нь өсгөдөг
Хүчин чадлын тоо:

```sql
7 ** 2
```

Бусад зарим хэлэнд `^` гэсэн каретыг экспонентацилахад ашигладаг, харин Python дээр
Энэ нь XOR гэж нэрлэгддэг битийн оператор юм.
Хэрэв та битийн операторуудын талаар сайн мэдэхгүй бол үр дүн нь гэнэтийн байж магадгүй:

```sql
7 ^ 2
```

Би энэ номонд битийн операторуудын талаар ярихгүй, гэхдээ та энэ тухай уншиж болно
тэдгээрийг <http://wiki.python.org/moin/BitwiseOperators> дээр.

## Илэрхийлэл

Оператор болон тоонуудын цуглуулгыг **илэрхийлэл** гэж нэрлэдэг.
Илэрхийлэлд дурын тооны оператор, тоо байж болно.
Жишээлбэл, энд хоёр оператор агуулсан илэрхийлэл байна.

```sql
6 + 6 ** 2
```

Экспонентжилт нь хасахаас өмнө явагддагийг анхаарна уу.
Python нь математикийн хичээл дээр сурсан байж болох үйлдлүүдийн дарааллыг дагадаг: экспоентаци нь үржүүлэх, хуваахаас өмнө, нэмэх хасахаас өмнө болдог.

Дараах жишээнд үржүүлэх нь нэмэхээс өмнө хийгддэг.

```sql
12 + 5 * 6
```

Хэрэв та эхлээд нэмэхийг хүсвэл хаалт хэрэглэж болно.

```sql
(12 + 5) * 6
```

Илэрхийлэл бүр **утгатай** байдаг.
Жишээлбэл, `6 * 7` илэрхийлэл нь `42` утгатай байна.

## Арифметик функцууд

Python нь арифметик операторуудаас гадна тоонуудтай ажилладаг цөөн хэдэн **функцүүдийг** өгдөг.
Жишээлбэл, `round` функц нь хөвөгч таслалтай тоог авч, хамгийн ойрын бүхэл тоо хүртэл тоймлона.

```sql
round(42.4)
```

```sql
round(42.6)
```

`abs` функц нь тооны үнэмлэхүй утгыг тооцдог.
Эерэг тооны хувьд үнэмлэхүй утга нь тоо өөрөө юм.

```sql
abs(42)
```

Сөрөг тооны хувьд үнэмлэхүй утга эерэг байна.

```sql
abs(-42)
```

Бид ийм функцийг ашиглахдаа функцийг **дуудаж байна** гэж хэлдэг.
Функцийг дууддаг илэрхийлэл нь **функцийн дуудлага** юм.

Функцийг дуудах үед хаалт шаардлагатай.
Хэрэв та тэдгээрийг орхисон бол алдааны мэдэгдэл хүлээн авна.

Хоёрдахь мөр нь алдааг агуулсан код бөгөөд доор нь тэмдэглэгээ (`^`) байгаа бөгөөд алдааг хаана илрүүлснийг заана.

Сүүлийн мөрөнд энэ нь **синтаксийн алдаа** гэдгийг харуулж байгаа бөгөөд энэ нь илэрхийллийн бүтцэд ямар нэг зүйл буруу байна гэсэн үг юм.
Энэ жишээн дээрх асуудал нь функцийн дуудлагад хаалт шаардлагатай байдаг.

Хаалтанд *болон* утгыг орхивол юу болохыг харцгаая.

```sql
abs
```

Функцийн нэр нь дангаараа утгатай хууль ёсны илэрхийлэл юм.
Үүнийг харуулах үед утга нь `abs` функц болохыг илтгэх бөгөөд энэ нь миний дараа тайлбарлах зарим нэмэлт мэдээллийг агуулна.

## Тэмдэгт мөр

Python нь тооноос гадна үсгүүдийн дарааллыг төлөөлж чаддаг бөгөөд эдгээр үсгүүдийг зүүлт дээрх бөмбөлгүүдийг шиг холбосон тул **мөр** гэж нэрлэдэг.
Мөр бичихийн тулд бид шулуун хашилтын дотор үсгүүдийн дарааллыг тавьж болно.

```sql
'Hello'
```

Мөн давхар хашилт хэрэглэхийг хуульчилсан.

```sql
"world"
```

Давхар ишлэл нь таслах тэмдэг агуулсан мөр бичихэд хялбар болгодог бөгөөд энэ нь шулуун ишлэлтэй ижил тэмдэг юм.

```sql
"it's a small "
```

Мөн мөр нь хоосон зай, цэг таслал, цифр агуулсан байж болно.

```sql
'Well, '
```

`+` оператор нь мөртэй ажилладаг; энэ нь хоёр мөрийг нэг мөрөнд нэгтгэдэг бөгөөд үүнийг **холбох** гэж нэрлэдэг.

```sql
'Well, ' + "it's a small " + 'world.'
```

`*` оператор нь мөн мөртэй ажилладаг; Энэ нь мөрийн олон хуулбарыг хийж, тэдгээрийг нэгтгэдэг.

```sql
'Spam, ' * 4
```

Бусад арифметик операторууд мөртэй ажилладаггүй.

Python нь мөрийн уртыг тооцоолох `len` хэмээх функцээр хангадаг.`

```sql
len('Spam')
```

`len` нь хашилтын хоорондох үсгийг тоолох боловч ишлэлийг биш гэдгийг анхаарна уу.

Мөр үүсгэхдээ шууд хашилт ашиглахаа мартуузай.
Буцах тэмдэг гэж нэрлэгддэг арын ишлэл нь синтаксийн алдаа үүсгэдэг.

```sql
%%expect SyntaxError

`Hello`
```

Буржгар ишлэл гэж нэрлэгддэг ухаалаг ишлэлүүд нь бас хууль бус юм.

```sql
%%expect SyntaxError

‘Hello’
```

## Утга ба төрөл

Одоогоор бид гурван төрлийн утгатай танилцлаа:

* `2` нь бүхэл тоо,

* `42.0` нь хөвөгч таслалтай тоо,

* `'Hello'` нь тэмдэгт мөр.

Утгуудын ангиллыг **төрөл** гэж нэрлэдэг. Утга бүр тодорхой төрөлтэй. Өөрөөр хэлбэл, тухайн төрөлд «харьяалагдана».

Python-ы `type` функцээр дурын утгын төрлийг мэдэж болно. Бүхэл тооны төрөл нь `int` юм.

```sql
type(2)
```

Хөвөгч таслалтай төрөл нь `float` юм.

```sql
type(42.0)
```

Мөрний төрөл нь `str`.

```sql
type('Hello, World!')
```

`int`, `float`, `str` төрлүүдийг функц болгон ашиглаж болно.
Жишээлбэл, `int` нь хөвөгч таслалтай тоог авч бүхэл тоо болгон хувиргах боломжтой (үргэлж доошоо доош бүхэлчилдэг).

```sql
int(42.9)
```

Мөн `float` бүхэл тоог хөвөгч таслалтай тоот хөрвүүлэх боломжтой.

```sql
float(42)
```

Одоо энд нэг зүйлийг төөрөлдүүлж болох юм.
Хэрэв та цифрүүдийн дарааллыг хашилтанд оруулбал юу авах вэ?

```sql
'126'
```

Энэ нь тоо шиг харагддаг ч үнэндээ мөр юм.

```sql
type('126')
```

Хэрэв та үүнийг тоо шиг ашиглахыг оролдвол алдаа гарч болзошгүй.

```sql
%%expect TypeError

'126' / 3
```

Энэ жишээ нь `TypeError` үүсгэдэг бөгөөд энэ нь илэрхийлэл дэх **операнд** гэж нэрлэгддэг утгууд нь буруу төрөлтэй байна гэсэн үг юм.
Алдааны мессеж нь `/` оператор эдгээр утгын төрлийг дэмждэггүй болохыг харуулж байна, `str` болон `int`.

Хэрэв танд цифр агуулсан мөр байгаа бол та `int`-г ашиглан бүхэл тоо руу хөрвүүлж болно.

```sql
int('126') / 3
```

Хэрэв танд цифрүүд болон аравтын бутархай бүхий тэмдэгт мөр байгаа бол та үүнийг `float` ашиглан хөвөгч таслалтай тоот хувиргаж болно.

```sql
float('12.6')
```

Та том бүхэл тоо бичихдээ таслал хэрэглэхийг хүсч магадгүй юм
`1,000,000` шиг цифрүүдийн бүлгүүдийн хооронд.
Энэ нь Python-д хууль ёсны илэрхийлэл боловч үр дүн нь бүхэл тоо биш юм.

```sql
1,000,000
```

Python `1,000,000`-г бүхэл тоон таслалаар тусгаарлагдсан дараалал гэж тайлбарладаг.
Энэ төрлийн дарааллын талаар бид дараа нь илүү ихийг мэдэх болно.

Та их тоог уншихад хялбар болгохын тулд доогуур зураасыг ашиглаж болно.

```sql
1_000_000
```

## Формал ба байгалийн хэл

**Байгалийн хэл**гэдэг нь англи, испани, франц зэрэг хүмүүсийн ярьдаг хэл юм. Тэдгээрийг хүмүүс зохион бүтээгээгүй; тэд байгалийн жамаар хөгжсөн.**Формал хэл** гэдэг нь хүмүүс тодорхой хэрэглээнд зориулан бүтээсэн хэл юм. 
Жишээлбэл, математикчдын ашигладаг тэмдэглэгээ нь формал хэл бөгөөд ялангуяа тоо, тэмдэгтүүдийн хоорондын харилцааг сайн илэрхийлдэг.
Үүний нэгэн адил програмчлалын хэл нь тооцооллыг илэрхийлэхэд зориулагдсан формал хэл юм.

Хэдийгээр албан ёсны болон байгалийн хэл нь зарим онцлог шинж чанартай байдаг
нийтлэг чухал ялгаанууд байдаг:

* Хоёрдмол байдал: Байгалийн хэл нь хоёрдмол утгатай бөгөөд хүмүүс үүнийг шийддэг
    контекст шинж тэмдэг болон бусад мэдээллийг ашиглах. Формал хэл нь
    бараг эсвэл бүрэн хоёрдмол утгагүй байхаар зохион бүтээгдсэн бөгөөд энэ нь гэсэн үг юм
    Аливаа программ нь контекстээс үл хамааран яг нэг утгатай.

* Илүүдэл: Тодорхой бус байдлыг нөхөж, үл ойлголцлыг багасгахын тулд,
    байгалийн хэлэнд илүүдлийг ашигладаг. Үүний үр дүнд тэд ийм байна
    ихэвчлэн дэлгэрэнгүй. Формал хэл нь илүүц багатай, илүү товч юм.

* Шууд утга: Байгалийн хэл нь хэлц үг, зүйрлэлээр дүүрэн байдаг. Формал хэл нь яг юу хэлж байгаагаа илэрхийлдэг.

Бид бүгд байгалийн хэлээр ярьж өссөн учраас формал хэлэнд дасан зохицоход заримдаа хэцүү байдаг.
Формал хэл нь байгалийн хэлээс илүү нягт байдаг тул тэдгээрийг уншихад илүү их хугацаа шаардагддаг.
Мөн бүтэц нь чухал тул дээрээс доош, зүүнээс баруун тийш унших нь үргэлж сайн байдаггүй.
Эцэст нь нарийн ширийн зүйл чухал. Үг үсгийн жижиг алдаа болон
Та байгалийн хэлээр холдож болох цэг таслалыг хийж болно
формал хэлний томоохон ялгаа.

## Алдаа олж засварлах

Программистууд алдаа гаргадаг. Хачирхалтай шалтгааны улмаас програмчлалын алдааг **алдаа**гэж нэрлэдэг бөгөөд тэдгээрийг хянах үйл явцыг**алдаа олж засварлах** гэж нэрлэдэг.

Програмчлал, ялангуяа алдаа олж засварлах нь заримдаа хүчтэй сэтгэл хөдлөлийг бий болгодог. Хэрэв та хэцүү алдаатай тэмцэж байгаа бол уурлаж, гунигтай эсвэл ичиж магадгүй юм.

Эдгээр хариу үйлдэлд бэлтгэх нь тэдгээрийг даван туулахад тань туслах болно. Нэг арга бол компьютерийг хурд, нарийвчлал гэх мэт тодорхой давуу талтай ажилтан, мөн бусдыг өрөвдөх сэтгэлгүй, том дүр зургийг ойлгох чадваргүй гэх мэт сул тал гэж үзэх явдал юм.

Таны ажил бол сайн менежер байх явдал юм: давуу талыг ашиглах, сул талыг багасгах арга замыг хайж олох. Таны хариу үйлдэл үр дүнтэй ажиллах чадварт тань саад учруулахгүйгээр асуудлыг шийдвэрлэхийн тулд сэтгэл хөдлөлөө ашиглах арга замыг хайж ол.

Дибаг хийж сурах нь урам хугарах боловч програмчлалаас гадна олон үйл ажиллагаанд хэрэг болох утгатэй ур чадвар юм. Бүлэг бүрийн төгсгөлд алдаа олж засварлах талаар миний санал болгосон үүнтэй адил хэсэг байдаг. Тэд тусална гэж найдаж байна!

## Нэр томьёоны тайлбар

**арифметик оператор:**
Нэмэх, үржүүлэх зэрэг арифметик үйлдлийг илэрхийлдэг `+`, `*` гэх мэт тэмдэг.

**бүхэл тоо:**
Бүхэл тоонуудыг илэрхийлэх өгөгдлийн төрөл.

**хөвөгч таслалтай тоо:**
Бутархай хэсэгтэй тоонуудыг илэрхийлэх өгөгдлийн төрөл.

**бүхэл хуваалт:**
Хоёр тоог хуваагаад үр дүнг доош нь бүхэлчилдэг `//` оператор.

**илэрхийлэл:**
Хувьсагч, утга, операторын хослол.

**утга:**
Бүхэл тоо, хөвөгч таслалтай тоо, тэмдэгт мөр болон цаашид үзэх бусад төрлийн утга.

**функц:**
Тодорхой үйлдэл гүйцэтгэдэг, нэртэй заавруудын дараалал. Функц аргумент хүлээн авч эсвэл авахгүй, үр дүн буцааж эсвэл буцаахгүй байж болно.

**функцийн дуудлага:**
Функцийг ажиллуулдаг илэрхийлэл буюу илэрхийллийн хэсэг. Функцийн нэрийн араас дугуй хаалтад аргументуудын жагсаалтыг бичнэ.

**бичиглэлийн алдаа:**
Програмын бүтцийг задлан таних боломжгүй болгож, улмаар ажиллуулахад саад болдог алдаа.

**тэмдэгт мөр:**
Тэмдэгтүүдийн дарааллыг илэрхийлэх төрөл.

**залгах:**
Хоёр тэмдэгт мөрийг төгсгөлөөр нь залгах үйлдэл.

**төрөл:**
Утгуудын ангилал. Бид бүхэл тоо (`int`), хөвөгч таслалтай тоо (` float`), тэмдэгт мөр (`str`) гэсэн төрлүүдийг үзсэн.

**операнд:**
Операторын үйлдэл хийж буй утгуудын нэг.

**байгалийн хэл:**
Байгалийн жамаар хөгжсөн, хүмүүсийн ярьдаг хэл.

**формал хэл:**
Математикийн санаа, компьютерийн програмыг илэрхийлэх зэрэг тодорхой зорилгоор хүмүүсийн зохиосон хэл. Програмчлалын бүх хэл формал хэл юм.

**програмын алдаа:**
Програмд гарсан алдаа.

**алдаа олж засварлах:**
Алдааг илрүүлж, засах үйл явц.

### Виртуал туслахаас асуух

Та энэ номыг уншиж байхдаа суралцахад тань туслах виртуал туслах эсвэл чатбот ашиглах хэд хэдэн арга бий.

* Хэрэв та бүлгийн сэдвийн талаар илүү ихийг мэдэхийг хүсвэл эсвэл тодорхойгүй зүйл байвал тайлбар асууж болно.

* Хэрэв танд ямар нэгэн дасгал хийхэд хэцүү байгаа бол тусламж хүсч болно.

Бүлэг болгонд би танд виртуал туслахаар хийж болох дасгалуудыг санал болгох болно, гэхдээ би та бүхнийг бие даан туршиж үзээд өөрт тохирох зүйлийг харахыг зөвлөж байна.

* Өмнө нь би битийн операторуудыг дурдсан боловч `7 ^ 2`-ийн утга яагаад 5 байдгийг тайлбарлаагүй. "Python-д битийн операторууд юу вэ?" гэж асуугаад үзээрэй. эсвэл "`7 XOR 2` ямар утгатай вэ?"

* Би мөн үйл ажиллагааны дарааллыг дурдсан. Дэлгэрэнгүй мэдээллийг "Python дээр үйлдлийн дараалал юу вэ?"

* Бидний хөвөгч таслалтай тоот тоог хамгийн ойрын бүхэл тоо хүртэл дугуйлахад ашигладаг `round` функц нь хоёр дахь аргумент авч болно. "Дугуй функцийн аргументууд юу вэ?" гэж асуугаад үзээрэй. эсвэл "Би яаж пи-г аравтын бутархайг гурван орон болгох вэ?"

* Миний дурдаагүй өөр нэг арифметик оператор байна; "Python дахь модулийн оператор гэж юу вэ?" гэж асуугаад үзээрэй.

Ихэнх виртуал туслахууд Python-ийн талаар мэддэг тул үүнтэй төстэй асуултуудад найдвартай хариулдаг.
Гэхдээ эдгээр хэрэгслүүд алдаа гаргадаг гэдгийг санаарай.
