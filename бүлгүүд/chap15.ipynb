# Класс ба метод

Python бол **объект хандалтат хэл**. Өөрөөр хэлбэл, дараах онцлогтой объект хандалтат програмчлалыг дэмжих боломжуудтай:

- Тооцооллын ихэнхийг объект дээр хийх үйлдлээр илэрхийлдэг.

- Объект нь ихэвчлэн бодит ертөнцийн зүйлийг төлөөлдөг бөгөөд метод нь тэдгээрийн харилцан үйлчлэлийг илэрхийлдэг.

- Програм нь класс болон методын тодорхойлолтуудыг агуулдаг.

Жишээлбэл, өмнөх бүлэгт бид өдрийн цагийг илэрхийлэх `Time` класс, мөн цагтай холбоотой үйлдлүүдийг гүйцэтгэх функцүүд тодорхойлсон. Гэвч `Time` классын тодорхойлолт болон түүний дараах функцүүдийн тодорхойлолтын хооронд шууд холбоо байгаагүй.

Функцийг классын тодорхойлолт доторх **метод** болгон дахин бичсэнээр энэ холбоог илэрхий болгож болно.


## Метод тодорхойлох

Өмнөх бүлэгт бид `Time` нэртэй классыг тодорхойлж, өдрийн цагийг харуулдаг `print_time` нэртэй функцийг бичсэн.

```python
class Time:
    """Represents the time of day."""

def print_time(time):
    s = f'{time.hour:02d}:{time.minute:02d}:{time.second:02d}'
    print(s)
```

`print_time`-г арга болгохын тулд функцийг зөөхөд л хангалттай
классын тодорхойлолт доторх тодорхойлолт. Өөрчлөлтийг анзаараарай
догол.

Үүний зэрэгцээ бид параметрийн нэрийг `time`-аас `self` болгон өөрчлөх болно.
Энэ өөрчлөлтийг хийх шаардлагагүй, гэхдээ `self` гэж нэрлэх аргын эхний параметр нь ердийн зүйл юм.

```python
class Time:
    """Represents the time of day."""    

    def print_time(self):
        s = f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}'
        print(s)
```

Энэ функцийг дуудахын тулд та `Time` объектыг аргумент болгон дамжуулах ёстой.
`Time` объект хийхэд бидний ашиглах функц энд байна.

```python
def make_time(hour, minute, second):
    time = Time()
    time.hour = hour
    time.minute = minute
    time.second = second
    return time
```

Энд `Time` жишээ байна.

```python
start = make_time(9, 40, 0)
```

Одоо `print_time` руу залгах хоёр арга бий. Эхний (ба бага нийтлэг)
арга бол функцийн синтаксийг ашиглах явдал юм.

```python
Time.print_time(start)
```

Энэ хувилбарт `Time` нь классын нэр, `print_time` нь аргын нэр, `start` нь параметр болгон дамждаг.
Хоёр дахь (ба илүү хэлц үг) арга бол аргын синтаксийг ашиглах явдал юм:

```python
start.print_time()
```

Энэ хувилбарт `start` нь аргыг дуудаж байгаа объектыг **хүлээн авагч** гэж нэрлэдэг бөгөөд энэ нь аргыг дуудах нь тухайн объект руу мессеж илгээхтэй адил төстэй зүйрлэлд үндэслэнэ.

Синтаксаас үл хамааран аргын зан байдал ижил байна.
Хүлээн авагч нь эхний параметрт хуваарилагдсан тул аргын дотор `self` нь `start`-тай ижил объектыг хэлнэ.

## Өөр нэг метод

Өмнөх бүлгийн `time_to_int` функц энд байна.

```python
def time_to_int(time):
    minutes = time.hour * 60 + time.minute
    seconds = minutes * 60 + time.second
    return seconds
```

Мөн арга болгон дахин бичсэн хувилбарыг энд оруулав.


```python
%%add_method_to Time

    def time_to_int(self):
        minutes = self.hour * 60 + self.minute
        seconds = minutes * 60 + self.second
        return seconds
```

Эхний мөрөнд `add_method_to` тусгай командыг ашигладаг бөгөөд энэ нь өмнө нь тодорхойлсон ангид аргыг нэмдэг.
Энэ тушаал нь Jupyter дэвтэр дээр ажилладаг боловч Python-ийн нэг хэсэг биш тул бусад орчинд ажиллахгүй.
Ер нь классын бүх аргууд нь классын тодорхойлолт дотор байдаг тул ангитай нэгэн зэрэг тодорхойлогддог.
Гэхдээ энэ номын хувьд нэг аргыг нэг дор тодорхойлох нь тустай.

Өмнөх жишээний нэгэн адил методын тодорхойлолтыг доголдуулсан бөгөөд параметрийн нэр нь `self` байна.
Үүнээс гадна арга нь функцтэй ижил байна.
Үүнийг бид хэрхэн дуудаж байгааг энд харуулав.

```python
start.time_to_int()
```

Бид функцийг "дуудаж", аргыг "дуудаж байна" гэж хэлэх нь түгээмэл боловч тэдгээр нь ижил утгатай.

## Статик методууд

Өөр нэг жишээ болгон `int_to_time` функцийг авч үзье.
Өмнөх бүлгийн хувилбарыг энд оруулав.

```python
def int_to_time(seconds):
    minute, second = divmod(seconds, 60)
    hour, minute = divmod(minute, 60)
    return make_time(hour, minute, second)
```

Энэ функц нь `seconds`-г параметр болгон авч, шинэ `Time` объектыг буцаана.
Хэрэв бид үүнийг `Time` классын арга болгон хувиргавал `Time` объект дээр дуудах хэрэгтэй.
Гэхдээ хэрэв бид шинэ `Time` объект үүсгэх гэж байгаа бол бид үүнийг юугаар дуудах ёстой вэ?

Бид энэ тахиа-өндөгний асуудлыг **статик метод** ашиглан шийдэж болох бөгөөд энэ нь классын жишээг дуудах шаардлагагүй арга юм.
Бид энэ функцийг статик методар хэрхэн дахин бичихийг эндээс үзнэ үү.

```python
%%add_method_to Time

    def int_to_time(seconds):
        minute, second = divmod(seconds, 60)
        hour, minute = divmod(minute, 60)
        return make_time(hour, minute, second)
```

Энэ нь статик метод учраас `self`-г параметр болгон оруулаагүй болно.
Үүнийг дуудахын тулд бид классын объект болох `Time`-г ашигладаг.

```python
start = Time.int_to_time(34800)
```

Үр дүн нь `9:40`-г илэрхийлэх шинэ объект юм.

```python
start.print_time()
```

Одоо бид `Time.from_seconds`-тэй болсон тул `add_time`-г арга болгон бичихэд ашиглаж болно.
Өмнөх бүлгийн функцийг энд харуулав.

```python
def add_time(time, hours, minutes, seconds):
    duration = make_time(hours, minutes, seconds)
    seconds = time_to_int(time) + time_to_int(duration)
    return int_to_time(seconds)
```

Мөн арга болгон дахин бичсэн хувилбарыг энд оруулав.

```python
%%add_method_to Time

    def add_time(self, hours, minutes, seconds):
        duration = make_time(hours, minutes, seconds)
        seconds = time_to_int(self) + time_to_int(duration)
        return Time.int_to_time(seconds)
```

`add_time` нь статик метод биш учраас `self` параметртэй.
Энэ бол энгийн арга бөгөөд үүнийг **инстанцийн арга** гэж бас нэрлэдэг.
Үүнийг дуудахын тулд бидэнд `Time` жишээ хэрэгтэй.

```python
end = start.add_time(1, 32, 0)
print_time(end)
```

## `__str__` метод

Арга бичихдээ бараг л хүссэн нэрээ сонгож болно.
Гэсэн хэдий ч зарим нэр нь онцгой утгатай байдаг.
Жишээлбэл, хэрэв объект `__str__` нэртэй аргатай бол Python тухайн объектыг мөр болгон хөрвүүлэхэд ашигладаг.
Жишээлбэл, энд цаг хугацааны объектын `__str__` арга байна.

```python
%%add_method_to Time

    def __str__(self):
        s = f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}'
        return s
```

Энэ арга нь өмнөх бүлгээс `print_time`-тай төстэй бөгөөд зөвхөн мөрийг хэвлэхээс илүүтэйгээр буцааж өгдөг.

Та энэ аргыг ердийн аргаар дуудаж болно.

```python
end.__str__()
```

Гэхдээ Python үүнийг танд зориулж дуудаж болно.
Хэрэв та `Time` объектыг мөр болгон хөрвүүлэхдээ `str` суулгасан функцийг ашигладаг бол Python `Time` ангид `__str__` аргыг ашигладаг.

```python
str(end)
```

Хэрэв та `Time` объектыг хэвлэвэл энэ нь мөн адил болно.

```python
print(end)
```

`__str__` зэрэг аргуудыг **тусгай метод** гэж нэрлэдэг.
Тэдний нэр хоёр доогуур зураасаар эхэлж, төгсдөг тул та тэдгээрийг тодорхойлж болно.

## __init__ метод

Тусгай аргуудаас хамгийн онцгой нь `__init__` бөгөөд энэ нь шинэ объектын шинж чанаруудыг эхлүүлдэг тул гэгддэг.
`Time` классын `__init__` арга нь дараах байдалтай байж болно.

```python
%%add_method_to Time

    def __init__(self, hour=0, minute=0, second=0):
        self.hour = hour
        self.minute = minute
        self.second = second
```

Одоо бид `Time` объектыг үүсгэх үед Python `__init__`-г дуудаж, аргументуудыг дамжуулдаг.
Тиймээс бид нэгэн зэрэг объект үүсгэж, шинж чанаруудыг эхлүүлэх боломжтой.

```python
time = Time(9, 40, 0)
print(time)
```

Энэ жишээнд параметрүүд нь сонголттой тул хэрэв та `Time`-г аргументгүйгээр залгавал,
та үндсэн утгуудыг авах болно.

```python
time = Time()
print(time)
```

Хэрэв та нэг аргумент оруулбал `hour`-г хүчингүй болгоно:

```python
time = Time(9)
print(time)
```

Хэрэв та хоёр аргумент оруулбал тэдгээр нь `hour` болон `minute`-г хүчингүй болгоно.

```python
time = Time(9, 45)
print(time)
```

Хэрэв та гурван аргумент оруулбал тэдгээр нь бүх гурван үндсэн утгыг хүчингүй болгоно
утгууд.

Би шинэ класс бичихдээ бараг үргэлж `__init__` бичдэг бөгөөд энэ нь объект үүсгэхэд хялбар болгодог, `__str__` нь алдаа олж засварлахэд хэрэгтэй байдаг.

## Операторын үйлдлийг дахин тодорхойлох

Бусад тусгай аргуудыг тодорхойлсноор та зан төлөвийг тодорхойлж болно
программист тодорхойлсон төрлүүдийн операторууд. Жишээлбэл, хэрэв та a
`Time` ангид зориулсан `__add__` нэртэй аргыг ашиглавал та `+` ашиглаж болно.
Цагийн объектууд дээрх оператор.

Энд `__add__` арга байна.

```python
%%add_method_to Time

    def __add__(self, other):
        seconds = self.time_to_int() + other.time_to_int()
        return Time.int_to_time(seconds)
```

Бид үүнийг ингэж ашиглаж болно.

```python
duration = Time(1, 32)
end = start + duration
print(end)
```

Эдгээр гурван мөр кодыг ажиллуулахад маш их зүйл тохиолддог:

* Бид `Time` объектыг үүсгэх үед `__init__` аргыг дууддаг.

* Бид `+` операторыг `Time` объекттой ашиглах үед түүний `__add__` аргыг дууддаг.

* Мөн бид `Time` объектыг хэвлэх үед түүний `__str__` аргыг дууддаг.

Программист тодорхойлсон төрлүүдтэй ажиллахын тулд операторын зан төлөвийг өөрчлөхийг **операторын хэт ачаалал** гэж нэрлэдэг.
`+` гэх мэт оператор бүрийн хувьд `__add__` гэх мэт холбогдох тусгай метод байдаг.

## Алдаа олж засварлах

`Time` объект зөв байх нөхцөл нь `minute`, `second`-ийн утгууд `0`-ээс `60`-ын хооронд, өөрөөр хэлбэл `0`-ийг багтааж, `60`-ыг багтаахгүй мужид байх, мөн `hour` эерэг байх явдал юм.
Түүнчлэн `hour`, `minute` бүхэл тоо байх ёстой боловч `second` бутархай хэсэгтэй байж болно.
Ийм шаардлагуудыг үргэлж үнэн байх ёстой учраас **инвариант** гэж нэрлэдэг. Өөрөөр хэлбэл, эдгээр нөхцөл биелэхгүй бол ямар нэг зүйл буруу болсон гэсэн үг.

Инвариантуудыг шалгах код бичих нь алдааг илрүүлж, шалтгааныг нь олоход тусалдаг.
Жишээлбэл, Time объект хүлээн аваад аль нэг инвариант зөрчигдвөл `False` буцаадаг `is_valid` метод бичиж болно.

```python
%%add_method_to Time

    def is_valid(self):
        if self.hour < 0 or self.minute < 0 or self.second < 0:
            return False
        if self.minute >= 60 or self.second >= 60:
            return False
        if not isinstance(self.hour, int):
            return False
        if not isinstance(self.minute, int):
            return False
        return True
```

Дараа нь арга бүрийн эхэнд та аргументуудыг шалгаж, тэдгээр нь хүчинтэй эсэхийг шалгах боломжтой.

```python
%%add_method_to Time

    def is_after(self, other):
        assert self.is_valid(), 'self is not a valid Time'
        assert other.is_valid(), 'self is not a valid Time'
        return self.time_to_int() > other.time_to_int()
```

`assert` заавар нь дараах илэрхийллийг үнэлдэг. Хэрэв үр дүн нь `True` бол юу ч хийхгүй; хэрэв үр дүн нь `False` бол `AssertionError` үүсгэдэг.
Энд нэг жишээ байна.

```python
duration = Time(minute=132)
print(duration)
```

```python
%%expect AssertionError

start.is_after(duration)
```

`assert` мэдэгдлүүд нь хэвийн нөхцөлтэй харьцдаг кодыг алдааг шалгадаг кодоос ялгадаг тул ашигтай байдаг.

## Нэр томьёоны тайлбар

**объект хандалтат хэл:**
Объект хандалтат програмчлалыг, ялангуяа хэрэглэгчийн тодорхойлсон төрөл болон удамшлыг дэмждэг хэл.

**метод:**
Классын тодорхойлолт дотор тодорхойлж, тухайн классын жишээн дээр дууддаг функц.

**хүлээн авагч:**
Методыг нь дуудаж буй объект.

**статик метод:**
Хүлээн авагч объектгүйгээр дуудаж болох метод.

**жишээний метод:**
Хүлээн авагч объекттойгоор дуудах шаардлагатай метод.

**тусгай метод:**
Оператор болон зарим функц тухайн объекттой хэрхэн ажиллахыг өөрчилдөг метод.

**операторын үйлдлийг дахин тодорхойлох:**
Тусгай метод ашиглан оператор хэрэглэгчийн тодорхойлсон төрөлтэй хэрхэн ажиллахыг өөрчлөх үйл явц.

**инвариант:**
Програм ажиллах явцад үргэлж үнэн байх ёстой нөхцөл.


## Дасгалууд

### Виртуал туслахаас асуух

Статик аргуудын талаар нэмэлт мэдээлэл авахыг хүсвэл виртуал туслахаас асууна уу:

* "Жишээ арга ба статик аргын хооронд ямар ялгаа байдаг вэ?"

* "Яагаад статик аргуудыг статик гэж нэрлэдэг вэ?"

Хэрэв та виртуал туслахаас статик метод үүсгэхийг хүсэх юм бол үр дүн нь `@staticmethod`-аас эхлэх бөгөөд энэ нь статик метод гэдгийг илтгэх "чимэглэгч" юм.
Чимэглэлийн талаар энэ номонд оруулаагүй ч хэрэв та сонирхож байвал VA-аас нэмэлт мэдээлэл авах боломжтой.

Энэ бүлэгт бид хэд хэдэн функцийг аргууд болгон дахин бичсэн.
Виртуал туслахууд ерөнхийдөө ийм төрлийн кодын хувиргалтанд сайн байдаг.
Жишээ болгон дараах функцийг VA-д буулгаад түүнээс "Энэ функцийг `Time` классын арга болгон дахин бичнэ үү" гэж асуугаарай.
