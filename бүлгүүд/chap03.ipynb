# Функц

Өмнөх бүлэгт бид `int` болон `float` гэх мэт Python-оос олгосон хэд хэдэн функцийг ашигласан бөгөөд `math` модулиас олгосон `sqrt`, `pow` зэрэг хэд хэдэн функцийг ашигласан.
Энэ бүлэгт та өөрийн функцүүдийг хэрхэн үүсгэж, тэдгээрийг ажиллуулах талаар сурах болно.
Нэг функц нөгөөг нь хэрхэн дуудаж болохыг бид харах болно.
Жишээ болгон бид Monty Python дуунуудын үгийг харуулах болно.
Эдгээр тэнэг жишээнүүд нь чухал шинж чанарыг харуулж байна -- өөрийн функцийг бичих чадвар нь програмчлалын үндэс суурь юм.

Энэ бүлэгт мөн тооцооллыг давтан хийхэд ашигладаг `for` давталт гэсэн шинэ мэдэгдлийг танилцуулж байна.

## Шинэ функц тодорхойлох

**Функцийн тодорхойлолт** нь шинэ функцийн нэр болон функцийг дуудах үед ажиллах хэллэгүүдийн дарааллыг заана. Энд нэг жишээ байна:

````python
def print_lyrics():
    print("I'm a lumberjack, and I'm okay.")
    print("I sleep all night and I work all day.")
````

`def` нь функцийн тодорхойлолт гэдгийг илтгэх түлхүүр үг юм.
Функцийн нэр нь `print_lyrics`.
Хууль ёсны хувьсагчийн нэр болох аливаа зүйл нь мөн хуулийн функцийн нэр юм.

Нэрийн ард байгаа хоосон хаалт нь энэ функц ямар ч аргумент авдаггүйг харуулж байна.

Функцийн тодорхойлолтын эхний мөрийг **толгой**, үлдсэн хэсгийг нь **бие** гэж нэрлэдэг.
Толгой хэсэг нь хоёр цэгээр төгсөж, бие нь доголтой байх ёстой. Дүрмээр бол догол нь үргэлж дөрвөн зайтай байдаг. 
Энэ функцийн үндсэн хэсэг нь хоёр хэвлэх заавар юм; ерөнхийдөө функцийн бие нь ямар ч төрлийн хэллэг агуулж болно.

Функцийг тодорхойлсноор **функцийн объект** үүсэх ба бид үүнийг ингэж харуулах боломжтой.

````python
print_lyrics
````

Гаралт нь `print_lyrics` нь аргумент авдаггүй функц болохыг харуулж байна.
`__main__` нь `print_lyrics` агуулсан модулийн нэр юм.

Одоо бид функцийг тодорхойлсны дараа бид үүнийг суулгасан функц гэж нэрлэдэгтэй адил дуудаж болно.

````python
print_lyrics()
````

Функц ажиллаж байх үед "The Lumberjack Song"-ийн эхний хоёр мөрийг харуулсан бие дэх мэдэгдлүүдийг гүйцэтгэдэг.

## Параметрүүд

Бидний үзсэн зарим функцүүд аргумент шаарддаг; жишээ нь, та `abs` руу залгахдаа аргумент болгон дугаар дамжуулдаг.
Зарим функцүүд нэгээс олон аргумент авдаг; жишээ нь, `math.pow` нь суурь ба илтгэгч гэсэн хоёрыг авдаг.

Энд аргумент авдаг функцийн тодорхойлолт байна.

````python
def print_twice(string):
    print(string)
    print(string)
````

Хаалтанд байгаа хувьсагчийн нэр нь **параметр** байна.
Функцийг дуудах үед аргументын утгыг параметрт өгнө.
Жишээлбэл, бид `print_twice`-г ингэж дуудаж болно.

````python
print_twice('Dennis Moore, ')
````

Энэ функцийг ажиллуулах нь аргументыг параметрт оноож, дараа нь функцийн биеийг гүйцэтгэхтэй адил үр дүнтэй.

````python
string = 'Dennis Moore, '
print(string)
print(string)
````

Та мөн хувьсагчийг аргумент болгон ашиглаж болно.

````python
line = 'Dennis Moore, '
print_twice(line)
````

Энэ жишээнд `line`-ийн утгыг `string` параметрт зааж өгсөн болно.

## Функц дуудах

Функцийг тодорхойлсны дараа та үүнийг өөр функц дотор ашиглаж болно.
Үзүүлэхийн тулд бид "Спам дуу" (<https://www.songfacts.com/lyrics/monty-python/the-spam-song>) дууны үгийг хэвлэх функцүүдийг бичих болно.

> Спам, Спам, Спам, Спам,  
> Спам, Спам, Спам, Спам,  
> Спам, Спам,  
> (Хөөрхөн спам, гайхалтай спам!)  
> Спам, Спам,

Бид дараах функцээс эхлэх бөгөөд үүнд хоёр параметр орно.


````python
def repeat(word, n):
    print(word * n)
````

Бид энэ функцийг ашиглан дууны эхний мөрийг хэвлэх боломжтой.

````python
spam = 'Spam, '
repeat(spam, 4)
````

Эхний хоёр мөрийг харуулахын тулд бид `repeat` ашигладаг шинэ функцийг тодорхойлж болно.

````python
def first_two_lines():
    repeat(spam, 4)
    repeat(spam, 4)
````

Тэгээд үүнийг ингэж нэрлэ.

````python
first_two_lines()
````

Сүүлийн гурван мөрийг харуулахын тулд бид `repeat` ашигладаг өөр функцийг тодорхойлж болно.

````python
def last_three_lines():
    repeat(spam, 2)
    print('(Lovely Spam, Wonderful Spam!)')
    repeat(spam, 2)
````

````python
last_three_lines()
````

Эцэст нь бид шүлгийг бүхэлд нь хэвлэх нэг функцээр бүгдийг нь нэгтгэж чадна.

````python
def print_verse():
    first_two_lines()
    last_three_lines()
````

````python
print_verse()
````

Бид `print_verse`-г ажиллуулахад `first_two_lines` гэж нэрлэдэг бөгөөд энэ нь `repeat`, `print` гэж нэрлэдэг.
Энэ бол маш олон функц юм.

Мэдээжийн хэрэг, бид цөөн тооны функцээр ижил зүйлийг хийж болох байсан, гэхдээ энэ жишээний гол зорилго нь функцүүд хэрхэн хамт ажиллаж болохыг харуулах явдал юм.

## Давтан гүйцэтгэх

Хэрэв бид нэгээс олон шүлгийг харуулахыг хүсвэл `for` мэдэгдлийг ашиглаж болно.
Энгийн жишээ энд байна.

````python
for i in range(2):
    print(i)
````

Эхний мөр нь хоёр цэгээр төгссөн толгой юм.
Хоёр дахь мөр нь доголтой байх ёстой бие юм.

Эхний мөр нь `for` гэсэн түлхүүр үг, `i` нэртэй шинэ хувьсагч, `in` гэсэн өөр түлхүүр үгээр эхэлдэг. 
Энэ нь `range` функцийг ашиглан `0` ба `1` гэсэн хоёр утгын дарааллыг үүсгэдэг.
Python дээр бид тоолж эхлэхдээ ихэвчлэн `0`-аас эхэлдэг.

`for` мэдэгдлийг ажиллуулах үед `range`-аас `i` руу эхний утгыг оноож, дараа нь `print` функцийг биед ажиллуулдаг бөгөөд энэ нь `0`-г харуулдаг.

Биеийн төгсгөлд хүрмэгц энэ нь толгой руу буцаж эргэлддэг тул энэ мэдэгдлийг **давталт** гэж нэрлэдэг.
Хоёр дахь удаагаа давталтаар дамжуулан `range`-аас `i` руу дараагийн утгыг оноож, харуулна.
Дараа нь, энэ нь `range`-ийн сүүлчийн утга учир давталт дуусна.

Дууны хоёр бадаг хэвлэхийн тулд `for` давталтыг хэрхэн ашиглаж болохыг эндээс үзнэ үү.

````python
for i in range(2):
    print("Verse", i)
    print_verse()
    print()
````

Та функц дотор `for` давталт тавьж болно.
Жишээлбэл, `print_n_verses` нь бүхэл тоо байх ёстой `n` нэртэй параметрийг авч, өгөгдсөн тооны шүлгийг харуулна.

````python
def print_n_verses(n):
    for i in range(n):
        print_verse()
        print()
````

Энэ жишээн дээр бид давталтын үндсэн хэсэгт `i` ашигладаггүй, гэхдээ толгой хэсэгт хувьсагчийн нэр байх ёстой.

## Хувьсагч ба параметрүүд локал байдаг

Функц дотор хувьсагч үүсгэх үед энэ нь **local** бөгөөд энэ нь
Энэ нь зөвхөн функц дотор байдаг гэсэн үг юм.
Жишээлбэл, дараах функц нь хоёр аргумент авч, тэдгээрийг нэгтгэж, үр дүнг хоёр удаа хэвлэдэг.

````python
def cat_twice(part1, part2):
    cat = part1 + part2
    print_twice(cat)
````

Үүнийг ашигладаг жишээ энд байна:

````python
line1 = 'Always look on the '
line2 = 'bright side of life.'
cat_twice(line1, line2)
````

`cat_twice` ажиллаж байх үед `cat` нэртэй локал хувьсагчийг үүсгэдэг бөгөөд функц дуусах үед устгадаг.
Хэрэв бид үүнийг харуулахыг оролдвол `NameError` гарч ирнэ:

````python
print(cat)
````

Функцээс гадуур `cat` тодорхойлогдоогүй байна. 

Параметрүүд нь мөн локал шинж чанартай байдаг.
Жишээлбэл, `cat_twice`-аас гадна `part1`, `part2` гэж байдаггүй.

## Стекийн диаграммууд

Аль хувьсагчийг хаана ашиглаж болохыг хянахын тулд заримдаа **стек диаграм** зурах нь ашигтай байдаг. 
Төлөвийн диаграммуудын нэгэн адил стек диаграм нь хувьсагч бүрийн утгыг харуулдаг боловч хувьсагч бүрийн хамаарах функцийг харуулдаг.

Функц бүрийг **фрэйм**-ээр илэрхийлнэ.
Фрейм нь гадна талдаа функцийн нэр, дотор талд функцийн параметр болон локал хувьсагчтай хайрцаг юм.

Өмнөх жишээний стек диаграмыг энд харуулав.

````python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# Жишээ өгөгдөл
line1 = "Hello, "
line2 = "World!"

# Frame-ийн өгөгдөл
d1 = dict(line1=line1, line2=line2)
d2 = dict(part1=line1, part2=line2, cat=line1 + line2)
d3 = dict(s=line1 + line2)
d4 = {"?": line1 + line2}

frames = [
    ("__main__", d1),
    ("cat_twice", d2),
    ("print_twice", d3),
    ("print", d4)
]

# Диаграмм үүсгэх
fig, ax = plt.subplots(figsize=(9, 7))
ax.set_xlim(0, 10)
ax.set_ylim(0, 11)
ax.axis("off")

def draw_frame(ax, name, variables, x, y):
    row_height = 0.45
    height = len(variables) * row_height + 0.5

    # Frame-ийн хүрээ
    rect = Rectangle(
        (x, y - height),
        8, height,
        fill=False,
        edgecolor="black",
        linewidth=1.2
    )
    ax.add_patch(rect)

    # Функцийн нэр
    ax.text(
        x + 0.2, y - 0.3,
        name,
        fontsize=12,
        fontweight="bold",
        family="monospace"
    )

    # Хувьсагчид
    for i, (key, value) in enumerate(variables.items()):
        yy = y - 0.8 - i * row_height

        ax.text(
            x + 0.5, yy,
            key,
            fontsize=11,
            family="monospace"
        )

        ax.text(
            x + 2.2, yy,
            repr(value),
            fontsize=11,
            family="monospace"
        )

# Stack Diagram
y_positions = [10.5, 8.2, 5.5, 3.2]

for (name, variables), y in zip(frames, y_positions):
    draw_frame(ax, name, variables, 1, y)

plt.title("Function Stack Diagram", fontsize=14)
plt.show()
````

аль гэх мэт. Доод талаас нь уншихад `print`-г `print_twice` гэж нэрлэсэн бөгөөд үүнийг `cat_twice` гэж нэрлэдэг бөгөөд үүнийг `__main__` гэж нэрлэдэг бөгөөд энэ нь хамгийн дээд талын хүрээний тусгай нэр юм.
Аливаа функцээс гадуур хувьсагч үүсгэх үед энэ нь `__main__`-д хамаарна.

`print` хүрээн дэх асуултын тэмдэг нь бид параметрийн нэрийг мэдэхгүй байгааг харуулж байна.

## Функц яагаад хэрэгтэй вэ?

Яагаад хөтөлбөрийг хуваах нь утгатэй зүйл болох нь тодорхойгүй байж магадгүй юм
функцүүд.
Үүнд хэд хэдэн шалтгаан бий:

- Шинэ функц үүсгэх нь танд бүлгийг нэрлэх боломжийг олгоно
    мэдэгдлүүд нь таны програмыг уншиж, алдаа олж засварлахэд хялбар болгодог.

- Функцүүд нь дахин давтагдах кодыг арилгах замаар програмыг жижигрүүлдэг.
    Дараа нь өөрчлөлт хийвэл нэг газар л хийх ёстой.

- Урт программыг функцүүдэд хуваах нь эд ангиудыг алдаа олж засварлах боломжийг олгодог
    нэг нэгээр нь хийж, дараа нь тэдгээрийг бүхэлд нь угсарна.

- Сайн зохион бүтээсэн функцүүд нь олон программуудад ихэвчлэн хэрэг болдог. Нэг удаа чи
    нэгийг бичиж, дибаг хийвэл та үүнийг дахин ашиглаж болно.

## Алдаа олж засварлах

Алдаа олж засварлах нь урам хугарах боловч хэцүү, сонирхолтой, заримдаа бүр хөгжилтэй байдаг.
Мөн энэ нь таны сурч болох хамгийн чухал ур чадваруудын нэг юм.

Зарим талаар алдаа олж засварлах нь мөрдөгчтэй адил юм.
Танд сэжүүр өгөгдсөн бөгөөд та үүнд хүргэсэн үйл явдлуудыг дүгнэх хэрэгтэй
Таны харж буй үр дүн.

Алдаа олж засварлах нь туршилтын шинжлэх ухаантай адил юм.
Та юу буруу болж байгаа талаар ойлголттой болсны дараа та програмаа өөрчилж, дахин оролдоно уу.
Хэрэв таны таамаглал зөв байсан бол та өөрчлөлтийн үр дүнг урьдчилан таамаглаж, ажлын хөтөлбөрт нэг алхам ойртоно.
Хэрэв таны таамаглал буруу байсан бол та шинэ таамаг дэвшүүлэх хэрэгтэй.

Зарим хүмүүсийн хувьд програмчлал болон алдаа олж засварлах нь ижил зүйл юм; өөрөөр хэлбэл програмчлал гэдэг нь программыг хүссэн зүйлээ хийх хүртэл аажмаар алдаа олж засварлах үйл явц юм.
Гол санаа нь та ажлын программаас эхлээд жижиг өөрчлөлтүүд хийж, тэдгээрийг алдаа олж засварлах хэрэгтэй.

Хэрэв та алдаа олж засварлахэд маш их цаг зарцуулж байгаа бол энэ нь тест хийхээсээ өмнө хэт их код бичиж байгаагийн шинж юм.
Хэрэв та жижиг алхмуудыг хийвэл илүү хурдан хөдөлж чадна гэдгийг мэдэж болно.

Функц үүсгэдэг заавар.

**толгой мөр:**
Функцийн тодорхойлолтын эхний мөр.

**их бие:**
Функцийн тодорхойлолт доторх заавруудын дараалал.

**функцийн объект:**
Функцийн тодорхойлолтоор үүсэх утга. Функцийн нэр нь функцийн объектыг заадаг хувьсагч юм.

**параметр:**
Аргумент болгон дамжуулсан утгыг функц дотор заахад хэрэглэдэг нэр.

**давталт:**
Нэг эсвэл хэд хэдэн зааврыг ихэвчлэн давтан гүйцэтгэдэг заавар.

**локал хувьсагч:**
Функц дотор тодорхойлогдсон, зөвхөн тухайн функцээс хандах боломжтой хувьсагч.

**стекийн диаграмм:**
Функцийн дуудлагуудын стек, тэдгээрийн хувьсагч болон зааж буй утгуудыг харуулсан зураглал.

**фрейм:**
Функцийн дуудлагыг төлөөлөх, стекийн диаграмм дахь хайрцаг. Тухайн функцийн локал хувьсагч, параметрүүдийг агуулна.

**дуудлагын мөр:**
Онцгой нөхцөл үүсэхэд хэвлэгддэг, тэр үед гүйцэтгэгдэж байсан функцүүдийн жагсаалт.

### Дасгал

Мөр болон бүхэл тоо авч, мөрний хуулбарыг ашиглан өгөгдсөн өндөртэй пирамид зурдаг `triangle` функцийг бич. `'L'` мөрийг ашиглан 4 түвшинтэй пирамидын жишээг энд үзүүлэв.

````python
def triangle(string, height):
    """Make a triangle shape by printing a string repeatedly.
    
    string: characters to be repeated
    height: number of lines in the triangle
    """
    for i in range(height+1):
        print(string * i)
````

````python
triangle('L', 5)
````

### Дасгал

Мөр ба хоёр бүхэл тоо авч, мөрийн хуулбарыг ашиглан өгөгдсөн өндөр өргөнтэй тэгш өнцөгт зурдаг `rectangle` функц бич. `'L'` мөрийг ашиглан 4 түвшинтэй пирамидын жишээг энд үзүүлэв.

````python
def rectangle(string, width, height):
    """Make a rectangle by printing a string repeatedly.
    
    string: characters to be repeated
    width: number of repetitions on each line
    height: number of lines
    """
    for i in range(height+1):
        print(string * width)
````

````python
rectangle('H', 5, 4)
````
